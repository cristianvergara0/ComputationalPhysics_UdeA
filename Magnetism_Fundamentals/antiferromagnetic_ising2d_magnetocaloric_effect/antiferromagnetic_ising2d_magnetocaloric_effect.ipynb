## 1. Introducción

Markna et al. (2026) estudian por Monte Carlo el efecto magnetocalórico (EMC) y el comportamiento crítico de la manganita GdMn$_{1-x}$Ni$_x$O$_3$ con un modelo de Heisenberg clásico tridimensional: planos con acoplamiento ferromagnético $J_1$ entre primeros vecinos y antiferromagnético $J_2$ entre segundos vecinos, acoplados entre sí, más un término de anisotropía. Para el compuesto sin dopar reportan $T_N\approx52\,\mathrm{K}$, un máximo de la variación de entropía magnética cerca de la transición y exponentes críticos próximos a los del Ising 3D.

Este trabajo reduce ese sistema a un solo plano descrito por un modelo de Ising bidimensional con la misma competencia $J_1$–$J_2$. El objetivo es entender, dentro de este modelo y con control de los errores estadísticos y sistemáticos:

## 2. Modelo, estados fundamentales y observables

### 2.1 Hamiltoniano

Se trabaja con espines $S_i=\pm1$ en una red cuadrada de $L\times L$ sitios, $N=L^2$, con el Hamiltoniano

$$
\mathcal H=-J_1\sum_{\langle i,j\rangle}S_iS_j-J_2\sum_{\langle\langle i,j\rangle\rangle}S_iS_j-h\sum_iS_i ,
$$

donde $\langle i,j\rangle$ recorre los pares de primeros vecinos (NN) y $\langle\langle i,j\rangle\rangle$ los de segundos vecinos (NNN, las cuatro diagonales). Con este signo, $J_1=1>0$ es ferromagnético y $J_2=-0.6<0$ antiferromagnético. Se usan unidades reducidas $k_B=J_1=1$: la temperatura se mide en $J_1/k_B$, el campo en $J_1$ y la entropía en $k_B$. El cociente de frustración es $g=|J_2|/J_1=0.6$.

Con fronteras periódicas (PBC) la red se cierra sobre sí misma y todos los sitios tienen cuatro NN y cuatro NNN; con fronteras libres (FBC) no hay enlaces fuera de la red, y los sitios de borde y de esquina tienen menos vecinos.

### 2.2 Estados fundamentales

A $T=0$ y $h=0$ compiten tres órdenes simples. Con la energía por espín $e=E/N$:

$$
e_{\mathrm{FM}}=-2J_1-2J_2-h,\qquad e_{\mathrm{checkerboard}}=2J_1-2J_2,\qquad e_{\mathrm{stripe}}=2J_2 .
$$

En el orden *stripe* (franjas) los espines alternan a lo largo de una dirección y se repiten a lo largo de la otra: de los cuatro enlaces NN de cada sitio, dos son paralelos y dos antiparalelos, de modo que $J_1$ no contribuye, y los cuatro enlaces diagonales son antiparalelos. El stripe es el estado fundamental cuando $|J_2|>J_1/2$, es decir $g>1/2$, y está cuatro veces degenerado: dos orientaciones de las franjas por dos fases. El checkerboard (orden de Néel), que sería el estado fundamental con $J_1$ antiferromagnético, es aquí el de mayor energía.

Un campo uniforme no cambia la energía del stripe, cuya magnetización total es nula, pero baja la del estado saturado. Las dos energías se igualan en

$$
h_c=4|J_2|-2J_1=0.4 .
$$

La Tabla 1 comprueba las tres energías con la implementación de la red, y la Tabla 2 busca, por enumeración exacta en toros pequeños, si existe alguna fase intermedia entre el stripe y el estado saturado. No aparece ninguna: el estado de mínima energía pasa directamente de magnetización nula a saturación en $h_c$.

```python
L_ref = 8
tables_ref = lattice.build_neighbor_tables(L_ref, periodic=True)
gs_formula = lattice.ground_state_energies(J1, J2)
reference_states = {
    "ferromagnético": lattice.ferromagnetic_configuration(L_ref),
    "checkerboard": lattice.checkerboard_configuration(L_ref),
    "stripe": lattice.stripe_configuration(L_ref),
}
rows = []
for name, conf in reference_states.items():
    key = {"ferromagnético": "ferro"}.get(name, name)
    M_tot, SAB, SX, SY = kernels.order_sums(conf, L_ref)
    rows.append({
        "configuración": name,
        "E/N (red 8×8, PBC)": f"{lattice.energy_numpy(conf, J1, J2, 0.0, *tables_ref) / L_ref**2:+.3f}",
        "E/N (fórmula)": f"{gs_formula[key]:+.3f}",
        "M/N": f"{M_tot / L_ref**2:.0f}",
        "M_s": f"{abs(SAB) / (2 * L_ref**2):.2f}",
        "m_stripe": f"{np.hypot(SX, SY) / L_ref**2:.2f}",
    })
show_table(pd.DataFrame(rows), 1, "Energía por espín y parámetros de orden de las configuraciones de referencia "
           "(J1 = 1, J2 = -0.6, h = 0).", name="estados_de_referencia")
```

```python
fields_gs = (0.0, 0.2, 0.38, 0.42, 0.6, 1.0)
rows = []
for Lx, Ly in ((4, 4), (6, 3), (4, 6)):
    states = lattice.minimum_energy_states(Lx, Ly, J1, J2, fields_gs)
    row = {"toro": f"{Lx}×{Ly}"}
    for h_val in fields_gs:
        e_min, mu_min = states[h_val]
        row[f"h = {h_val:g}"] = f"{e_min:+.3f} ({mu_min:.2f})"
    rows.append(row)
show_table(pd.DataFrame(rows), 2,
           "Energía por espín del estado de mínima energía y, entre paréntesis, su magnetización por espín, "
           "por enumeración exacta de todas las configuraciones de cada toro. El cambio de M/N = 0 a 1 ocurre "
           f"entre h = 0.38 y 0.42, alrededor de h_c = {H_C:.2f}.", name="estados_minimos_toros")
```

### 2.3 Parámetros de orden y longitud de correlación

El enunciado define la **magnetización de subred** dividiendo la red en las subredes $A$ ($x+y$ par) y $B$ ($x+y$ impar),

$$
M_s=\frac{1}{2N}\left|\sum_{i\in A}S_i-\sum_{j\in B}S_j\right| ,
$$

que con esta normalización vale $1/2$ en el estado checkerboard y $0$ en el stripe (Tabla 1). $M_s$ es el parámetro de orden del antiferromagneto de Néel, pero no detecta el orden que aparece aquí. Para el stripe se usan las dos componentes

$$
m_x=\frac1N\sum_{x,y}(-1)^xS_{x,y},\qquad m_y=\frac1N\sum_{x,y}(-1)^yS_{x,y},\qquad m=\sqrt{m_x^2+m_y^2},
$$

que corresponden a los vectores de ordenamiento $\mathbf Q_x=(\pi,0)$ y $\mathbf Q_y=(0,\pi)$. A partir de cualquiera de los dos parámetros de orden se definen la susceptibilidad y el cumulante de Binder (Binder, 1981),

$$
\chi=\frac{N}{T}\left(\langle m^2\rangle-\langle m\rangle^2\right),\qquad U=1-\frac{\langle m^4\rangle}{3\langle m^2\rangle^2}.
$$

Como $m$ tiene dos componentes, $U$ tiende a $2/3$ en la fase ordenada y a $1/3$ (no a $0$, como en un parámetro de orden escalar) en la desordenada.

La **longitud de correlación de segundo momento** se obtiene del factor de estructura $S(\mathbf q)=\langle|\sum_iS_ie^{i\mathbf q\cdot\mathbf r_i}|^2\rangle/N$ en el vector de ordenamiento y en el vector de onda más cercano, $\mathbf Q+\mathbf k$ con $|\mathbf k|=2\pi/L$ (Cooper et al., 1982):

$$
\xi=\frac{1}{2\sin(\pi/L)}\sqrt{\frac{S(\mathbf Q)}{S(\mathbf Q+\mathbf k)}-1}.
$$

Se suman los dos canales $\mathbf Q_x$ y $\mathbf Q_y$ y se promedian $\mathbf k$ paralelo y perpendicular a $\mathbf Q$. El cociente $\xi/L$ es adimensional y, como $U$, se cruza para distintos tamaños en la temperatura crítica.

La energía da el **calor específico** por espín, $C=N(\langle e^2\rangle-\langle e\rangle^2)/T^2$, y su cumulante $V=1-\langle e^4\rangle/(3\langle e^2\rangle^2)$ (Challa et al., 1986), cuyo mínimo tiende a $2/3$ en una transición continua y a un valor menor en una de primer orden.

### 2.4 Termodinámica del efecto magnetocalórico

Al aplicar isotérmicamente un campo $H$, la entropía cambia en $\Delta S_M(T,H)=S(T,H)-S(T,0)$. Con la magnetización total $M=\sum_iS_i$, la relación de Maxwell $(\partial S/\partial h)_T=(\partial M/\partial T)_h$ da

$$
\Delta S_M(T,H)=\int_0^H\left(\frac{\partial M}{\partial T}\right)_{h}dh ,\qquad
\left(\frac{\partial \langle M\rangle}{\partial T}\right)_{h}=\frac{\langle ME\rangle-\langle M\rangle\langle E\rangle}{T^2},
$$

donde la segunda igualdad es exacta a campo fijo ($E$ incluye el término $-hM$) y evita derivar numéricamente en $T$. Con un solo campo y la regla del trapecio, la integral se reduce a $\Delta S_M\approx\tfrac H2(\partial M/\partial T)_H$, porque $(\partial M/\partial T)_{h=0}=0$ por simetría.

El mismo $\Delta S_M$ se obtiene integrando la entropía en temperatura a campo fijo, con $dS=d\langle E\rangle/T$:

$$
S(T,h)-S(T_{\mathrm{ref}},h)=-\int_T^{T_{\mathrm{ref}}}\frac{1}{T'}\frac{d\langle E\rangle}{dT'}\,dT',\qquad
\Delta S_M(T,H)=\Delta S_M(T_{\mathrm{ref}},H)+\big[S(T,H)-S(T_{\mathrm{ref}},H)\big]-\big[S(T,0)-S(T_{\mathrm{ref}},0)\big].
$$

Los dos caminos son equivalentes en el límite continuo, pero no con una malla finita de campos cuando el sistema atraviesa una transición inducida por el campo; la Sección 9.3 muestra por qué el segundo es el adecuado en la fase ordenada.

El proceso adiabático conserva la entropía, de modo que el **cambio de temperatura adiabático** es $\Delta T_{ad}=T'-T$ con $S(T',H)=S(T,0)$. Su versión linealizada habitual es $\Delta T_{ad}\approx-T\,\Delta S_M(T,H)/C_H(T)$, con $C_H$ la capacidad calorífica total al campo $H$ (Pecharsky et al., 2001). El **poder de enfriamiento relativo** es $\mathrm{RCP}=|\Delta S_M|_{\mathrm{pico}}\times\mathrm{FWHM}$, con FWHM el ancho a media altura del pico (Franco et al., 2018). Las magnitudes totales se reportan divididas por $N$ (por espín) cuando se comparan tamaños.

## 3. Método Monte Carlo y estimadores

### 3.1 Algoritmo de Metrópolis

Cada paso Monte Carlo (MCS) consiste en $N$ intentos de volteo en sitios elegidos al azar. Voltear $S_i$ cambia la energía en

$$
\Delta E=2S_i\Big(J_1\sum_{j\in\mathrm{NN}(i)}S_j+J_2\sum_{j\in\mathrm{NNN}(i)}S_j+h\Big),
$$

y el volteo se acepta con probabilidad $\min(1,e^{-\Delta E/T})$ (Metropolis et al., 1953; Landau y Binder, 2021). Como las sumas de vecinos solo toman valores enteros entre $-4$ y $4$, las 162 probabilidades posibles se tabulan una vez por temperatura. La energía, la magnetización y las sumas que definen $M_s$, $m_x$ y $m_y$ se actualizan con el cambio de cada volteo aceptado, sin recorrer la red. El núcleo está en `isingj1j2/kernels.py` (`metropolis_mcs`, `sample_temperature`).

**Barridos con memoria.** Cada réplica recorre una lista de temperaturas en orden y usa la configuración final de una temperatura como inicial de la siguiente; en cada temperatura se descartan $N_{\mathrm{term}}$ MCS y luego se mide una vez cada $s$ MCS durante $N_{\mathrm{muestra}}$ MCS. El protocolo por defecto enfría desde una configuración aleatoria a alta temperatura. Para comprobar que los resultados no dependen del camino, la Sección 9.7 calienta desde el estado fundamental de cada campo.

**Réplicas, semillas y paralelismo.** Cada punto de las curvas combina $R$ réplicas independientes, cada una con su propia configuración inicial y su propia semilla, derivada de una semilla base con `numpy.random.SeedSequence`. La misma semilla fija la configuración inicial y la dinámica de Metrópolis (el generador de Numba es propio de cada proceso y se siembra al empezar cada réplica), de modo que cada réplica es reproducible independientemente de en qué proceso se ejecute. Las réplicas no dependen unas de otras y se ejecutan en paralelo en procesos separados; el barrido de cada réplica, en cambio, es secuencial, porque cada temperatura parte de la configuración de la anterior.

### 3.2 Estimadores y errores

Las magnitudes de fluctuación ($C$, $\chi$, $\partial M/\partial T$) se calculan con la varianza de la serie completa de cada réplica y se promedian entre réplicas; su error es el error estándar entre réplicas. Las medias y los cocientes de momentos ($U$, $V$, $\xi/L$) se calculan con los momentos promediados sobre todas las réplicas, lo que equivale a concatenar sus series, y su error es el jackknife eliminando una réplica cada vez. Calcular las varianzas réplica a réplica evita que una réplica que no haya equilibrado (por ejemplo, atrapada con dominios de franjas de orientaciones distintas a baja temperatura) infle la varianza con la dispersión de las medias entre réplicas; a cambio, el estimador tiene un sesgo relativo de orden $2\tau_{\mathrm{int}}/N_{\mathrm{muestra}}$, menor que el 5 % con las longitudes de muestreo de este trabajo (Sección 7.4). Las magnitudes derivadas de curvas (posición y altura de picos, cruces, pendientes, exponentes, leyes de campo) llevan error bootstrap entre réplicas: en cada remuestreo se sortean con reposición las réplicas de cada tamaño o campo por separado y se recalcula todo, de modo que el error incluye la incertidumbre de ubicar esas magnitudes. Se reporta el semiancho del intervalo central del 68 % de los remuestreos. Todos los errores son estadísticos; los sistemáticos, como las correcciones al escalado, se discuten aparte.

Promediar varianzas calculadas dentro de bloques cortos de la serie, en lugar de usar la varianza de la serie completa, subestima $C$ y $\chi$ cerca de la transición; la Sección 5.3 cuantifica ese sesgo.

El tiempo de autocorrelación integrado se estima con la ventana automática de Madras y Sokal (1988):

$$
\tau_{\mathrm{int}}=\frac12+\sum_{t=1}^{W}\rho(t),\qquad W=\min\{W:\ W\geq6\,\tau_{\mathrm{int}}(W)\},
$$

con $\rho(t)$ la autocorrelación normalizada de la serie.

La celda siguiente mide la velocidad de esta máquina; cada simulación imprime su costo estimado antes de lanzarse y el tiempo real al terminar.

### 4.2 Muestreo frente a la enumeración exacta de una red $4\times4$

En un toro $4\times4$ las $2^{16}$ configuraciones pueden enumerarse, de modo que los promedios térmicos se calculan exactamente. La Tabla 4 compara los momentos muestreados con los exactos a una temperatura y un campo en que todos son no nulos. Los momentos que contienen $\mu$ se miden sobre las mismas configuraciones y están fuertemente correlacionados entre sí, de modo que sus desviaciones respecto del valor exacto no son independientes: una fluctuación de una o dos desviaciones estándar en $\langle\mu\rangle$ aparece también en $\langle\mu^2\rangle$ y en $\langle e\mu\rangle$. La misma enumeración verifica la identidad $\partial\langle M\rangle/\partial T=\mathrm{Cov}(M,E)/T^2$ contra la derivada numérica de $\langle M\rangle$ exacta.

```python
L_ex, T_ex, h_ex = 4, 1.5, 0.3
E_int_ex, M_ex, SX_ex, SY_ex = lattice.enumerate_states(L_ex, L_ex, J1, J2)
exact = lattice.exact_thermal_moments(E_int_ex, M_ex, SX_ex, SY_ex, L_ex**2, T_ex, h_ex)

validation_tasks = [make_task(L_ex, [T_ex], h=h_ex, nterm=2000, nsamp=200_000, nblocks=20,
                              seed=task_seed("validacion_exacta", r), replica=r) for r in range(32)]
validation = run_experiment("validacion_enumeracion_exacta", validation_tasks)
mom_val = estimators.replica_moments(validation.stack(range(len(validation)), "blocks"))[:, 0, :]
mean_val, sem_val = estimators.mean_sem(mom_val)

rows = []
for key, label in (("e", "⟨e⟩"), ("e2", "⟨e²⟩"), ("mu", "⟨μ⟩"), ("mu2", "⟨μ²⟩"), ("emu", "⟨eμ⟩"),
                   ("m", "⟨m⟩"), ("m2", "⟨m²⟩"), ("m4", "⟨m⁴⟩")):
    k = OBS[key]
    rows.append({"magnitud": label, "exacto": f"{exact[key]:.5f}", "Monte Carlo": fmt(mean_val[k], sem_val[k]),
                 "(MC − exacto)/σ": f"{(mean_val[k] - exact[key]) / sem_val[k]:+.1f}"})
show_table(pd.DataFrame(rows), 4, f"Momentos térmicos en un toro 4×4 (T = {T_ex}, h = {h_ex}): enumeración exacta "
           "frente a 32 réplicas de Monte Carlo (error estándar entre réplicas).", name="enumeracion_exacta")

dT_fd = 1e-4
mom_exact = lambda t: lattice.exact_thermal_moments(E_int_ex, M_ex, SX_ex, SY_ex, L_ex**2, t, h_ex)
dMdT_numeric = L_ex**2 * (mom_exact(T_ex + dT_fd)["mu"] - mom_exact(T_ex - dT_fd)["mu"]) / (2 * dT_fd)
dMdT_cov = L_ex**4 * (exact["emu"] - exact["e"] * exact["mu"]) / T_ex**2
print(f"dM/dT exacta: derivada numérica = {dMdT_numeric:.6f}, Cov(M, E)/T² = {dMdT_cov:.6f} "
      f"(diferencia relativa {abs(dMdT_cov / dMdT_numeric - 1):.1e})")
```

### 4.3 Factor de estructura frente a la transformada de Fourier

La componente de Fourier en $\mathbf Q+\mathbf k$ se calcula con una suma explícita sobre la red en cada medición. La celda la compara con la transformada rápida de Fourier de NumPy para configuraciones aleatorias y comprueba que un stripe perfecto tiene todo su peso en $\mathbf Q$.

```python
max_rel = 0.0
for L_t in (6, 8, 12, 16):
    for seed_t in range(3):
        spins_t = lattice.random_configuration(L_t, np.random.default_rng(100 * L_t + seed_t))
        cos_t, sin_t = np.cos(2 * np.pi * np.arange(L_t) / L_t), np.sin(2 * np.pi * np.arange(L_t) / L_t)
        skl, skt = kernels.stripe_fourier(spins_t, L_t, cos_t, sin_t)
        P = np.abs(np.fft.fft2(spins_t.reshape(L_t, L_t).astype(float))) ** 2 / L_t**4
        half = L_t // 2
        skl_fft = P[half - 1, 0] + P[0, half - 1]
        skt_fft = P[half, L_t - 1] + P[L_t - 1, half]
        max_rel = max(max_rel, abs(skl / skl_fft - 1), abs(skt / skt_fft - 1))
stripe_t = lattice.stripe_configuration(16)
skl_s, skt_s = kernels.stripe_fourier(stripe_t, 16, np.cos(2 * np.pi * np.arange(16) / 16), np.sin(2 * np.pi * np.arange(16) / 16))
print(f"Máxima diferencia relativa con la FFT (12 configuraciones aleatorias, L = 6-16): {max_rel:.1e}")
print(f"Stripe perfecto (L = 16): m² = 1, |ψ(Q+k)|²/N² = {skl_s:.1e} (k paralelo), {skt_s:.1e} (k perpendicular)")
```

Las tres comprobaciones se cumplen a la precisión de máquina o dentro del error estadístico: las actualizaciones incrementales son exactas, el muestreo reproduce los promedios de Boltzmann y el factor de estructura coincide con la FFT. La Sección 8 añade una validación externa: la temperatura crítica en función de $g$ frente a los valores publicados por matriz de transferencia.

## 5. Equilibración y autocorrelación

### 5.1 Termalización

Antes de promediar hay que descartar la relajación desde la configuración inicial. La Figura 1 muestra, para $L=24$ con PBC, la evolución de la energía por espín y del orden stripe desde una configuración aleatoria a cinco temperaturas: en la fase desordenada ($T=3.0$ y $1.5$), cerca de la transición ($T=0.98$) y en la fase ordenada ($T=0.6$ y $0.3$). Las curvas sólidas son promedios en ventanas de 200 MCS; la Tabla 5 compara el promedio entre los MCS 5 000 y 10 000 con el de los MCS 10 000 a 20 000, en unidades del error de este último estimado con $\tau_{\mathrm{int}}$.

```python
L_th = 24
T_therm = (3.0, 1.5, 0.98, 0.6, 0.3)
n_mcs_th = simulation.scaled(20_000)
tables_th = lattice.build_neighbor_tables(L_th, periodic=True)
traces = {}
for k, T_val in enumerate(T_therm):
    seed_th = task_seed("termalizacion", k)
    spins_th = lattice.random_configuration(L_th, np.random.default_rng(seed_th))
    kernels.seed_rng(seed_th)
    traces[T_val] = kernels.thermalization_trace(spins_th, L_th, T_val, J1, J2, 0.0, *tables_th, n_mcs_th)

def window_mean(x, w=200):
    n = len(x) // w
    return (np.arange(n) + 0.5) * w, x[: n * w].reshape(n, w).mean(axis=1)

fig, axes = plt.subplots(1, 2, figsize=(12, 4.3))
colors_T = dict(zip(T_therm, plotting.WONG))
for T_val, tr in traces.items():
    for ax, col in zip(axes, (0, 2)):
        t_w, y_w = window_mean(tr[:, col])
        ax.plot(np.arange(1, len(tr) + 1), tr[:, col], color=colors_T[T_val], alpha=0.15, lw=0.5)
        ax.plot(t_w, y_w, color=colors_T[T_val], lw=1.8, label=f"T = {T_val}")
for ax, ylabel in zip(axes, (r"$e=E/N$", r"$m$ (orden stripe)")):
    ax.set_xscale("log")
    ax.axvline(5000, color="0.4", ls=":", lw=1)
    ax.set_xlabel("MCS")
    ax.set_ylabel(ylabel)
axes[0].legend(ncol=2)
axes[0].set_title("(a) energía por espín")
axes[1].set_title("(b) parámetro de orden stripe")
fig.suptitle(f"Figura 1. Relajación desde una configuración aleatoria (L = {L_th}, PBC, h = 0)", y=1.02)
plotting.savefig(fig, "fig01_termalizacion")
plt.show()

rows = []
for T_val, tr in traces.items():
    row = {"T": T_val}
    for col, name in ((0, "e"), (2, "m")):
        x = tr[:, col]
        a, b = x[n_mcs_th // 4: n_mcs_th // 2], x[n_mcs_th // 2:]
        tau, _, _ = estimators.tau_int(b)
        err_b = b.std(ddof=1) * np.sqrt(2 * max(tau, 0.5) / len(b)) if np.isfinite(tau) else np.nan
        row[f"⟨{name}⟩ (MCS 10k–20k)"] = fmt(b.mean(), err_b)
        row[f"desvío de {name} (MCS 5k–10k), en σ"] = f"{(a.mean() - b.mean()) / err_b:+.1f}" if np.isfinite(err_b) else "n/d"
    rows.append(row)
show_table(pd.DataFrame(rows), 5, "Deriva residual después de 5 000 MCS de termalización: diferencia entre el promedio de los "
           "MCS 5 000–10 000 y el de los MCS 10 000–20 000, en unidades del error de este último.",
           name="deriva_termalizacion")
```

En la fase desordenada la relajación toma unas decenas de MCS, y después de 5 000 MCS las derivas residuales de la Tabla 5 son de unas dos desviaciones estándar o menos a todas las temperaturas. Cerca de la transición y en la fase ordenada la configuración aleatoria inicial forma dominios de franjas con orientaciones o fases distintas, que se eliminan lentamente. A la temperatura más baja pueden no eliminarse nunca: si la relajación a $T=0.3$ termina en un estado con varios dominios de franjas separados por paredes, la energía queda por encima de la del stripe ($e_{\mathrm{stripe}}=-1.2$) y el orden stripe por debajo de $1$, y la dinámica local ya no sale de ese estado en la escala de la simulación. La Tabla 5 muestra entonces una deriva nula aunque el estado no sea el de equilibrio: una deriva pequeña es una condición necesaria de la equilibración, pero no suficiente.

Por eso los barridos de producción no parten de una configuración aleatoria a baja temperatura, sino que enfrían desde la fase desordenada y cada temperatura hereda la configuración ya equilibrada de la anterior, de modo que el sistema atraviesa la transición gradualmente. Las $N_{\mathrm{term}}=4000$–$5000$ MCS por temperatura de los barridos de producción superan en más de un orden de magnitud los tiempos de autocorrelación de la Sección 5.2 para $L\leq24$, incluso en la transición, y para los tamaños mayores se escalan con $L$ (Sección 7.1).

### 5.2 Tiempo de autocorrelación

Mediciones sucesivas de una cadena de Metrópolis están correlacionadas, y el número efectivo de mediciones independientes es $n/(2\tau_{\mathrm{int}})$. Para medirlo se hace un barrido con memoria largo ($N_{\mathrm{term}}=20\,000$ y $N_{\mathrm{muestra}}=10^5$ MCS por temperatura, con una medición por MCS) para $L=16$, $24$ y $32$ y cinco temperaturas, y se guardan las series de $e$ y de $m^2$.

```python
AC_L = (16, 24, 32)
AC_T = (1.30, 1.05, 0.976, 0.90, 0.70)
ac_tasks = [make_task(L, AC_T, nterm=20_000, nsamp=100_000, stride=1, nblocks=20, keep_series=True,
                      seed=task_seed("autocorrelacion", L), replica=0) for L in AC_L]
ac_run = run_experiment("autocorrelacion", ac_tasks)

ac_series = {L: ac_run.results[ac_run.select(L=L)[0]]["series"] for L in AC_L}
rows, tau_tab = [], {}
for L in AC_L:
    for it, T_val in enumerate(AC_T):
        tau_e = estimators.tau_int(ac_series[L][it, :, 0])
        tau_m = estimators.tau_int(ac_series[L][it, :, 1])
        tau_tab[(L, T_val)] = (tau_e[0], tau_m[0])
        rows.append({"L": L, "T": T_val, "τ_int(e) [MCS]": fmt(tau_e[0], tau_e[1]),
                     "τ_int(m²) [MCS]": fmt(tau_m[0], tau_m[1])})
df_tau = pd.DataFrame(rows)
show_table(df_tau.pivot(index="T", columns="L").sort_index(ascending=False).reset_index(), 6,
           "Tiempo de autocorrelación integrado de la energía y del cuadrado del orden stripe "
           "(ventana automática de Madras y Sokal, 1988), en MCS.", name="tau_int")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4.3))
it_c = AC_T.index(0.976)
for L in AC_L:
    rho = estimators.autocorrelation(ac_series[L][it_c, :, 1], max_lag=3000)
    axes[0].plot(np.arange(len(rho)), rho, color=plotting.color_for_L(L), label=f"L = {L}")
    axes[1].plot(AC_T, [tau_tab[(L, T_val)][1] for T_val in AC_T], "o-", color=plotting.color_for_L(L), label=f"L = {L}")
axes[0].set_yscale("log")
axes[0].set_ylim(1e-2, 1.1)
axes[0].set_xlabel("t [MCS]")
axes[0].set_ylabel(r"$\rho_{m^2}(t)$")
axes[0].set_title("(a) autocorrelación en T = 0.976")
axes[1].set_yscale("log")
axes[1].set_xlabel(r"$T$ [$J_1/k_B$]")
axes[1].set_ylabel(r"$\tau_{\mathrm{int}}(m^2)$ [MCS]")
axes[1].set_title("(b) tiempo de autocorrelación integrado")
for ax in axes:
    ax.legend()
fig.suptitle("Figura 2. Autocorrelación del cuadrado del orden stripe (PBC, h = 0)", y=1.02)
plotting.savefig(fig, "fig02_autocorrelacion")
plt.show()
```

El tiempo de autocorrelación tiene un máximo pronunciado en la región crítica, donde crece con el tamaño (*critical slowing down*), y es de pocas decenas de MCS lejos de ella. La Sección 7.4 mide su crecimiento con $L$ hasta $L=64$ en la temperatura crítica.

### 5.3 Varianza de la serie completa frente a varianzas por bloques

Una forma habitual de asignar errores a $C$ y $\chi$ es dividir el muestreo en bloques, calcular la varianza dentro de cada bloque y promediar. Ese promedio no estima la varianza de la serie: cada bloque mide las fluctuaciones respecto de su propia media y pierde las que son más lentas que él, de modo que su valor esperado es, para un bloque de $b$ MCS, aproximadamente $(1-2\tau_{\mathrm{int}}/b)$ veces la varianza verdadera. La Figura 3 calcula ese estimador sobre las series de la Sección 5.2 para distintos tamaños de bloque, y la Tabla 7 da el cociente para bloques de 500 MCS, el tamaño que resulta de dividir $10^4$ MCS de muestreo en 20 bloques.

```python
def block_ratio(x, b):
    '''Cociente (promedio de las varianzas dentro de bloques de b mediciones) / (varianza de la serie completa).'''
    n = (len(x) // b) * b
    xb = x[:n].reshape(-1, b)
    return xb.var(axis=1).mean() / x[:n].var()

def m_from_m2(series_m2):
    return np.sqrt(np.maximum(series_m2.astype(float), 0.0))

n_series = ac_series[AC_L[0]].shape[1]
b_values = np.unique(np.round(np.logspace(np.log10(50), np.log10(n_series / 4), 25)).astype(int))
fig, ax = plt.subplots(figsize=(7.2, 4.6))
for L in AC_L:
    x = m_from_m2(ac_series[L][it_c, :, 1])
    ax.plot(b_values, [block_ratio(x, b) for b in b_values], "o-", ms=3.5, color=plotting.color_for_L(L), label=f"L = {L}")
ax.axvline(500, color="0.4", ls=":", lw=1)
ax.axhline(1.0, color="0.3", lw=0.8)
ax.set_xscale("log")
ax.set_xlabel("tamaño del bloque b [MCS]")
ax.set_ylabel(r"$\overline{\mathrm{Var}_b(m)}\,/\,\mathrm{Var}(m)$")
ax.set_title("Figura 3. Estimador por bloques de χ en T = 0.976 (PBC)")
ax.legend()
plotting.savefig(fig, "fig03_sesgo_bloques")
plt.show()

rows = []
for L in AC_L:
    row = {"L": L}
    for it, T_val in enumerate(AC_T):
        x_m = m_from_m2(ac_series[L][it, :, 1])
        x_e = ac_series[L][it, :, 0].astype(float)
        row[f"χ, T = {T_val}"] = f"{block_ratio(x_m, 500):.2f}"
        row[f"C, T = {T_val}"] = f"{block_ratio(x_e, 500):.2f}"
    rows.append(row)
show_table(pd.DataFrame(rows), 7, "Cociente entre el estimador por bloques de 500 MCS y la varianza de la serie completa, "
           "para χ (varianza de m) y para C (varianza de e). Un cociente 1 indica ausencia de sesgo.",
           name="sesgo_estimador_bloques")
```

Lejos de la transición el cociente es cercano a 1, pero en $T\approx T_c$ el estimador por bloques de 500 MCS pierde una fracción apreciable de la varianza, y la pérdida crece con $L$ porque crece $\tau_{\mathrm{int}}$. El sesgo afecta al valor central, no solo a la barra de error, y precisamente donde se miden las alturas de los picos de $\chi$ y $C$, de las que salen $\gamma/\nu$ y $\alpha/\nu$: con ese estimador ambos exponentes quedarían subestimados y tanto más cuanto mayores fueran los tamaños incluidos. Por eso todo el análisis usa la varianza de la serie completa, y los errores se obtienen de la dispersión entre réplicas independientes, que no depende de suponer bloques independientes.

## 6. Parte A: barrido térmico con $L=16$ y $24$

### 6.1 Simulación

Los cuatro sistemas $(L,\text{fronteras})\in\{16,24\}\times\{\mathrm{PBC},\mathrm{FBC}\}$ se enfrían desde $T=3.0$ hasta $0.1$ con paso $0.1$ y $h=0$, con $N_{\mathrm{term}}=5000$ y $N_{\mathrm{muestra}}=10\,000$ MCS por temperatura y 8 réplicas por sistema. La primera réplica de cada sistema guarda la configuración final de cada temperatura para las instantáneas de la Sección 6.4.

```python
PA_T = np.round(np.arange(3.0, 0.05, -0.1), 2)            # 3.0 -> 0.1, descendente
PA_SYSTEMS = [(L, periodic) for periodic in (True, False) for L in (16, 24)]
PA_REPLICAS = 8
pa_tasks = [make_task(L, PA_T, periodic=periodic, nterm=5000, nsamp=10_000, nblocks=20,
                      keep_snapshots=(r == 0), seed=task_seed("parte_a", L, int(periodic), r), replica=r)
            for (L, periodic) in PA_SYSTEMS for r in range(PA_REPLICAS)]
parte_a = run_experiment("parte_a", pa_tasks)


def bc_name(periodic):
    return "PBC" if periodic else "FBC"


def system_curves(result, L, periodic, **criteria):
    '''Curvas (T ascendente) de un sistema: {magnitud: (valor, error)} con jackknife entre réplicas.'''
    idx = result.select(L=L, periodic=periodic, **criteria)
    T_task = result.temperatures(idx[0])
    order = np.argsort(T_task)
    blocks = result.stack(idx, "blocks")[:, order]
    return T_task[order], estimators.curves(blocks, T_task[order], L), blocks


PA = {(L, p): system_curves(parte_a, L, p) for (L, p) in PA_SYSTEMS}
```

### 6.2 Curvas pedidas por el enunciado

La Figura 4 muestra $M_s(T)$, $\chi_s(T)$ y $U_s(T)$, calculados con la magnetización de subred del enunciado, y la Figura 5 la energía y el calor específico. Las barras de error son el jackknife entre réplicas (Sección 3.2); en la mayoría de los puntos son menores que los marcadores.

```python
def plot_grid(quantities, title, fname, logy=()):
    '''Una fila por magnitud y una columna por condición de frontera; en cada panel, L = 16 y L = 24.'''
    fig, axes = plt.subplots(len(quantities), 2, figsize=(12, 3.4 * len(quantities)), squeeze=False, sharex=True)
    for row, (key, ylabel) in enumerate(quantities):
        for col, periodic in enumerate((True, False)):
            ax = axes[row, col]
            for L in (16, 24):
                T_s, cur, _ = PA[(L, periodic)]
                y, err = cur[key]
                ax.errorbar(T_s, y, yerr=err, marker="o", ms=3.5, lw=1.2, elinewidth=0.8,
                            color=plotting.color_for_L(L), label=f"L = {L}")
            if key in logy:
                ax.set_yscale("log")
            ax.set_ylabel(ylabel)
            ax.set_title(f"{ylabel}, {bc_name(periodic)}")
            if row == len(quantities) - 1:
                ax.set_xlabel(r"$T$ [$J_1/k_B$]")
            ax.legend()
    fig.suptitle(title, y=1.0)
    fig.tight_layout()
    plotting.savefig(fig, fname)
    plt.show()


plot_grid([("ms", r"$M_s$"), ("chi_s", r"$\chi_s$"), ("U_s", r"$U_s$")],
          "Figura 4. Magnetización de subred, su susceptibilidad y su cumulante de Binder (definición del enunciado)",
          "fig04_parte_a_subred")
```

```python
plot_grid([("e", r"$E/N$"), ("Cv", r"$C$")],
          "Figura 5. Energía y calor específico por espín", "fig05_parte_a_energia")
```

$M_s$ permanece en el nivel de ruido, de orden $N^{-1/2}$, en todo el rango de temperaturas, y $\chi_s$ no tiene ningún pico en la región de la transición: la magnetización de subred no la detecta, porque el estado fundamental es el stripe y no el checkerboard (Sección 2.2). El cumulante $U_s$ toma valores muy negativos a baja temperatura, lo que no indica una transición sino que $M_s$ es casi siempre nula y solo ocasionalmente toma valores no nulos, de modo que $\langle M_s^4\rangle\gg3\langle M_s^2\rangle^2$. La energía, en cambio, sí muestra la transición: cae con fuerza cerca de $T\approx1$, donde el calor específico tiene un máximo. A la temperatura más baja coincide con la del stripe perfecto, $e_{\mathrm{stripe}}=2J_2=-1.2$, con PBC. Con FBC la referencia es $2J_2(L-1)^2/L^2$, porque faltan los enlaces diagonales del borde, y la energía queda ligeramente por encima de ella (Tabla 8). La razón está en los dos bordes donde terminan las franjas: un espín de esos bordes tiene un vecino paralelo y dos antiparalelos entre los primeros vecinos y solo dos segundos vecinos, de modo que voltearlo cuesta $\Delta E=2(2|J_2|-J_1)=0.4$, frente a $4.8$ en el interior. A $T=0.1$ una fracción $e^{-4}\approx2\,\%$ de esos $\approx2L$ espines está volteada, lo que eleva la energía en $0.4\cdot2L\,e^{-4}/N$, alrededor de una milésima por espín para $L=16$ y algo menos para $L=24$, que es la diferencia que se observa. Las mismas excitaciones de borde producen la pequeña estructura de $\chi_s$ con FBC cerca de $T=0.2$ y el aporte del calor específico a baja temperatura. Una réplica que el enfriamiento dejara con dominios de franjas daría una diferencia mucho mayor. La Figura 6 repite las curvas con el parámetro de orden stripe.

### 6.3 Estimaciones de la Parte A

La Tabla 8 resume, para cada sistema, la posición y la altura del máximo de $\chi$ y de $C$ (vértice de una parábola por los tres nodos más altos), el cruce de $U$ entre $L=16$ y $L=24$ y la energía a $T=0.1$. Los errores son bootstrap entre réplicas; la malla de paso $0.1$ limita además la resolución de las posiciones, de modo que estas cifras sirven para ubicar la transición, que la Sección 7 determina con precisión. En el caso del cruce de $U$ la limitación es más fuerte: $U$ pasa de casi $2/3$ a casi $1/3$ entre dos nodos consecutivos, y una interpolación lineal entre ellos no tiene sentido. Por eso se da el intervalo de la malla en que $U(24)-U(16)$ cambia de signo, junto con la fracción de remuestreos bootstrap que lo sitúan en el mismo intervalo.

```python
PEAK_WINDOW = (0.5, 2.0)     # los máximos se buscan lejos de los extremos de la malla


def peak_stats(groups, T_s, L):
    obs_p = fss.pooled_observables(T_s, groups["mom"], L)
    sel = (T_s >= PEAK_WINDOW[0]) & (T_s <= PEAK_WINDOW[1])
    out = {}
    out["Tchi"], out["chimax"], _ = fss.parabolic_peak(T_s[sel], obs_p["chi"][sel])
    out["TCv"], out["Cvmax"], _ = fss.parabolic_peak(T_s[sel], obs_p["Cv"][sel])
    return out

rows, pa_peaks = [], {}
for (L, periodic), (T_s, cur, blocks) in PA.items():
    mom_R = estimators.replica_moments(blocks)
    boot = estimators.bootstrap({"mom": mom_R}, lambda g: peak_stats(g, T_s, L))
    pa_peaks[(L, periodic)] = boot
    e_low, e_low_err = cur["e"][0][0], cur["e"][1][0]
    e_stripe = 2 * J2 * (1.0 if periodic else ((L - 1) / L) ** 2)     # con FBC faltan los enlaces diagonales del borde
    rows.append({"sistema": f"L = {L}, {bc_name(periodic)}", "T del máximo de χ": fmt_pair(boot["Tchi"]),
                 "χ máx.": fmt_pair(boot["chimax"]), "T del máximo de C": fmt_pair(boot["TCv"]),
                 "C máx.": fmt_pair(boot["Cvmax"]), "E/N en T = 0.1": fmt(e_low, e_low_err),
                 "E/N del stripe perfecto": f"{e_stripe:.4f}"})
def u_crossing_node(groups, T_s):
    '''Índice del nodo inferior del intervalo de malla en que U(24) - U(16) cambia de signo (-1 si no hay cruce).'''
    o16 = fss.pooled_observables(T_s, groups[16], 16)
    o24 = fss.pooled_observables(T_s, groups[24], 24)
    # U puede valer casi 2/3 en el nodo ordenado más cercano y bajar transitoriamente de 1/3 en el desordenado,
    # de ahí el rango admitido; fuera de la ventana las curvas coinciden dentro del ruido.
    T_x = fss.find_crossing(T_s, o16["U"], o24["U"], y_range=(0.2, 0.6666), t_window=(0.6, 1.4))
    return int(np.searchsorted(T_s, T_x, side="right") - 1) if np.isfinite(T_x) else -1


rng_pa = np.random.default_rng(2026)
for periodic in (True, False):
    T_s = PA[(16, periodic)][0]
    groups_U = {L: estimators.replica_moments(PA[(L, periodic)][2]) for L in (16, 24)}
    i_x = u_crossing_node(groups_U, T_s)
    same = np.mean([u_crossing_node({L: g[rng_pa.integers(0, len(g), len(g))] for L, g in groups_U.items()}, T_s) == i_x
                    for _ in range(400)])
    text_x = (f"entre {T_s[i_x]:.1f} y {T_s[i_x + 1]:.1f} ({100 * same:.0f} % de los remuestreos)" if i_x >= 0
              else "sin cruce en la malla")
    rows.append({"sistema": f"cruce de U (16, 24), {bc_name(periodic)}", "T del máximo de χ": text_x,
                 "χ máx.": "", "T del máximo de C": "", "C máx.": "", "E/N en T = 0.1": "", "E/N del stripe perfecto": ""})
show_table(pd.DataFrame(rows), 8, "Parte A: máximos de χ y C, intervalo de la malla en que se cruzan los cumulantes de Binder "
           "stripe de L = 16 y 24, y energía a la temperatura más baja frente a la del stripe perfecto (2 J2 con PBC; "
           "2 J2 (L-1)²/L² con FBC, porque faltan los enlaces del borde). Error bootstrap entre réplicas.", name="parte_a_resumen")
```

**Pregunta 1: efecto de las condiciones de frontera.** Con PBC, las curvas de $L=16$ y $L=24$ son muy parecidas fuera de la región crítica y los máximos de $\chi$ y de $C$ quedan cerca de $T\approx1$ para ambos tamaños. Con FBC el orden stripe cae a temperaturas menores, los máximos están claramente por debajo de los de PBC y se corren hacia arriba al aumentar $L$, el máximo de $C$ es mucho más bajo y el de $\chi$ crece más despacio con el tamaño (Sección 10). Los espines de borde tienen menos vecinos, así que el orden es más débil cerca de la superficie, y la fracción de espines de borde, $\approx4/L$, decrece con el tamaño: el corrimiento de la temperatura pseudocrítica con FBC decae como $1/L$, mientras que con PBC decae como $L^{-1/\nu}$, más rápido (Sección 10). PBC elimina la superficie y es la elección adecuada para estimar el límite termodinámico; FBC describe una muestra finita con bordes.

**Pregunta 2: temperatura de transición.** Con la malla de paso $0.1$, el máximo de $\chi$ con PBC queda en $T\approx1.0$ para los dos tamaños y el cruce de $U$ entre $L=16$ y $24$ cae entre los nodos $0.9$ y $1.0$; con FBC el máximo de $\chi$ queda bastante más abajo para estos tamaños. Estas cifras dependen de la resolución de la malla y del tamaño: el cruce entre dos tamaños tiene la resolución de un intervalo de malla, y la posición del máximo de $\chi$ de un sistema finito no es la temperatura crítica. La Sección 7 determina la transición con una malla fina, siete tamaños y los cruces de $\xi/L$ y de $U$: $T_c\approx0.97$ (Tabla 11).

## 7. Transición de fase: escalado de tamaño finito

### 7.1 Contexto y diseño de las simulaciones

El modelo de este trabajo es el modelo de Ising $J_1$–$J_2$ de la red cuadrada con $J_1$ ferromagnético y $J_2$ antiferromagnético. A campo nulo es equivalente al caso con ambos acoplamientos antiferromagnéticos, porque invertir los espines de una subred del tablero de ajedrez cambia el signo de $J_1$ y deja $J_2$ intacto. Para $g>1/2$ el estado fundamental es el stripe, que rompe una simetría $\mathbb Z_4$, y la naturaleza de la transición térmica es un problema discutido desde hace décadas. Kalz et al. (2008, 2011) encontraron histogramas de energía con dos picos y propusieron transiciones de primer orden débil para $1/2<g\lesssim1$. Jin et al. (2012, 2013), con matriz de transferencia y Monte Carlo, y Kalz y Honecker (2012) concluyeron que la transición es de primer orden muy débil solo para $1/2<g<g^*$, con $g^*=0.67\pm0.01$, y continua para $g>g^*$, sobre la línea crítica de Ashkin–Teller que une el punto de Potts de 4 estados ($g=g^*$) con el límite de dos redes de Ising desacopladas ($g\to\infty$). Li y Jiang (2024) concluyen que para $g\geq0.58$ la transición es probablemente continua. El valor $g=0.6$ de este trabajo cae en la región en disputa.

A lo largo de la línea de Ashkin–Teller el exponente $\nu$ varía con $g$, de $2/3$ (Potts de 4 estados) a $1$ (Ising), mientras que $\gamma/\nu=7/4$ y $\beta/\nu=1/8$ se mantienen (universalidad débil; Suzuki, 1974). Por eso la comparación con las clases de universalidad se hace con los cocientes $1/\nu$, $\gamma/\nu$, $\alpha/\nu$ y $\beta/\nu$, que da directamente el escalado de tamaño finito, y no con $\beta$ o $\gamma$ por separado, que solo coinciden con los del Ising 2D ($\nu=1$, $\beta=1/8$, $\gamma=7/4$; Onsager, 1944; Yang, 1952) si $\nu=1$. En el punto de Potts de 4 estados $\nu=2/3$, $\alpha=2/3$, $\beta=1/12$ y $\gamma=7/6$ (Wu, 1982). Una transición de primer orden en $d=2$ tendría, en cambio, $1/\nu=\alpha/\nu=\gamma/\nu=2$ y $\beta/\nu=0$ (Fisher y Berker, 1982).

Se usan dos conjuntos de simulaciones con PBC y $h=0$, ambos con 8 réplicas por tamaño:

- **Ventana ancha**: $L=12,16,20,24,32$ entre $T=0.80$ y $1.10$, con paso $0.005$ entre $0.945$ y $1.035$ y $0.02$ fuera. Cubre las dos fases y permite ver la forma completa de las curvas.
- **Ventana estrecha**: $L=24,32,48,64$ entre $T=0.964$ y $0.994$ con paso $0.003$. La región crítica se estrecha como $L^{-1/\nu}$ y para $L=64$ mide unas milésimas en $T$.

En ambos casos se mide una vez cada 2 MCS ($L\leq32$) o cada 5 MCS ($L=48,64$). Como el tiempo de autocorrelación en la transición crece como una potencia de $L$ con exponente dinámico $z\approx2$ (Sección 7.4), los MCS por temperatura se escalan como $(L/24)^{2.3}$ para $L>24$, a partir de $N_{\mathrm{term}}=4000$ y $N_{\mathrm{muestra}}=12\,000$, de modo que el número de mediciones independientes por temperatura sea comparable en todos los tamaños o mayor en los grandes.

```python
def fss_counts(L, stride, nterm_ref=4000, nsamp_ref=12_000, z_scale=2.3, nblocks=20):
    '''MCS de termalización y de muestreo de cada tamaño: crecen como (L/24)^z para L > 24.'''
    factor = max(1.0, (L / 24.0) ** z_scale)
    nsamp = int(round(nsamp_ref * factor / (stride * nblocks))) * stride * nblocks
    return int(round(nterm_ref * factor)), nsamp


FSS_T = np.round(np.concatenate([np.arange(1.10, 1.039, -0.02), np.arange(1.035, 0.9449, -0.005),
                                 np.arange(0.94, 0.799, -0.02)]), 4)
FSS_L = (12, 16, 20, 24, 32)
FSS_T_NARROW = np.round(np.arange(0.994, 0.9629, -0.003), 4)
FSS_L_NARROW = (24, 32, 48, 64)
FSS_STRIDE = {12: 2, 16: 2, 20: 2, 24: 2, 32: 2, 48: 5, 64: 5}
FSS_REPLICAS = 8

fss_tasks, fss_tasks_narrow = [], []
for L in FSS_L:
    nterm, nsamp = fss_counts(L, FSS_STRIDE[L])
    fss_tasks += [make_task(L, FSS_T, nterm=nterm, nsamp=nsamp, stride=FSS_STRIDE[L], measure_sq=True,
                            keep_series=(r < 2), seed=task_seed("fss_pbc", L, r), replica=r)
                  for r in range(FSS_REPLICAS)]
for L in FSS_L_NARROW:
    nterm, nsamp = fss_counts(L, FSS_STRIDE[L])
    fss_tasks_narrow += [make_task(L, FSS_T_NARROW, nterm=nterm, nsamp=nsamp, stride=FSS_STRIDE[L], measure_sq=True,
                                   keep_series=(r < 4), seed=task_seed("fss_pbc_estrecha", L, r), replica=r)
                         for r in range(FSS_REPLICAS)]
for L in sorted(set(FSS_L) | set(FSS_L_NARROW)):
    nterm, nsamp = fss_counts(L, FSS_STRIDE[L])
    print(f"L = {L:3d}: N_term = {nterm:6d}, N_muestra = {nsamp:6d} MCS, una medición cada {FSS_STRIDE[L]} MCS")

fss_run = run_experiment("fss_pbc", fss_tasks)
fss_run_narrow = run_experiment("fss_pbc_estrecha", fss_tasks_narrow)
```

```python
def fss_dataset(result, L_values, **criteria):
    '''{L: (T ascendente, momentos por réplica (R, nT, NOBS))} para el análisis de tamaño finito.'''
    data = {}
    for L in L_values:
        idx = result.select(L=L, **criteria)
        T_task = result.temperatures(idx[0])
        order = np.argsort(T_task)
        data[L] = (T_task[order], estimators.replica_moments(result.stack(idx, "blocks"))[:, order])
    return data


FSS_DATA = fss_dataset(fss_run, FSS_L)
FSS_DATA_NARROW = fss_dataset(fss_run_narrow, FSS_L_NARROW)
T_fss = FSS_DATA[FSS_L[0]][0]
fss_curves = {L: estimators.curves_from_moments(FSS_DATA[L][1], T_fss, L) for L in FSS_L}
```

### 7.3 Temperatura crítica y exponentes

Para cada conjunto se buscan los cruces de $\xi/L$ y de $U$ entre todos los pares de tamaños con $L_2/L_1\geq1.3$ (pares de tamaños casi iguales dan curvas casi superpuestas y cruces muy ruidosos), solo donde ambas curvas están fuera de la zona saturada de la fase ordenada ($U<0.655$, $\xi/L<1.8$). $T^*$ es la mediana de los cruces de $\xi/L$ de los pares con $L_1\geq16$ (ventana ancha) o $L_1\geq24$ (ventana estrecha). En $T^*$ se evalúan las pendientes $d(\xi/L)/dT$ y $dU/dT$ y el parámetro de orden, y en los máximos las alturas de $\chi$ y de $C$; sus ajustes log–log frente a $L$ dan $1/\nu$, $\gamma/\nu$, $\alpha/\nu$ y $\beta/\nu$ (Sección 7.1). Se ajusta además un colapso de $\xi/L$ frente a $(T-T_c)L^{1/\nu}$ con $T_c$ y $1/\nu$ libres. Todos los errores son bootstrap entre réplicas (400 remuestreos), con todo el procedimiento repetido en cada remuestreo.

```python
def fss_boot(data, **kwargs):
    groups = {L: data[L][1] for L in data}
    T_map = {L: data[L][0] for L in data}
    return estimators.bootstrap(groups, lambda g: fss.fss_statistics({L: (T_map[L], g[L]) for L in g}, **kwargs))


FSS_WINDOWS = {
    "12–32 (ventana ancha)": (FSS_DATA, dict(L1_min=16, collapse_halfwidth=0.035, U_halfwindow=0.03)),
    "24–64 (ventana estrecha)": (FSS_DATA_NARROW, dict(L1_min=24, collapse_halfwidth=0.015)),
}
fss_results = {name: fss_boot(data, **kw) for name, (data, kw) in FSS_WINDOWS.items()}
# Deriva de los exponentes con la ventana de tamaños (ventana estrecha)
for sub in ((24, 32, 48), (32, 48, 64)):
    fss_results[f"{sub[0]}–{sub[-1]} (ventana estrecha)"] = fss_boot(
        {L: FSS_DATA_NARROW[L] for L in sub}, L1_min=24, collapse_halfwidth=0.015, do_collapse=False)
FSS_MAIN = fss_results["24–64 (ventana estrecha)"]
T_STAR = FSS_MAIN["T_star"][0]
T_STAR_ERR = FSS_MAIN["T_star"][1]

rows = []
for name, (data, _) in FSS_WINDOWS.items():
    res = fss_results[name]
    for L in data:
        rows.append({"conjunto": name, "L": L, "T del máximo de χ": fmt_pair(res[f"Tchi_{L}"]),
                     "χ máx.": fmt_pair(res[f"chimax_{L}"]), "T del máximo de C": fmt_pair(res[f"TCv_{L}"]),
                     "C máx.": fmt_pair(res[f"Cvmax_{L}"]), "⟨m⟩(T*)": fmt_pair(res[f"m_{L}"]),
                     "ξ/L(T*)": fmt_pair(res[f"xiL_{L}"]), "|d(ξ/L)/dT|(T*)": fmt(abs(res[f"dxi_{L}"][0]), res[f"dxi_{L}"][1])})
show_table(pd.DataFrame(rows), 9, "Magnitudes por tamaño (PBC, h = 0): máximos de χ y C (posición por parábola), "
           "y parámetro de orden, ξ/L y pendiente de ξ/L en la T* de cada conjunto.", name="fss_por_tamano")

rows = []
for name, (data, _) in FSS_WINDOWS.items():
    res = fss_results[name]
    for a, b in fss.size_pairs(list(data)):
        rows.append({"conjunto": name, "par (L1, L2)": f"({a}, {b})", "cruce de ξ/L": fmt_pair(res[f"Txi_{a}_{b}"]),
                     "cruce de U": fmt_pair(res[f"TU_{a}_{b}"])})
show_table(pd.DataFrame(rows), 10, "Temperaturas de cruce de ξ/L y de U por pares de tamaños.", name="fss_cruces")
```

### 7.4 Orden de la transición

Una transición de primer orden se reconoce, con tamaños suficientemente grandes, por tres señales: el histograma de la energía en la transición desarrolla dos picos cuya separación no se anula al crecer $L$; el mínimo del cumulante de la energía, $V_{\min}$, tiende a un valor menor que $2/3$, de modo que $2/3-V_{\min}$ satura en lugar de anularse (Challa et al., 1986); y los exponentes efectivos derivan hacia $1/\nu=\alpha/\nu=\gamma/\nu=d$. En una transición continua, en cambio, $2/3-V_{\min}\propto L^{\alpha/\nu-d}$. Si la transición es de primer orden muy débil, estas señales solo aparecen cuando $L$ supera la longitud de correlación en la transición, que puede ser mucho mayor que los tamaños simulables (Jin et al., 2013; Li y Jiang, 2024).

```python
def vmin_stats(data):
    out = {}
    for L in data:
        obs_v = fss.pooled_observables(data[L][0], data[L][1], L)
        T_v, neg_v, _ = fss.parabolic_peak(data[L][0], -obs_v["V"])
        out[f"Vgap_{L}"] = 2.0 / 3.0 + neg_v
        out[f"TV_{L}"] = T_v
    return out

vmin_broad = estimators.bootstrap({L: FSS_DATA[L][1] for L in FSS_L}, lambda g: vmin_stats({L: (T_fss, g[L]) for L in g}))
vmin_narrow = estimators.bootstrap({L: FSS_DATA_NARROW[L][1] for L in FSS_L_NARROW},
                                   lambda g: vmin_stats({L: (FSS_DATA_NARROW[L][0], g[L]) for L in g}))
VGAP = {L: vmin_broad[f"Vgap_{L}"] for L in FSS_L if L < 24}
VGAP.update({L: vmin_narrow[f"Vgap_{L}"] for L in FSS_L_NARROW})

# Tiempos de autocorrelación en el nodo más cercano a T* (series de las réplicas de la ventana estrecha)
tau_rows = {}
for L in FSS_L_NARROW:
    idx = [i for i in fss_run_narrow.select(L=L) if "series" in fss_run_narrow.results[i]]
    T_task = fss_run_narrow.temperatures(idx[0])
    it = int(np.argmin(np.abs(T_task - T_STAR)))
    taus_e = [estimators.tau_int(fss_run_narrow.results[i]["series"][it, :, 0])[0] for i in idx]
    taus_m = [estimators.tau_int(fss_run_narrow.results[i]["series"][it, :, 1])[0] for i in idx]
    n_meas = fss_run_narrow.results[idx[0]]["series"].shape[1]
    tau_rows[L] = (T_task[it], FSS_STRIDE[L] * np.array(taus_e), FSS_STRIDE[L] * np.array(taus_m), n_meas * FSS_STRIDE[L])

rows = []
L_sorted = sorted(VGAP)
for k, L in enumerate(L_sorted):
    row = {"L": L, "2/3 − V_min": fmt_pair(VGAP[L], 5),
           "pendiente local de ln(2/3 − V_min)": "n/d" if k == 0 else
           f"{np.log(VGAP[L][0] / VGAP[L_sorted[k - 1]][0]) / np.log(L / L_sorted[k - 1]):+.2f}"}
    if L in tau_rows:
        T_node, te, tm, nsamp_L = tau_rows[L]
        row["τ_int(e) en T* [MCS]"] = fmt(*estimators.mean_sem(te))
        row["τ_int(m²) en T* [MCS]"] = fmt(*estimators.mean_sem(tm))
        row["N_muestra / τ_int(m²)"] = f"{nsamp_L / tm.mean():.0f}"
    rows.append(row)
show_table(pd.DataFrame(rows), 14, "Cumulante de la energía y tiempos de autocorrelación en la transición. Para L < 24 se usa la "
           "ventana ancha y para L ≥ 24 la estrecha; τ_int se mide en el nodo más cercano a T* sobre las réplicas con series.",
           name="orden_transicion")

L_tau = np.array(sorted(tau_rows))
tau_m_stats = np.array([estimators.mean_sem(tau_rows[L][2]) for L in L_tau])
z_fit = fss.weighted_loglog_fit(L_tau, tau_m_stats[:, 0], tau_m_stats[:, 1])     # (z, error, ordenada, chi²/gdl)
gap_fit = fss.loglog_fit(L_sorted, [VGAP[L][0] for L in L_sorted])
print(f"τ_int(m²) en T* ~ L^z con z = {fmt(z_fit[0], z_fit[1])} (L = {L_tau.min()}–{L_tau.max()}, ajuste ponderado; "
      f"χ²/gdl = {z_fit[3]:.1f}, error escalado por su raíz si es mayor que 1)")
print(f"2/3 − V_min ~ L^a con a = {gap_fit[0]:.2f} (L = {L_sorted[0]}–{L_sorted[-1]}); una transición continua con "
      f"α/ν = {FSS_MAIN['alpha_nu'][0]:.2f} predice a = α/ν − 2 = {FSS_MAIN['alpha_nu'][0] - 2:.2f}; "
      "una de primer orden, a → 0")
```

### 7.5 Relaciones de escala

Si los cocientes de la Tabla 11 describen una sola transición continua, deben cumplir las relaciones de Fisher, $\gamma/\nu=2-\eta$; de Rushbrooke dividida por $\nu$, $\alpha/\nu+2\beta/\nu+\gamma/\nu=2/\nu$; y de hiperescalamiento, $2/\nu-\alpha/\nu=d=2$. Como $\chi$, $C$, $\xi$ y $\langle m\rangle$ son mediciones distintas, que las relaciones se cumplan es una prueba de consistencia interna. La Tabla 15 las evalúa con los exponentes de la ventana estrecha, con $1/\nu$ de la pendiente de $\xi/L$, y propaga los errores en cuadratura suponiéndolos independientes.

```python
def relation_rows(res):
    inv_nu, gnu, anu, bnu = res["inv_nu_xi"], res["gamma_nu"], res["alpha_nu"], res["beta_nu"]
    eta = (2 - gnu[0], gnu[1])
    lhs_r = (anu[0] + 2 * bnu[0] + gnu[0], np.sqrt(anu[1] ** 2 + 4 * bnu[1] ** 2 + gnu[1] ** 2))
    rhs_r = (2 * inv_nu[0], 2 * inv_nu[1])
    hyper = (2 * inv_nu[0] - anu[0], np.hypot(2 * inv_nu[1], anu[1]))
    return [
        {"relación": "Fisher: η = 2 − γ/ν (η = 1/4 en la línea de Ashkin–Teller)", "lado izquierdo": fmt_pair(eta),
         "lado derecho": "0.25", "diferencia (σ)": f"{(eta[0] - 0.25) / eta[1]:+.1f}"},
        {"relación": "Rushbrooke: α/ν + 2β/ν + γ/ν = 2/ν", "lado izquierdo": fmt_pair(lhs_r),
         "lado derecho": fmt_pair(rhs_r), "diferencia (σ)": f"{report.deviation_sigma(lhs_r, rhs_r):+.1f}"},
        {"relación": "hiperescalamiento: 2/ν − α/ν = d = 2", "lado izquierdo": fmt_pair(hyper),
         "lado derecho": "2", "diferencia (σ)": f"{(hyper[0] - 2) / hyper[1]:+.1f}"},
    ]

show_table(pd.DataFrame(relation_rows(FSS_MAIN)), 15, "Relaciones de escala con los exponentes de la ventana estrecha "
           "(L = 24–64).", name="relaciones_escala")
```

### 7.7 Lectura de los resultados

**Temperatura crítica.** Los cruces de $\xi/L$ y de $U$ de todos los pares de tamaños se agrupan alrededor de un mismo valor, y las dos ventanas dan temperaturas compatibles, con la de tamaños grandes más precisa (Tablas 10 y 11): $T_c\approx0.97$ para $g=0.6$. Los máximos de $\chi$ y de $C$ quedan por encima y se acercan a $T_c$ al crecer $L$, como corresponde a temperaturas pseudocríticas de un sistema finito con PBC.

**Exponentes.** $\gamma/\nu$ y $\beta/\nu$ son compatibles con $7/4$ y $1/8$, los valores comunes al Ising 2D y a toda la línea de Ashkin–Teller, de modo que no discriminan entre ellas. El exponente que discrimina es $\alpha/\nu$: su valor cercano a $1$ descarta al Ising 2D, en el que el calor específico solo diverge logarítmicamente ($\alpha/\nu=0$), y es compatible con el del Potts de 4 estados, extremo de la línea de Ashkin–Teller en $g^*\approx0.67$. Las tres estimaciones de $1/\nu$ caen entre $1.3$ y $1.75$ y son compatibles con $3/2$, pero con errores de $0.17$ a $0.25$ ninguna discrimina por sí sola: según el estimador, también admiten $1$ o $2$ a dos desviaciones estándar (Tablas 11 y 12). El colapso de $\xi/L$ con $L=12$–$32$ tampoco discrimina: su calidad es prácticamente la misma con $1/\nu=1$ y con $3/2$, y el mejor colapso está entre ambos valores (Tabla 13). Las relaciones de Fisher, Rushbrooke e hiperescalamiento se cumplen dentro de los errores (Tabla 15). Los errores son solo estadísticos; cerca del punto de Potts las correcciones al escalado son logarítmicas y pueden ser comparables a ellos, de modo que los exponentes deben leerse como efectivos de la ventana de tamaños usada.

**Orden de la transición.** Hasta $L=64$ el histograma de la energía en el máximo de $C$ es ancho y aplanado para los tamaños intermedios, pero no desarrolla dos picos que se separen al crecer $L$: el de $L=64$ es unimodal (Figura 11a). El cumulante de Binder sí muestra un rasgo que suele asociarse a una transición de primer orden: justo por encima de $T_c$ baja de su valor en la fase desordenada, $1/3$, y el mínimo es más profundo cuanto mayor es $L$ (Figura 8). Jin et al. (2012) observan ese mismo mínimo creciente en este modelo a ambos lados de $g^*$, y también en el Potts de 4 estados y en el modelo de Ashkin–Teller, cuyas transiciones son continuas, y lo interpretan como un comportamiento pseudo de primer orden propio de la vecindad del punto de Potts; por sí solo no decide el orden de la transición. $2/3-V_{\min}$ decrece con la pendiente que corresponde a una transición continua con el $\alpha/\nu$ medido, y los exponentes efectivos no derivan hacia los valores de primer orden al pasar de la ventana $24$–$48$ a la $32$–$64$: $1/\nu$, $\gamma/\nu$ y $\alpha/\nu$ no se acercan a $2$, y los cambios de los cuatro cocientes están dentro de uno o dos errores (Tablas 11 y 14, Figura 11). Esto es consistente con la conclusión de Li y Jiang (2024) para $g\geq0.58$ y con un comportamiento efectivo muy cercano al del Potts de 4 estados. No descarta una transición de primer orden extremadamente débil como la que proponen Jin et al. (2013) para $g<g^*$, cuya longitud de correlación en la transición sería mayor que $L=64$; decidirlo exigiría tamaños mucho mayores y algoritmos que no sufran el *critical slowing down* del Metrópolis local, cuyo tiempo de autocorrelación en la transición crece como $L^{z}$ con $z\approx2$ (Tabla 14). El valor de $z$ tiene una incertidumbre grande, porque $\tau_{\mathrm{int}}$ se estima en un solo nodo de temperatura y con pocas réplicas por tamaño, y los puntos se apartan de la recta más de lo que indican sus errores.

**Pregunta 5: escalado de tamaño finito.** Con PBC la posición de los máximos de $\chi$ y de $C$ se corre hacia $T_c$ al crecer $L$ y sus alturas crecen como potencias de $L$, con $\gamma/\nu$ cercano a $7/4$ y $\alpha/\nu$ cercano a $1$ (Figura 9). En el límite termodinámico $\chi$ y $C$ divergen en $T_c$; para $L$ finito quedan redondeadas en una ventana de anchura $\propto L^{-1/\nu}$. Los cruces de $\xi/L$ y de $U$ son la estimación más estable de $T_c$, porque son magnitudes adimensionales cuyas correcciones de tamaño finito son pequeñas.

**Pregunta 6: clases de universalidad.** La clase de universalidad depende de la dimensión, de la simetría del parámetro de orden y del alcance de las interacciones, no de los detalles del Hamiltoniano (Stanley, 1999). Este modelo es bidimensional y su parámetro de orden rompe una simetría $\mathbb Z_4$; sus exponentes efectivos son compatibles con los del Potts de 4 estados, incompatibles con el Ising 2D por el valor de $\alpha/\nu$ y lejanos de las clases tridimensionales. El sistema de Markna et al. (2026) es tridimensional y sus exponentes, $\beta\approx0.33$ y $\gamma\approx1.24$–$1.31$, están cerca de los del Ising 3D; los propios autores lo atribuyen a la anisotropía de tipo Ising de la manganita, que rompe la simetría continua del modelo de Heisenberg a gran escala. No hay razón para que los exponentes de los dos modelos coincidan, y no coinciden (Sección 11).

## 8. Diagrama de fase $T_c(g)$ y comparación con la literatura

Variar $g=|J_2|/J_1$ es, dentro de este modelo, el análogo más cercano a cambiar la composición del material, y permite una comprobación externa del Hamiltoniano, del signo de $J_2$ y del procedimiento de cruces frente a valores publicados. Jin et al. (2013) calcularon $T_c(g)$ por matriz de transferencia para $g\geq0.67$ (su Tabla I da $|J_1|/T_c$). Además hay dos límites exactos: $T_c\to0$ cuando $g\to1/2$, donde el estado fundamental se vuelve altamente degenerado, y $T_c\to2g/\ln(1+\sqrt2)\approx2.269\,g$ cuando $g\to\infty$, porque la red se separa en dos redes cuadradas independientes con acoplamiento $|J_2|$ (Onsager, 1944).

El cálculo tiene dos etapas para no presuponer ningún valor de la literatura: un barrido grueso con $L=16$ entre $1.2g$ y $3.0g$ ubica el máximo de $\chi$ de cada $g$, y un barrido fino entre $0.80$ y $1.08$ veces esa posición, con $L=16$, $24$ y $32$, 6 réplicas y $N_{\mathrm{muestra}}=16\,000$ MCS por temperatura, da los cruces de $U$ y de $\xi/L$ para los pares $(16,24)$ y $(24,32)$. La misma función se usa en la Sección 10 con fronteras libres.

```python
PD_G = (0.60, 0.67, 0.70, 0.75, 1.00, 2.00, 5.00)
PD_L = (16, 24, 32)
PD_INV_TC_JIN = {0.67: 0.8335, 0.70: 0.7758, 0.75: 0.69866, 1.00: 0.48029, 2.00: 0.22468, 5.00: 0.088406}  # Jin et al. (2013), Tabla I
PD_TC_JIN = {g: 1.0 / v for g, v in PD_INV_TC_JIN.items()}
TC_DECOUPLED = 2.0 / np.log(1.0 + np.sqrt(2.0))


def phase_diagram(periodic, g_values, stage_a_factors, stage_b_window, tag):
    '''Etapas A (ubicar el pico de χ con L = 16) y B (barrido fino con tres tamaños) para cada g.'''
    tasks_a = [make_task(16, np.round(stage_a_factors * g, 5), periodic=periodic, J2=-g, nterm=1500, nsamp=4000,
                         stride=2, measure_sq=True, seed=task_seed(f"{tag}_A", int(1000 * g), r), g=g, replica=r)
               for g in g_values for r in range(2)]
    run_a = run_experiment(f"{tag}_etapa_A", tasks_a)
    peak_a = {}
    for g in g_values:
        data_a = fss_dataset(run_a, (16,), g=g)
        T_a, mom_a = data_a[16]
        peak_a[g], _, edge = fss.parabolic_peak(T_a, fss.pooled_observables(T_a, mom_a, 16)["chi"])
        if edge:
            print(f"aviso: con g = {g} el máximo de χ cayó en el borde de la ventana de la etapa A")
    tasks_b = []
    for g in g_values:
        T_b = np.round(np.linspace(stage_b_window[1], stage_b_window[0], 25) * peak_a[g], 5)
        for L in PD_L:
            tasks_b += [make_task(L, T_b, periodic=periodic, J2=-g, nterm=3000, nsamp=16_000, stride=2, measure_sq=True,
                                  seed=task_seed(f"{tag}_B", int(1000 * g), L, r), g=g, replica=r) for r in range(6)]
    run_b = run_experiment(f"{tag}_etapa_B", tasks_b)
    out = {}
    for g in g_values:
        data_b = fss_dataset(run_b, PD_L, g=g)
        T_map = {L: data_b[L][0] for L in PD_L}
        out[g] = estimators.bootstrap({L: data_b[L][1] for L in PD_L},
                                      lambda gr: fss.pair_crossings({L: (T_map[L], gr[L]) for L in gr}, [(16, 24), (24, 32)]))
    return peak_a, out


pd_peak_a, pd_pbc = phase_diagram(True, PD_G, np.round(np.arange(3.0, 1.19, -0.05), 3), (0.80, 1.08), "diagrama_fase_pbc")
```

```python
def pd_table(results, label_bc):
    rows = []
    for g, res in results.items():
        row = {"g": g}
        for key, name in (("TU_16_24", "cruce de U (16, 24)"), ("TU_24_32", "cruce de U (24, 32)"),
                          ("Txi_16_24", "cruce de ξ/L (16, 24)"), ("Txi_24_32", "cruce de ξ/L (24, 32)")):
            row[name] = fmt_pair(res[key], 4)
        row["máximo de χ, L = 32"] = fmt_pair(res["Tchi_32"], 4)
        if g in PD_TC_JIN:
            lit = PD_TC_JIN[g]
            row["T_c, Jin et al. (2013)"] = f"{lit:.4f}"
            for key, name in (("TU_24_32", "desv. U (24, 32) [%]"), ("Txi_24_32", "desv. ξ/L (24, 32) [%]")):
                v, e = res[key]
                row[name] = f"{100 * (v / lit - 1):+.2f} ± {100 * e / lit:.2f}"
        else:
            row["T_c, Jin et al. (2013)"] = "n/d"
        row["límite desacoplado 2.269 g"] = f"{TC_DECOUPLED * g:.3f}"
        rows.append(row)
    return pd.DataFrame(rows)

show_table(pd_table(pd_pbc, "PBC"), 17, "Temperatura crítica en función de g = |J2|/J1 (PBC): cruces de U y de ξ/L entre pares de tamaños, "
           "máximo de χ de L = 32 y valores de matriz de transferencia de Jin et al. (2013).", name="diagrama_fase_pbc")

fig, axes = plt.subplots(1, 2, figsize=(12.5, 4.5))
g_arr = np.array(PD_G)
for key, marker, label in (("TU_24_32", "o", "cruce de U (24, 32)"), ("Txi_24_32", "s", "cruce de ξ/L (24, 32)")):
    v = np.array([pd_pbc[g][key][0] for g in PD_G])
    e = np.array([pd_pbc[g][key][1] for g in PD_G])
    axes[0].errorbar(g_arr, v, yerr=e, fmt=marker, ms=5, capsize=2, label=label)
    axes[1].errorbar(g_arr, v / g_arr, yerr=e / g_arr, fmt=marker, ms=5, capsize=2, label=label)
g_lit = np.array(sorted(PD_TC_JIN))
axes[0].plot(g_lit, [PD_TC_JIN[g] for g in g_lit], "k*", ms=10, label="Jin et al. (2013)")
axes[1].plot(g_lit, [PD_TC_JIN[g] / g for g in g_lit], "k*", ms=10, label="Jin et al. (2013)")
g_line = np.linspace(0.5, 5.2, 100)
axes[0].plot(g_line, TC_DECOUPLED * g_line, "--", color="0.5", lw=1, label=r"$2g/\ln(1+\sqrt{2})$")
axes[1].axhline(TC_DECOUPLED, ls="--", color="0.5", lw=1)
axes[1].axvline(0.67, ls=":", color="0.5", lw=1)
axes[1].text(0.68, 0.04, r"$g^*$", color="0.4", transform=axes[1].get_xaxis_transform())
axes[0].set_xlabel(r"$g=|J_2|/J_1$")
axes[0].set_ylabel(r"$T_c$ [$J_1/k_B$]")
axes[1].set_xlabel(r"$g$")
axes[1].set_ylabel(r"$T_c/g$")
axes[1].set_xscale("log")
axes[1].set_xticks([0.6, 0.7, 1.0, 2.0, 5.0])
axes[1].xaxis.set_major_formatter(matplotlib.ticker.ScalarFormatter())
axes[1].xaxis.set_minor_formatter(matplotlib.ticker.NullFormatter())
axes[0].legend(fontsize=9)
fig.suptitle("Figura 12. Diagrama de fase del modelo J1–J2 (PBC, h = 0)", y=1.02)
plotting.savefig(fig, "fig12_diagrama_fase")
plt.show()
```

Para $g\geq0.67$ los cruces reproducen los valores de matriz de transferencia con desviaciones del orden del uno por ciento o menores (Tabla 17), lo que valida el Hamiltoniano, el signo de los acoplamientos, el motor de Metrópolis y el procedimiento de cruces con una referencia independiente. ## 9. Efecto magnetocalórico

### 9.1 Simulaciones

El EMC se estudia con $L=24$ y PBC, como pide el enunciado, para 21 campos $H=0,0.05,\dots,1.0$ y 10 réplicas por campo. Cada réplica enfría desde una configuración aleatoria a $T=3.0$ hasta $T=0.5$, con paso $0.1$ en la cola paramagnética ($T\geq1.6$), $0.05$ entre $1.5$ y $1.35$ y $0.02$ por debajo de $1.3$, donde están los dos picos del efecto, y con $N_{\mathrm{term}}=4000$ y $N_{\mathrm{muestra}}=12\,000$ MCS por temperatura. Se mide una vez cada 2 MCS la energía, la magnetización uniforme, su producto y el orden stripe.

Para comprobar la robustez de los resultados se hacen tres experimentos más con la misma malla térmica y campos $H=0,0.1,\dots,1.0$: un barrido ascendente desde el estado fundamental de cada campo (stripe si $h<h_c$, saturado si $h\geq h_c$), con 4 réplicas, más 2 réplicas que parten del estado opuesto en $h=0.3$, $0.4$ y $0.5$; el mismo barrido descendente para $L=12$, $16$, $32$ y $48$, con 4 réplicas, para estudiar la dependencia con el tamaño; y el barrido descendente con FBC, que se analiza en la Sección 10.

```python
EMC_T = np.round(np.concatenate([np.arange(3.0, 1.55, -0.1), np.arange(1.50, 1.345, -0.05),
                                 np.arange(1.30, 0.499, -0.02)]), 3)            # descendente, 60 temperaturas
EMC_H = np.round(np.arange(0.0, 1.0001, 0.05), 3)
EMC_H_COARSE = np.round(np.arange(0.0, 1.0001, 0.1), 3)
EMC_COUNTS = dict(nterm=4000, nsamp=12_000, stride=2, nblocks=20)


def emc_tasks(tag, L, fields, replicas, periodic=True, ascending=False, init=None):
    temps = EMC_T[::-1] if ascending else EMC_T
    tasks = []
    for h_val in fields:
        start = init or ("random" if not ascending else ("stripe" if h_val < H_C - 1e-9 else "ferro"))
        tasks += [make_task(L, temps, periodic=periodic, h=h_val, init=start, **EMC_COUNTS,
                            seed=task_seed(tag, L, int(round(1000 * h_val)), r), replica=r)
                  for r in range(replicas)]
    return tasks


emc_main = run_experiment("emc_pbc_l24", emc_tasks("emc_pbc_l24", 24, EMC_H, 10))
asc_tasks = emc_tasks("emc_pbc_l24_ascendente", 24, EMC_H_COARSE, 4, ascending=True)
for h_val in (0.3, 0.4, 0.5):
    opposite = "ferro" if h_val < H_C - 1e-9 else "stripe"
    asc_tasks += emc_tasks(f"emc_pbc_l24_ascendente_{opposite}", 24, [h_val], 2, ascending=True, init=opposite)
emc_asc = run_experiment("emc_pbc_l24_ascendente", asc_tasks)
emc_sizes = run_experiment("emc_pbc_tamanos", sum((emc_tasks("emc_pbc_tamanos", L, EMC_H_COARSE, 4) for L in (12, 16, 32, 48)), []))
```

```python
def emc_dataset(result, L, periodic=True, fields=None, **criteria):
    '''(H, T ascendente, {h: momentos (R, nT, NOBS)}) de un experimento de campo uniforme.'''
    H, T_e, groups = mce.field_dataset(result, L=L, periodic=periodic, **criteria)
    if fields is not None:
        keep = [h for h in H if np.any(np.isclose(h, fields))]
        H, groups = np.array(keep), {float(h): groups[float(h)] for h in keep}
    return H, T_e, groups


def emc_bootstrap(H, T_e, groups, L):
    return estimators.bootstrap(groups, lambda g: mce.scalar_summary(H, T_e, g, L, T_STAR))


H24, T_emc, G24 = emc_dataset(emc_main, 24)
N24 = 24 * 24
EMC24 = mce.analyze(H24, T_emc, G24, 24, T_STAR)
EMC24_BOOT = emc_bootstrap(H24, T_emc, G24, 24)
print(f"T* = {T_STAR:.4f} (Sección 7) separa las dos ramas; {len(H24)} campos, "
      f"{len(G24[0.0])} réplicas por campo, {len(T_emc)} temperaturas")
```

### 9.2 Respuesta al campo uniforme

La Figura 13 muestra la magnetización por espín, el calor específico y la susceptibilidad stripe a varios campos. Por encima de $T_c$ el sistema es paramagnético y la magnetización crece al bajar la temperatura. En la fase stripe, en cambio, el orden de franjas tiene magnetización nula y compite con el campo: $\langle M\rangle$ disminuye al enfriar mientras el orden se establece, y la transición stripe se desplaza a temperaturas menores al aumentar $h$. Para $h>h_c=0.4$ el estado fundamental es el saturado y la transición stripe desaparece. El panel (d) traza la frontera de la fase stripe en el plano $(T,h)$ a partir del máximo de $\chi$ a cada campo.

```python
H_show = [0.0, 0.1, 0.2, 0.3, 0.35, 0.4, 0.5, 0.7, 1.0]
cols = plotting.field_colors(H_show)
fig, axes = plt.subplots(2, 2, figsize=(12.5, 8.4))
for h_val in H_show:
    k = int(np.argmin(np.abs(H24 - h_val)))
    cur = estimators.curves_from_moments(G24[float(H24[k])], T_emc, 24, ("mu", "Cv", "chi"))
    for ax, key in zip(axes.flat[:3], ("mu", "Cv", "chi")):
        y, err = cur[key]
        plotting.errorband(ax, T_emc, y, err, cols[h_val], label=f"h = {h_val:g}", ms=2.5)
T_ch, h_ch = [], []
for k, h_val in enumerate(H24):
    obs_h = fss.pooled_observables(T_emc, G24[float(h_val)], 24)
    sel = T_emc <= 1.2
    T_pk, chi_pk, edge = fss.parabolic_peak(T_emc[sel], obs_h["chi"][sel])
    if not edge and chi_pk > 5.0:          # solo campos con una transición stripe dentro de la malla
        T_ch.append(T_pk)
        h_ch.append(h_val)
axes[1, 1].plot(T_ch, h_ch, "o-", color=plotting.COLOR_PBC, label=r"máximo de $\chi$ (L = 24)")
axes[1, 1].plot([0], [H_C], "k*", ms=12, clip_on=False, label=r"$h_c=4|J_2|-2J_1$ (T = 0)")
axes[1, 1].axvline(T_STAR, color="0.5", ls=":", lw=1)
axes[1, 1].set_xlim(0, 1.3)
axes[1, 1].set_ylim(0, 0.7)
axes[1, 1].text(0.2, 0.12, "stripe", fontsize=12)
axes[1, 1].text(0.75, 0.55, "paramagnético /\npolarizado", fontsize=11)
for ax, ylabel, title in zip(axes.flat, (r"$\langle M\rangle/N$", r"$C_h$", r"$\chi$", r"$h$ [$J_1$]"),
                             ("(a) magnetización uniforme", "(b) calor específico a campo fijo",
                              "(c) susceptibilidad stripe", "(d) frontera de la fase stripe")):
    ax.set_xlabel(r"$T$ [$J_1/k_B$]")
    ax.set_ylabel(ylabel)
    ax.set_title(title)
axes[0, 0].legend(fontsize=8, ncol=2)
axes[1, 1].legend(fontsize=9, loc="lower left")
fig.suptitle("Figura 13. Respuesta al campo uniforme (L = 24, PBC)", y=1.0)
fig.tight_layout()
plotting.savefig(fig, "fig13_respuesta_campo")
plt.show()

T_iso, mu_iso, inv_delta_u, r2_iso = mce.uniform_isotherm(H24, T_emc, G24, 24, T_STAR)
iso_boot = estimators.bootstrap(G24, lambda g: {"x": mce.uniform_isotherm(H24, T_emc, g, 24, T_STAR)[2]})
print(f"Isoterma en T = {T_iso:.2f}: |M|/N ~ h^(1/δ_u) con 1/δ_u = {fmt_pair(iso_boot['x'])} (R² = {r2_iso:.4f}, 0 < h ≤ 0.5). "
      "Un campo conjugado del parámetro de orden daría 1/δ = 1/15 (Ising 2D).")
```

La isoterma $M(h)$ en $T_c$ es lineal o algo superlineal en $h$, no sublineal como la $h^{1/\delta}$ con $\delta=15$ que produciría un campo conjugado al parámetro de orden en el Ising 2D. El campo uniforme no se acopla al orden stripe, cuya magnetización es nula, sino a la magnetización total, que no es crítica. Esto determina la forma del EMC: las leyes de campo que se deducen de la teoría de escalado para un campo conjugado (Franco et al., 2018) no tienen por qué aplicarse aquí.

### 9.3 Variación de entropía: dos caminos de cálculo

La Figura 14 compara los dos caminos de la Sección 2.4. En la fase paramagnética ($T>T^*$) coinciden. En la fase stripe no: la relación de Maxwell con una malla discreta de campos produce picos agudos en $T$, y con $\Delta h=0.1$ aparece un máximo en $T\approx0.82$ que no se mueve al aumentar $H$ por encima de $h_c$; la integración de la entropía en $T$ da una curva suave.

La razón se ve en el panel (c). A campo fijo $h<h_c$, $(\partial M/\partial T)_h$ tiene un pico muy agudo en la temperatura $T_c(h)$ en que se cruza la frontera de la fase stripe (Figura 13d), porque ahí $\langle M\rangle$ salta del valor reducido de la fase ordenada al de la paramagnética. La integral sobre el campo de esos picos es suave en $T$: al variar $T$, el pico se desplaza en $h$ y siempre contribuye el mismo salto de magnetización. La regla del trapecio, en cambio, evalúa el integrando solo en los nodos de campo, y la contribución del salto aparece concentrada en los $T_c(h_k)$ de cada nodo. La altura de esas estructuras no converge con $L$: los picos de $(\partial M/\partial T)_h$ se estrechan y crecen, y lo que de ellos capta la regla del trapecio depende de cómo caen respecto de los nodos de campo; su posición queda fijada por la malla, no por la física. El panel (b) lo muestra con los tamaños de la Sección 9.7. La integración en $T$ a campo fijo solo usa $\langle E\rangle(T)$, que es acotada y monótona, y depende de la malla de campos únicamente a través de la constante $\Delta S_M(T_{\mathrm{ref}},H)$ en $T_{\mathrm{ref}}=3.0$, donde el integrando de Maxwell es suave. La Tabla 18 comprueba que el resultado es insensible a la elección de $T_{\mathrm{ref}}$ y a la resolución de la malla térmica. Por eso todo lo que sigue usa la integración en temperatura.

```python
coarse = np.isin(np.round(H24, 3), EMC_H_COARSE)
dS_max_coarse = mce.delta_S_maxwell(H24[coarse], EMC24["dMdT"][coarse])
fig, axes = plt.subplots(1, 3, figsize=(17, 4.6))
for h_val, col in zip((0.3, 1.0), (plotting.WONG[0], plotting.WONG[2])):
    k = int(np.argmin(np.abs(H24 - h_val)))
    kc = int(np.argmin(np.abs(H24[coarse] - h_val)))
    axes[0].plot(T_emc, EMC24["dS"][k] / N24, "-", color=col, lw=2.2, label=f"H = {h_val:g}, integración en T")
    axes[0].plot(T_emc, dS_max_coarse[kc] / N24, "--", color=col, lw=1.2, label=f"H = {h_val:g}, Maxwell, Δh = 0.1")
    axes[0].plot(T_emc, EMC24["dS_maxwell"][k] / N24, ":", color=col, lw=1.4, label=f"H = {h_val:g}, Maxwell, Δh = 0.05")
axes[0].axhline(0, color="0.3", lw=0.7)
axes[0].axvline(T_STAR, color="0.5", ls=":", lw=1)
axes[0].set_xlim(0.5, 2.0)
axes[0].set_xlabel(r"$T$ [$J_1/k_B$]")
axes[0].set_ylabel(r"$\Delta S_M/N$ [$k_B$]")
axes[0].set_title("(a) los dos caminos, L = 24")
axes[0].legend(fontsize=8.5)

size_rows = {}
for L in (12, 16, 24, 32, 48):
    res_L, f_L = (emc_main, EMC_H_COARSE) if L == 24 else (emc_sizes, None)
    H_L, T_L, G_L = emc_dataset(res_L, L, fields=f_L)
    an_L = mce.analyze(H_L, T_L, G_L, L, T_STAR)
    k = int(np.argmin(np.abs(H_L - 0.5)))
    maxwell_pk = mce.branch_peaks(T_L, an_L["dS_maxwell"][k], T_STAR)["dS_inv"] / L**2
    thermal_pk = an_L["peaks"][k]["dS_inv"] / L**2
    size_rows[L] = (maxwell_pk, thermal_pk, H_L, T_L, G_L, an_L)
Ls_emc = np.array(sorted(size_rows))
axes[1].plot(Ls_emc, [size_rows[L][0] for L in Ls_emc], "s--", label="Maxwell (Δh = 0.1)")
axes[1].plot(Ls_emc, [size_rows[L][1] for L in Ls_emc], "o-", label="integración en T")
axes[1].set_xlabel(r"$L$")
axes[1].set_ylabel(r"máximo de la rama inversa, $\Delta S_M/N$")
axes[1].set_title("(b) H = 0.5: dependencia con el tamaño")
axes[1].legend()

for h_val, col in zip((0.1, 0.2, 0.3, 0.35, 0.4, 0.5), plotting.WONG):
    k = int(np.argmin(np.abs(H24 - h_val)))
    axes[2].plot(T_emc, EMC24["dMdT"][k] / N24, color=col, lw=1.3, label=f"h = {h_val:g}")
axes[2].set_xlim(0.5, 1.4)
axes[2].set_xlabel(r"$T$ [$J_1/k_B$]")
axes[2].set_ylabel(r"$(\partial M/\partial T)_h/N$")
axes[2].set_title("(c) integrando de la relación de Maxwell")
axes[2].legend(fontsize=8)
fig.suptitle("Figura 14. Variación de entropía por la relación de Maxwell y por integración de la entropía en T", y=1.03)
plotting.savefig(fig, "fig14_caminos_entropia")
plt.show()

# Sensibilidad del camino térmico: temperatura de referencia y resolución de la malla térmica
th24 = mce.thermodynamics(H24, T_emc, G24, 24)
rows = []
for label, T_ref_val, step in (("referencia: T_ref = 3.0, malla completa", 3.0, 1), ("T_ref = 2.0", 2.0, 1),
                               ("T_ref = 1.5", 1.5, 1), ("malla térmica cada 2 nodos", 3.0, 2)):
    sel = np.arange(len(T_emc))
    sel = sel[(T_emc[sel] <= T_ref_val + 1e-9)]
    if step == 2:
        sel = np.unique(np.concatenate([sel[::-2][::-1], [sel[-1]]]))
    T_s = T_emc[sel]
    dS_m = mce.delta_S_maxwell(H24, th24["dMdT"][:, sel])
    dS_t = mce.delta_S_thermal(T_s, th24["E"][:, sel], dS_m[:, -1])
    ref_interp = np.array([np.interp(T_s, T_emc, EMC24["dS"][k]) for k in range(len(H24))])
    rows.append({"variante": label, "máx. |Δ(ΔS_M/N)| frente a la referencia": f"{np.max(np.abs(dS_t - ref_interp)) / N24:.1e}"})
show_table(pd.DataFrame(rows), 18, "Sensibilidad de ΔS_M obtenida por integración en T a la temperatura de referencia y a la "
           "resolución de la malla térmica (L = 24, todos los campos y temperaturas comunes).", name="sensibilidad_camino_termico")
```

### 9.4 Dos ramas: efecto magnetocalórico inverso y convencional

La Figura 15 muestra $\Delta S_M(T,H)$ para todos los campos y su valor en $T^*$. Por encima de $T^*$ el efecto es **convencional** ($\Delta S_M<0$): el campo alinea los espines de la fase paramagnética y reduce la entropía, y el mínimo de $\Delta S_M$ queda claramente por encima de $T^*$, más cerca de ella a campos altos que a campos bajos. Por debajo de $T^*$ el efecto es **inverso** ($\Delta S_M>0$): el campo compite con el orden stripe, reduce su temperatura de transición (Figura 13d) y aumenta la entropía. La rama inversa crece con $H$ mientras el campo debilita el orden de franjas, alcanza su máximo cuando $H$ se acerca a la frontera de la fase stripe a esa temperatura, en torno a $h_c$, y decrece para campos mayores, cuando el campo ya polariza el sistema. Para campos suficientemente altos desaparece y el efecto es convencional a todas las temperaturas. En consecuencia, $\Delta S_M(T^*,H)$ cambia de signo a un campo intermedio. Ninguna de las dos ramas tiene su extremo en $T^*$; la Tabla 19 da los picos de cada rama.

### 9.5 Leyes de campo y RCP

En el enfoque de escalado de Franco et al. (2018), con un campo conjugado al parámetro de orden, el máximo de $|\Delta S_M|$ escala como $H^n$ con $n=1+(\beta-1)/(\beta+\gamma)$ en la transición ($n=2/3$ en campo medio) y el RCP como $H^{p}$ con $p=1+1/\delta$. Aquí el campo no es conjugado (Sección 9.2) y la referencia natural es la respuesta lineal: si $M=\chi_u(T)H$ a campo bajo, la relación de Maxwell da $\Delta S_M=\tfrac12(d\chi_u/dT)H^2$, es decir $n=2$, y con un FWHM casi independiente de $H$, también $p\approx2$. La Tabla 21 ajusta las leyes de potencia, con $0.1\leq H\leq1$, en la rama convencional, que es limpia y monótona en $H$, y en el máximo global de $|\Delta S_M|$, que pasa de una rama a la otra al crecer $H$ y por eso no tiene por qué seguir una ley de potencia. Se excluye $H=0.05$, donde la rama convencional es del orden del ruido estadístico y su ancho no está bien definido. La Figura 16 muestra los ajustes y el exponente local $d\ln|y|/d\ln H$.

### 9.8 Lectura de los resultados

Los dos protocolos de barrido dan los mismos promedios y los mismos picos y exponentes dentro del error (Tablas 22 y 23): las dos ramas son propiedades de equilibrio y no efectos del camino. Las diferencias nodo a nodo de más de $2.58\sigma$ son algo más frecuentes que el 1 % esperado para una distribución normal, pero están repartidas en todo el rango de temperaturas, incluida la fase paramagnética, y son las esperables cuando el error de un protocolo se estima con solo 4 réplicas: con tan pocas réplicas el cociente $z$ sigue aproximadamente una $t$ de Student con colas pesadas. La excepción física está cerca de $h_c$ y por debajo de $T\approx0.6$, donde el estado inicial puede dejar huella en $\langle M\rangle$ (Tabla 24): con $h=0.3<h_c$, una réplica que parte del estado saturado puede conservar una magnetización residual que no aparece partiendo del stripe ni de una configuración aleatoria, señal de metaestabilidad entre las dos fases. Como la entropía se integra desde la temperatura más alta hacia abajo, esas diferencias solo afectan a $\Delta S_M$ en las temperaturas donde ocurren, no por encima de ellas. Con la integración en temperatura, las amplitudes por espín convergen con $L$: la de la rama inversa en $H=0.5$ cambia un 3 % entre $L=16$ y $48$ y menos de un 0.1 % entre $L=32$ y $48$, mientras que el máximo inverso obtenido con la relación de Maxwell y $\Delta h=0.1$ varía de forma errática, entre $0.14$ y $0.40$, en el mismo intervalo de tamaños (Tabla 25, Figura 14b). A campo bajo la rama inversa todavía crece con $L$, porque su pico está junto a $T^*$, donde el redondeo de tamaño finito es mayor. La rama convencional converge desde $L\approx24$: entre $L=24$ y $48$ su amplitud en $H=0.5$ cambia menos de un 2 % y $n$ queda entre $2.20$ y $2.23$, mientras que los dos tamaños menores dan amplitudes y valores de $n$ algo menores. $p$ tiene errores grandes en esos dos tamaños, donde el ancho de la rama convencional a campo bajo está mal definido. El campo del cambio de signo en $T^*$ baja ligeramente al crecer $L$, de $0.59$–$0.60$ para $L\leq16$ a $0.56$ para $L=48$.

**Pregunta 3: por qué el EMC es grande cerca de la transición.** $\Delta S_M$ es la integral sobre el campo de $(\partial M/\partial T)_h$, y esa derivada es grande donde la magnetización responde con fuerza a la temperatura: cerca de la transición y, aquí, sobre la frontera de la fase stripe en el plano $(T,h)$. En este modelo la respuesta tiene dos regímenes con signos opuestos. En la fase paramagnética el campo alinea los espines y reduce la entropía (EMC convencional), con un máximo por encima de $T_c$ que se acerca a ella al crecer $H$. En la fase stripe el campo compite con el orden antiferromagnético y aumenta la entropía (EMC inverso), con un máximo cuando el campo alcanza la frontera de la fase ordenada, en torno al campo crítico $h_c=4|J_2|-2J_1$ en que a $T=0$ el estado saturado iguala al stripe. Esta estructura de dos ramas, con cambio de signo en la transición, es la propia de los antiferromagnetos y no aparece con un ferromagneto simple. El "pico característico en $T_N$" que describe Markna et al. (2026) no se observa: ninguna de las dos ramas tiene su extremo en $T_c$.

**Pregunta 4: poder de enfriamiento relativo.** El RCP de la rama convencional crece con el campo como una ley de potencia limpia con $p\approx2$ (Tabla 21). El de la rama inversa es mayor que el convencional para $H\leq h_c$, entre 1.5 y 1.6 veces en $H=0.3$–$0.4$, y cae rápidamente por encima de $h_c$ (Tabla 19). Un RCP grande combina una variación de entropía apreciable con un intervalo de temperaturas amplio, y es por eso la figura de mérito habitual de un refrigerante magnético. Los valores de este trabajo están en unidades reducidas, son por espín y corresponden a un modelo bidimensional, de modo que no son comparables en magnitud con el RCP de Markna et al. (2026), $\approx46\,\mathrm{J\,kg^{-1}}$ a $10\,\mathrm{T}$ para el compuesto sin dopar. Lo que sí es comparable es la forma de las leyes de campo, y ahí la diferencia es cualitativa: con un campo uniforme que no es conjugado del parámetro de orden, este modelo está en el régimen de respuesta lineal, con $n\to2$ a campo bajo, en vez del $n\approx2/3$ de un campo conjugado (Sección 11).

## 10. Fronteras libres

Con fronteras libres los espines del borde tienen menos vecinos y el orden stripe es más débil cerca de la superficie. En un sistema finito la temperatura pseudocrítica queda entonces por debajo de $T_c$, con un corrimiento que decae como $1/L$ (la fracción de espines de borde es $\approx4/L$), mientras que con PBC decae como $L^{-1/\nu}$. La temperatura crítica del límite termodinámico no depende de las fronteras, de modo que las extrapolaciones en $L$ de las magnitudes de FBC deben converger a la $T_c$ de PBC. Esta sección repite con FBC el escalado de tamaño finito (con 8 réplicas por tamaño), el diagrama de fase y el EMC.

### 10.1 Escalado de tamaño finito

```python
FBC_T = np.round(np.arange(1.10, 0.619, -0.02), 3)
fbc_tasks = []
for L in FSS_L:
    nterm, nsamp = fss_counts(L, FSS_STRIDE[L])
    fbc_tasks += [make_task(L, FBC_T, periodic=False, nterm=nterm, nsamp=nsamp, stride=FSS_STRIDE[L], measure_sq=True,
                            seed=task_seed("fss_fbc", L, r), replica=r) for r in range(FSS_REPLICAS)]
fss_fbc_run = run_experiment("fss_fbc", fbc_tasks)
FBC_DATA = fss_dataset(fss_fbc_run, FSS_L)
T_fbc = FBC_DATA[FSS_L[0]][0]
fbc_curves = {L: estimators.curves_from_moments(FBC_DATA[L][1], T_fbc, L) for L in FSS_L}
# Con FBC el máximo de χ de los tamaños menores puede quedar por debajo de la malla (T < 0.62): esos tamaños
# no entran en el ajuste de γ/ν, y lo mismo para C.
FBC_EDGE = {key: [L for L in FSS_L if fss.parabolic_peak(T_fbc, fss.pooled_observables(T_fbc, FBC_DATA[L][1], L)[key])[2]]
            for key in ("chi", "Cv")}
FBC_CHI_L = [L for L in FSS_L if L not in FBC_EDGE["chi"]]
FBC_CV_L = [L for L in FSS_L if L not in FBC_EDGE["Cv"]]


def fbc_statistics(groups):
    data = {L: (T_fbc, groups[L]) for L in groups}
    out = fss.fss_statistics(data, L1_min=16, do_collapse=False, chi_sizes=FBC_CHI_L, cv_sizes=FBC_CV_L)
    Ls = sorted(groups)
    out["TCv_inf"], _ = fss.extrapolate_inverse_L(Ls, [out[f"TCv_{L}"] for L in Ls])
    out["TCv_inf16"], _ = fss.extrapolate_inverse_L(Ls[1:], [out[f"TCv_{L}"] for L in Ls[1:]])
    return out

FBC_BOOT = estimators.bootstrap({L: FBC_DATA[L][1] for L in FSS_L}, fbc_statistics)

fig, axes = plt.subplots(2, 2, figsize=(12, 8.2), sharex=True)
for ax, (key, label) in zip(axes.flat, (("xi_L", r"$\xi/L$"), ("U", r"$U$"), ("chi", r"$\chi$"), ("Cv", r"$C$"))):
    for L in FSS_L:
        y, err = fbc_curves[L][key]
        ax.errorbar(T_fbc, y, yerr=err, marker="o", ms=3, lw=1.1, elinewidth=0.7, color=plotting.color_for_L(L), label=f"L = {L}")
    ax.axvline(T_STAR, color="0.5", ls=":", lw=1)
    ax.set_ylabel(label)
    ax.legend(fontsize=9)
for ax in axes[1]:
    ax.set_xlabel(r"$T$ [$J_1/k_B$]")
axes[0, 0].set_ylim(0, 1.5)
fig.suptitle("Figura 19. Escalado de tamaño finito con fronteras libres (h = 0; línea punteada: T* de PBC)", y=1.0)
fig.tight_layout()
plotting.savefig(fig, "fig19_fbc_curvas")
plt.show()

def fbc_peak_cell(key, L, q):
    return f"< {T_fbc.min():.2f} (fuera de la malla)" if L in FBC_EDGE[q] and key.startswith("T") else (
        "n/d" if L in FBC_EDGE[q] else fmt_pair(FBC_BOOT[f"{key}_{L}"]))

rows = []
for L in FSS_L:
    rows.append({"L": L, "T del máximo de χ": fbc_peak_cell("Tchi", L, "chi"), "χ máx.": fbc_peak_cell("chimax", L, "chi"),
                 "T del máximo de C": fbc_peak_cell("TCv", L, "Cv"), "C máx.": fbc_peak_cell("Cvmax", L, "Cv")})
show_table(pd.DataFrame(rows), 26, "Máximos de χ y C con fronteras libres (h = 0). Un máximo que queda en el borde inferior de la "
           "malla térmica no está resuelto y no entra en los ajustes de la Tabla 27.", name="fbc_por_tamano")

rows = [
    {"estimación": "mediana de los cruces de ξ/L (pares con L1 ≥ 16)", "FBC": fmt_pair(FBC_BOOT["T_star"], 3),
     "PBC (Sección 7)": fmt_pair(FSS_MAIN["T_star"], 4)},
    {"estimación": "mediana de los cruces de U (pares con L1 ≥ 16)", "FBC": fmt_pair(FBC_BOOT["T_star_U"], 3),
     "PBC (Sección 7)": fmt_pair(FSS_MAIN["T_star_U"], 4)},
    {"estimación": "máximo de C, extrapolación lineal en 1/L (L = 12–32)", "FBC": fmt_pair(FBC_BOOT["TCv_inf"], 3),
     "PBC (Sección 7)": fmt_pair(FSS_MAIN["T_star"], 4)},
    {"estimación": "máximo de C, extrapolación lineal en 1/L (L = 16–32)", "FBC": fmt_pair(FBC_BOOT["TCv_inf16"], 3),
     "PBC (Sección 7)": fmt_pair(FSS_MAIN["T_star"], 4)},
    {"estimación": f"γ/ν efectivo (χ máx., L = {FBC_CHI_L[0]}–{FBC_CHI_L[-1]})", "FBC": fmt_pair(FBC_BOOT["gamma_nu"]),
     "PBC (Sección 7)": fmt_pair(FSS_MAIN["gamma_nu"])},
    {"estimación": f"α/ν efectivo (C máx., L = {FBC_CV_L[0]}–{FBC_CV_L[-1]})", "FBC": fmt_pair(FBC_BOOT["alpha_nu"]),
     "PBC (Sección 7)": fmt_pair(FSS_MAIN["alpha_nu"])},
]
show_table(pd.DataFrame(rows), 27, "Temperatura crítica estimada con fronteras libres frente a la de PBC, y exponentes efectivos de los "
           "máximos con FBC.", name="fbc_temperatura_critica")
```

El valor de $\alpha/\nu$, lejos del $0$ del Ising 2D, descarta esa clase y, junto con $1/\nu$, sitúa el comportamiento efectivo cerca del Potts de 4 estados, extremo de la línea de Ashkin–Teller que la literatura sitúa en $g^*\approx0.67$. Las relaciones de escala se cumplen dentro de los errores. Hasta $L=64$ no aparece ninguna señal concluyente de primer orden: no hay doble pico en la energía, ni saturación de $2/3-V_{\min}$, ni deriva de los exponentes, y el único rasgo de ese tipo, un mínimo del cumulante de Binder que se profundiza con $L$, aparece también en transiciones continuas cercanas al punto de Potts (Jin et al., 2012). Esto es consistente con una transición continua o con una de primer orden tan débil que su longitud de correlación supera los tamaños simulados; con estos datos no puede decidirse cuál de las dos lecturas de la literatura es la correcta para $g=0.6$. El diagrama de fase $T_c(g)$ reproduce los valores de matriz de transferencia de Jin et al. (2013) con desviaciones de alrededor del uno por ciento, lo que valida el Hamiltoniano y el método.

**El efecto magnetocalórico.** $\Delta S_M(T,H)$ tiene dos ramas separadas por la transición: inversa por debajo de $T_c$, donde el campo uniforme compite con el orden stripe y aumenta la entropía, y convencional por encima, donde el campo alinea los espines paramagnéticos. La rama inversa alcanza su máximo, de alrededor de $0.25\,k_B$ por espín con $H\approx0.3$–$0.35$, cuando el campo se acerca a la frontera de la fase stripe, cuya escala fija el campo crítico $h_c=4|J_2|-2J_1=0.4$ de la competencia entre las dos interacciones, y desaparece a campos mayores; en consecuencia, $\Delta S_M(T_c,H)$ cambia de signo a un campo intermedio y ninguna rama tiene su extremo en $T_c$. Con un campo uniforme, que no es conjugado del parámetro de orden stripe, la respuesta es lineal a campo bajo: la rama convencional escala con $n\to2$ y el RCP con $p\approx2$, y $\Delta T_{ad}$ en el pico convencional con $m\approx1.8$. El cálculo de $\Delta S_M$ por la relación de Maxwell con una malla discreta de campos produce, en la fase ordenada, estructuras que dependen de la malla y crecen con $L$; la integración de la entropía en temperatura las elimina y da amplitudes por espín que convergen con el tamaño. El resultado no depende del protocolo de barrido (enfriamiento o calentamiento desde el estado fundamental).

**El papel de la competencia entre interacciones.** La frustración fija el tipo de orden ($g>1/2$), reduce la temperatura crítica respecto de la de dos redes de Ising desacopladas y la anula al acercarse a $g=1/2$, determina la clase de universalidad (a través de la simetría $\mathbb Z_4$ del stripe) y fija, mediante $h_c=4|J_2|-2J_1$, la escala de campo de la rama inversa del EMC.

**Tamaño y fronteras.** Con PBC los efectos de tamaño finito siguen las leyes de escala de la Sección 7. Con FBC las temperaturas pseudocríticas quedan muy por debajo de $T_c$ y se acercan a ella al crecer $L$, el máximo de $C$ es más bajo, el de $\chi$ crece más despacio y la rama inversa del EMC es mucho más débil. Los cruces de cocientes adimensionales quedan, con $L\leq32$, alrededor de un 2 % por debajo de la $T_c$ de PBC, y el diagrama de fase con FBC sigue la curva de la literatura para $g$ grande. Las diferencias son efectos de superficie de un sistema finito; los espines de los bordes donde terminan las franjas, que se voltean con un costo de solo $0.4\,J_1$, dominan además la termodinámica de FBC a baja temperatura.

**Comparación con el artículo.** El modelo reducido reproduce la existencia de la transición y un EMC apreciable en un intervalo ancho de temperaturas, pero no los exponentes críticos (otra clase de universalidad) ni las leyes de campo ni el pico único en $T_N$ (otro acoplamiento entre campo y parámetro de orden). Esas diferencias no son defectos de resolución o de estadística, sino consecuencias de la dimensionalidad y de la simetría del orden stripe.

**Cuestiones abiertas.** El orden de la transición en $g=0.6$, que exigiría tamaños mucho mayores y algoritmos sin *critical slowing down*; la dependencia del EMC con $g$ a ambos lados de $g^*$; y el EMC con un campo conjugado al orden stripe, que debería recuperar leyes de campo de tipo crítico (Sección 13).

## 13. Limitaciones y perspectivas

**Limitaciones.**

- El modelo es una reducción fuerte del sistema del artículo: bidimensional, con espines de Ising y sin acoplamiento entre planos ni anisotropía. Sirve para estudiar la física de la frustración $J_1$–$J_2$ y la estructura del EMC en un antiferromagneto stripe, no para predecir valores del material.
- Los exponentes son efectivos para $L\leq64$ y sus errores son solo estadísticos. Cerca del punto de Potts de 4 estados las correcciones al escalado son logarítmicas y pueden ser comparables a esos errores.
- El orden de la transición en $g=0.6$ no queda decidido: hasta $L=64$ no hay señales concluyentes de primer orden, pero una transición de primer orden extremadamente débil solo se manifestaría con tamaños mayores que su longitud de correlación.
- El tiempo de autocorrelación del Metrópolis local crece en la transición como $L^{z}$ con $z\approx2$, lo que limita los tamaños accesibles con costo razonable; el propio $z$ está mal determinado con cuatro tamaños y pocas réplicas.
- El EMC se estudió con campos $H\leq1$, en una malla de paso $0.05$ ($L=24$) o $0.1$ (otros tamaños y FBC); la integración en temperatura no depende de esa malla salvo por la constante de referencia en $T=3.0$, pero las leyes de campo sí dependen de cuántos campos haya.
- Con fronteras libres solo se usaron tamaños hasta $L=32$: las extrapolaciones lineales en $1/L$ de las temperaturas pseudocríticas no son fiables con esos tamaños, y los cruces todavía quedan un 2 % por debajo de $T_c$.

**Algoritmos de clúster.** El problema que resuelven es el *critical slowing down*: los algoritmos de Swendsen y Wang (1987) y de Wolff (1989) construyen clústeres de espines conectados con una probabilidad que depende del acoplamiento y los voltean a la vez, lo que en el Ising ferromagnético reduce el exponente dinámico de $z\approx2.17$ a valores por debajo de $0.5$. Para este modelo serían útiles precisamente donde el Metrópolis local es más lento, cerca de $T_c$ y con $L$ grande. La dificultad es la frustración: con acoplamientos que compiten no todos los enlaces pueden satisfacerse a la vez, los clústeres de Fortuin–Kasteleyn percolan por encima de $T_c$ y llegan a ocupar casi toda la red, y el algoritmo pierde su eficiencia o deja de ser útil. Existen construcciones específicas para sistemas frustrados (Kandel et al., 1990), pero no son una generalización directa ni se garantiza que funcionen para este Hamiltoniano. Introducir un algoritmo de clúster exigiría reemplazar la regla de actualización del núcleo de Monte Carlo, validar que satisface balance detallado con la enumeración exacta de la Sección 4.2 y comprobar que reproduce los resultados de las Secciones 7 y 8 dentro del error antes de usarlo con tamaños mayores. Por esas razones se dejó fuera del núcleo de este trabajo.

**Alternativas más directas.** El *parallel tempering* o intercambio de réplicas (Hukushima y Nemoto, 1996) mantiene la actualización local pero intercambia configuraciones entre temperaturas vecinas, lo que acelera la decorrelación cerca de $T_c$ y en la fase ordenada, y requiere cambios menores: las réplicas a distintas temperaturas ya se simulan, y bastaría con correrlas simultáneamente y añadir el paso de intercambio. El muestreo de Wang y Landau (2001) estima directamente la densidad de estados $g(E)$, de la que se obtienen la entropía, el calor específico y los histogramas de energía a cualquier temperatura; sería especialmente adecuado tanto para el EMC, que es una diferencia de entropías, como para buscar la doble joroba de una transición de primer orden débil.

**Extensiones.** Estudiar el EMC en función de $g$ a ambos lados de $g^*\approx0.67$, donde la literatura sitúa el cambio de transición de primer orden a continua; aplicar un campo alternante conjugado al orden stripe, que sí debería producir leyes de campo del tipo de Franco et al. (2018); y añadir un acoplamiento entre planos para acercarse al sistema tridimensional del artículo.
