## Planteamiento del problema

El estudio del movimiento de sólidos rígidos constituye uno de los pilares fundamentales de la mecánica clásica. En su formulación más simple, el movimiento de un sólido puede describirse mediante la formulación lagrangiana, estudiando por separado la traslación y la rotación del cuerpo. Sin embargo, cuando un sólido se desplaza dentro de un fluido, su dinámica se vuelve considerablemente más compleja debido a la interacción con el medio.

Un caso particularmente interesante de esta interacción es el de un sólido rígido en rotación que se mueve a través de un fluido. En estas condiciones aparece una fuerza lateral, perpendicular a la velocidad de traslación, conocida como fuerza de Magnus. Este fenómeno es responsable de trayectorias curvas en numerosos sistemas físicos y desempeña un papel central en el comportamiento de objetos cotidianos como los balones de fútbol.

## Objetivos

* Modelar la dinámica de un balón de fútbol en un fluido.
* Usar métodos computacionales para resolver numéricamente las ecuaciones de movimiento.
* Analizar los distintos parámetros (velocidad angular, inclinación del eje de rotación y viento) y su influencia en la trayectoria del balón.
* Graficar dichas trayectorias y compararlas visualmente.


## Modelo físico

El efecto Magnus, llamado así en honor al físico y químico alemán Heinrich Gustav Magnus (1802-1870), es el nombre que recibe el fenómeno por el cual la rotación de un objeto afecta su trayectoria al desplazarse a través de un fluido, como el aire (Magnus, 1853).

Un objeto en rotación arrastra una fina capa de fluido a su alrededor. En el lado del objeto donde el giro tiene el mismo sentido que la corriente incidente, la velocidad relativa del fluido aumenta; en el lado opuesto, disminuye. Por el principio de Bernoulli, la presión se reduce donde la velocidad es mayor, de modo que aparece una diferencia de presión entre ambos lados del objeto y, con ella, una fuerza neta perpendicular a la dirección de la corriente. Esta fuerza desvía al objeto respecto de la trayectoria que tendría en ausencia de fluido.

La fuerza de Magnus sobre un objeto giratorio se define como

$$
\vec{F}_M = \frac{1}{2}\,\rho\,C_L\,A\,(\vec{\omega} \times \vec{v}_{\mathrm{rel}})
$$

donde $C_L$ es un coeficiente adimensional determinado por la rugosidad y la forma del sólido, $\rho$ es la densidad del fluido, $A$ es el área efectiva (la sección transversal) del sólido, $\vec{\omega}$ es la velocidad angular del sólido y $\vec{v}_{\mathrm{rel}}$ es la velocidad relativa entre el sólido y el fluido:

$$
\vec{v}_{\mathrm{rel}} = \dot{\vec{r}} - \vec{v}_{\mathrm{fluido}}
$$

Adicionalmente, el movimiento a través de un fluido implica una fuerza de arrastre que se opone al movimiento relativo:

$$
\vec{F}_D = -\frac{1}{2}\,\rho\,C_D\,A\,\lvert \vec{v}_{\mathrm{rel}} \rvert\,\vec{v}_{\mathrm{rel}}
$$

donde $C_D$ es el coeficiente de arrastre.

Como el objetivo es estudiar la trayectoria de un balón de fútbol, se modela el cuerpo rígido como una esfera ideal, indeformable y de densidad uniforme. Al tratarse de una esfera y dado que, en este contexto, los efectos del fluido sobre la rotación de la pelota son despreciables, se supone $\vec{\omega}$ constante. El fluido, por su parte, se supone continuo, incompresible y newtoniano, y los coeficientes $C_L$ y $C_D$ se toman constantes. Solo se consideran las fuerzas mencionadas y la gravitatoria (modelada como un campo uniforme); no se contempla ningún tipo de colisión durante el vuelo del balón.

Al tomar $\vec{\omega}$ constante, esta deja de ser una variable dinámica, de modo que la energía cinética del balón es

$$
T = \frac{1}{2}\,m\,\dot{\vec{r}}^{\,2}
$$

y el lagrangiano del sólido resulta

$$
\mathcal{L} = \frac{1}{2}\,m\,\dot{\vec{r}}^{\,2} - m g z
$$

Al introducir $\vec{F}_M$ y $\vec{F}_D$ como fuerzas generalizadas en las ecuaciones de Euler-Lagrange, las ecuaciones de movimiento quedan

$$
m\,\ddot{\vec{r}}
=
-\,m g\,\hat{z}
-\frac{1}{2}\,\rho\,C_D\,A\,\lvert \vec{v}_{\mathrm{rel}} \rvert\,\vec{v}_{\mathrm{rel}}
+\frac{1}{2}\,\rho\,C_L\,A\,(\vec{\omega} \times \vec{v}_{\mathrm{rel}})
$$

$\vec{F}_M$ y $\vec{F}_D$ se introducen como fuerzas generalizadas, y no como parte de un potencial, porque ninguna de las dos deriva de un potencial: $\vec{F}_D$ es disipativa (como la fricción dinámica) y $\vec{F}_M$ depende explícitamente de la velocidad del sólido. Ambas características son incompatibles con un potencial $V(\vec r)$ del que se derivarían por $-\nabla V$.


## Integración numérica de las ecuaciones de movimiento

Las ecuaciones de movimiento se resuelven numéricamente con `solve_ivp` de SciPy (Virtanen et al., 2020), usando por defecto el método explícito Runge-Kutta 4(5) de Dormand y Prince (RK45), un método de paso adaptativo que estima el error de truncamiento comparando una solución de orden 4 con una de orden 5 obtenidas a partir de las mismas evaluaciones de la función (Dormand & Prince, 1980).

Esta es una integración temporal: el estado en $t+\Delta t$ depende del estado en $t$, por lo que el proceso es intrínsecamente secuencial y no puede paralelizarse sin cambiar el algoritmo. En cambio, sí es posible acelerar cada evaluación de la función que calcula $\dot{\vec y}$ (ver más abajo) y paralelizar los barridos de parámetros, en los que cada combinación de velocidad angular, inclinación del eje o viento es independiente de las demás.


## Análisis de parámetros

Se simulan tres escenarios de viento: sin viento ($\vec{v}_{\mathrm{fluido}} = 0$), viento frontal ($\vec{v}_{\mathrm{fluido}} = (-5, 0, 0)$ m/s) y viento lateral ($\vec{v}_{\mathrm{fluido}} = (0, -5, 0)$ m/s). Para cada escenario se realiza un barrido en la magnitud de la velocidad angular (con inclinación del eje fija) y un barrido en la inclinación del eje de rotación (con magnitud de $\vec\omega$ fija).

Antes de entrar en los barridos, se ilustra una trayectoria de ejemplo con la configuración de un tiro con rosca ($|\vec{\omega}| = 50$ rad/s, $\theta = 20°$).


## Parámetros del modelo

```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt
from concurrent.futures import ThreadPoolExecutor

# --- Parámetros físicos (balón de fútbol estándar) ---
m = 0.43            # masa [kg]
R = 0.111           # radio [m]
A = np.pi * R**2    # área frontal / sección transversal [m^2]
rho = 1.225         # densidad del aire [kg/m^3]
g = 9.81            # aceleración de la gravedad [m/s^2]

C_D = 0.25          # coeficiente de arrastre (adimensional)
C_L = 0.2           # coeficiente de Magnus (adimensional)

# --- Condiciones iniciales, comunes a todas las simulaciones de este notebook ---
r0 = np.array([0.0, 0.0, 0.0])     # posición inicial del balón [m]
v0 = np.array([20.0, 0.0, 12.0])   # velocidad inicial del balón [m/s]
y0 = np.concatenate((r0, v0))      # estado inicial [x, y, z, vx, vy, vz]

# --- Configuración temporal de la integración ---
t_span = (0.0, 1.2)                    # intervalo de integración [s]
t_eval = np.linspace(0.0, 1.2, 1000)   # instantes en los que se guarda la solución
```

## Dinámica del balón

La función que sigue calcula $\dot{\vec y} = [\vec v,\, \vec a]$ para el sistema de primer orden equivalente a la ecuación de movimiento. Es la función que `solve_ivp` evalúa muchas veces por cada paso de integración, así que es la candidata natural para acelerar con **Numba** (`@njit`): al compilarla a código máquina la primera vez que se usa, las llamadas posteriores se ejecutan sin el overhead del intérprete de Python.

Como no todos los entornos donde puede correr este notebook tienen Numba instalado, el import se protege con `try/except`: si Numba no está disponible, `@njit` se reemplaza por un decorador que no hace nada, y el notebook sigue funcionando igual (aunque más lento).

`omega` y `v_fluido` se pasan como argumentos explícitos de la función (a través de `args=` en `solve_ivp`), en vez de leerse como variables globales. Esto es importante por dos razones:

1. **Compatibilidad con Numba:** una función compilada con `@njit` trata a las variables globales como constantes fijadas en tiempo de compilación; si `omega` cambiara después de la primera llamada, la función compilada seguiría usando el valor viejo. Pasar los parámetros como argumentos evita ese problema de raíz.
2. **Robustez frente al orden de ejecución:** al no depender de variables globales que se reescriben en un lugar del notebook y se leen desde otro, cada barrido de parámetros (ver más abajo) queda completamente autocontenido y no puede verse afectado por el estado dejado por una celda anterior.


```python
try:
    from numba import njit
    NUMBA_DISPONIBLE = True
except ImportError:
    # Decorador de respaldo: si Numba no está instalado, @njit simplemente
    # devuelve la función sin modificarla. Soporta tanto "@njit" como
    # "@njit(cache=True, nogil=True)".
    def njit(*args, **kwargs):
        def decorador(func):
            return func
        if len(args) == 1 and callable(args[0]) and not kwargs:
            return args[0]
        return decorador
    NUMBA_DISPONIBLE = False


@njit(cache=True, nogil=True)
def ecuaciones_rhs(t, y, omega, v_fluido, m, A, rho, g, C_D, C_L):
    """Calcula dy/dt para y = [x, y, z, vx, vy, vz].

    Físicamente: la derivada de la posición es la velocidad, y la derivada de
    la velocidad es la aceleración, que resulta de sumar la gravedad, el
    arrastre aerodinámico (F_D, opuesto a la velocidad relativa con el aire) y
    la fuerza de Magnus (F_M, perpendicular tanto a omega como a la velocidad
    relativa).

    Los componentes se calculan "a mano" (en vez de con np.cross / np.linalg.norm
    sobre arreglos) por dos razones:
    - Es matemáticamente idéntico (misma fórmula del producto cruz y de la norma
      euclidiana), así que no cambia ningún resultado numérico.
    - Numba en modo nopython no siempre soporta np.cross para arreglos 1D según
      la versión instalada, mientras que las operaciones escalares de abajo sí
      están garantizadas; esto hace la función robusta independientemente de la
      versión de Numba disponible.
    """
    vx, vy, vz = y[3], y[4], y[5]

    # Velocidad del balón relativa al aire en movimiento
    v_rel_x = vx - v_fluido[0]
    v_rel_y = vy - v_fluido[1]
    v_rel_z = vz - v_fluido[2]
    v_mod = np.sqrt(v_rel_x**2 + v_rel_y**2 + v_rel_z**2)

    # Fuerza de arrastre: opuesta a la velocidad relativa, crece con su cuadrado
    coef_D = -0.5 * rho * C_D * A * v_mod
    FD_x = coef_D * v_rel_x
    FD_y = coef_D * v_rel_y
    FD_z = coef_D * v_rel_z

    # Fuerza de Magnus: proporcional a omega x v_rel (producto cruz expandido)
    coef_M = 0.5 * rho * C_L * A
    FM_x = coef_M * (omega[1]*v_rel_z - omega[2]*v_rel_y)
    FM_y = coef_M * (omega[2]*v_rel_x - omega[0]*v_rel_z)
    FM_z = coef_M * (omega[0]*v_rel_y - omega[1]*v_rel_x)

    dydt = np.empty(6)
    dydt[0] = vx
    dydt[1] = vy
    dydt[2] = vz
    dydt[3] = (FD_x + FM_x) / m
    dydt[4] = (FD_y + FM_y) / m
    dydt[5] = (FD_z + FM_z) / m - g   # la gravedad solo actúa en z
    return dydt


def hit_ground(t, y, *args):
    """Evento para solve_ivp: se dispara cuando z cruza cero hacia abajo.

    No se compila con @njit: SciPy exige que las funciones de evento sean
    funciones Python normales, y de todas formas solo se evalúa unas pocas
    veces por integración (para refinar el instante del cruce), así que
    compilarla no traería ningún beneficio.

    Acepta *args porque, al usar args=(...) en solve_ivp, SciPy le pasa a
    TODAS las funciones asociadas (la EDO y los eventos) los mismos argumentos
    adicionales; hit_ground no los necesita, pero debe poder recibirlos.
    """
    return y[2]


hit_ground.terminal = True   # detener la integración al ocurrir el evento
hit_ground.direction = -1    # solo cuando z decrece (el balón cae, no cuando despega)
```

## Funciones para los barridos de parámetros

Cada barrido (en velocidad angular o en inclinación del eje) recorre varios valores de $\vec\omega$ que son completamente independientes entre sí: la trayectoria para un valor no depende de la trayectoria para otro. Esto es justo lo opuesto a la integración temporal dentro de una misma llamada a `solve_ivp` (que sí es secuencial): aquí se trata de un barrido de parámetros "vergonzosamente paralelo", así que el trabajo se reparte entre hilos con `ThreadPoolExecutor`. Como `ecuaciones_rhs` está compilada con `nogil=True`, los hilos pueden ejecutar tramos de código compilado en paralelo real (sin la limitación del GIL de Python) durante buena parte del tiempo de cómputo.

Además, cada barrido se integra **una sola vez** con `correr_barrido`, y el resultado se reutiliza en las cuatro gráficas correspondientes (trayectoria X-Y, trayectoria X-Z, métrica final vs. parámetro, y altura máxima vs. parámetro), evitando repetir la misma integración por cada gráfica.


```python
def correr_barrido(etiquetas, omegas, v_fluido):
    """Integra la trayectoria del balón para cada vector de velocidad angular
    en `omegas`, bajo un viento fijo `v_fluido`, y devuelve un diccionario
    {etiqueta: solución} que se reutiliza en todas las gráficas de ese barrido.

    `etiquetas` y `omegas` deben tener la misma longitud y estar en
    correspondencia uno a uno (etiquetas[i] identifica a omegas[i]).
    """
    def _resolver(par):
        etiqueta, omega = par
        sol = solve_ivp(
            ecuaciones_rhs, t_span, y0, t_eval=t_eval, events=hit_ground,
            args=(omega, v_fluido, m, A, rho, g, C_D, C_L),
        )
        return etiqueta, sol

    resultados = {}
    with ThreadPoolExecutor() as executor:
        for etiqueta, sol in executor.map(_resolver, zip(etiquetas, omegas)):
            resultados[etiqueta] = sol
    return resultados


def graficar_trayectorias(resultados, indices, xlabel, ylabel, titulo, etiqueta_fmt):
    """Grafica, para cada entrada de un barrido ya calculado, la proyección de
    la trayectoria sobre el plano indicado por `indices` (p. ej. (0, 1) para
    X-Y o (0, 2) para X-Z). Recibe resultados ya integrados (ver
    correr_barrido) en vez de volver a resolver las ecuaciones: cada llamada a
    solve_ivp es costosa, así que conviene pagarla una sola vez y reutilizarla
    en cuantas gráficas hagan falta.
    """
    i, j = indices
    plt.figure(figsize=(7, 5))
    for etiqueta, sol in resultados.items():
        plt.plot(sol.y[i], sol.y[j], label=etiqueta_fmt(etiqueta))
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.title(titulo)
    plt.legend()
    plt.grid()
    plt.show()


def graficar_metrica_final(valores_x, valores_y, xlabel, ylabel, titulo, linea_referencia=None):
    """Grafica una métrica escalar (p. ej. desviación lateral final o altura
    máxima) en función del parámetro barrido.
    """
    plt.figure(figsize=(7, 5))
    plt.plot(valores_x, valores_y, 'o-')
    if linea_referencia is not None:
        plt.axhline(linea_referencia, linestyle='--')
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.title(titulo)
    plt.grid()
    plt.show()
```

## Trayectoria de referencia: tiro con rosca

Antes de los barridos de parámetros, se ilustra la trayectoria para una configuración de tiro con rosca: $|\vec\omega| = 50$ rad/s y una inclinación del eje de rotación $\theta = 20°$, sin viento.


```python
# Configuración de ejemplo: "tiro con rosca"
omega_mag_ejemplo = 50.0             # magnitud de la velocidad angular [rad/s]
theta_ejemplo = np.deg2rad(20.0)     # inclinación del eje de rotación
omega_ejemplo = omega_mag_ejemplo * np.array([0.0, np.sin(theta_ejemplo), np.cos(theta_ejemplo)])
v_fluido_ejemplo = np.array([0.0, 0.0, 0.0])   # sin viento

sol_ejemplo = solve_ivp(
    ecuaciones_rhs, t_span, y0, t_eval=t_eval, events=hit_ground,
    args=(omega_ejemplo, v_fluido_ejemplo, m, A, rho, g, C_D, C_L),
)

fig, ejes = plt.subplots(1, 2, figsize=(11, 4))

ejes[0].plot(sol_ejemplo.y[0], sol_ejemplo.y[1])
ejes[0].set_xlabel("x [m]")
ejes[0].set_ylabel("y [m]")
ejes[0].set_title("Vista superior (X-Y)")
ejes[0].grid()

ejes[1].plot(sol_ejemplo.y[0], sol_ejemplo.y[2])
ejes[1].set_xlabel("x [m]")
ejes[1].set_ylabel("z [m]")
ejes[1].set_title("Vista lateral (X-Z)")
ejes[1].grid()

fig.suptitle("Trayectoria de ejemplo: tiro con rosca (|ω| = 50 rad/s, θ = 20°, sin viento)")
plt.tight_layout()
plt.show()
```

## Análisis paramétrico: sin viento

### Barrido en velocidad angular (sin viento)

```python
# Viento de esta sección
v_fluido = np.array([0.0, 0.0, 0.0])

# Barrido en la magnitud de omega, con inclinación del eje fija
omega_mag_valores = np.array([-30, -20, -10, 0, 10, 20, 30])   # rad/s
theta_fijo = np.deg2rad(15)   # inclinación fija del eje de rotación
direccion_eje = np.array([0.0, np.sin(theta_fijo), np.cos(theta_fijo)])
omegas_barrido = omega_mag_valores[:, None] * direccion_eje

# La etiqueta de cada resultado es la componente z real de omega (no la
# magnitud de entrada), ya que theta_fijo != 0 hace que
# omega_z = omega_mag * cos(theta_fijo).
etiquetas_omega = omegas_barrido[:, 2]

# Una sola integración por valor de omega_z; el resultado se reutiliza en
# las cuatro gráficas siguientes.
resultados_omega_sv = correr_barrido(etiquetas_omega, omegas_barrido, v_fluido)
```

```python
graficar_trayectorias(
    resultados_omega_sv, indices=(0, 1),
    xlabel="x [m]", ylabel="y [m]",
    titulo="Trayectoria X-Y para distintas velocidades angulares (sin viento)",
    etiqueta_fmt=lambda etq: f"ωz = {etq:.0f} rad/s",
)
```

```python
graficar_trayectorias(
    resultados_omega_sv, indices=(0, 2),
    xlabel="x [m]", ylabel="z [m]",
    titulo="Trayectoria X-Z para distintas velocidades angulares (sin viento)",
    etiqueta_fmt=lambda etq: f"ωz = {etq:.0f} rad/s",
)
```

```python
omega_vals = list(resultados_omega_sv.keys())
delta_y_final = [sol.y[1, -1] for sol in resultados_omega_sv.values()]

graficar_metrica_final(
    omega_vals, delta_y_final,
    xlabel="ωz [rad/s]", ylabel="Desviación lateral final Δy [m]",
    titulo="Velocidad angular vs desviación lateral final (sin viento)",
    linea_referencia=0,
)
```

```python
omega_plot = list(resultados_omega_sv.keys())
z_max_plot = [np.max(sol.y[2]) for sol in resultados_omega_sv.values()]

graficar_metrica_final(
    omega_plot, z_max_plot,
    xlabel="ωz [rad/s]", ylabel="Altura máxima z_max [m]",
    titulo="Velocidad angular vs altura máxima alcanzada (sin viento)",
    linea_referencia=6,   # altura máxima de referencia, caso sin rotación (ωz=0)
)
```

## Análisis paramétrico: viento frontal

### Barrido en velocidad angular (viento frontal)

```python
# Viento de esta sección
v_fluido = np.array([-5.0, 0.0, 0.0])

# Barrido en la magnitud de omega, con inclinación del eje fija
omega_mag_valores = np.array([-30, -20, -10, 0, 10, 20, 30])   # rad/s
theta_fijo = np.deg2rad(15)   # inclinación fija del eje de rotación
direccion_eje = np.array([0.0, np.sin(theta_fijo), np.cos(theta_fijo)])
omegas_barrido = omega_mag_valores[:, None] * direccion_eje

# La etiqueta de cada resultado es la componente z real de omega (no la
# magnitud de entrada), ya que theta_fijo != 0 hace que
# omega_z = omega_mag * cos(theta_fijo).
etiquetas_omega = omegas_barrido[:, 2]

# Una sola integración por valor de omega_z; el resultado se reutiliza en
# las cuatro gráficas siguientes.
resultados_omega_vf = correr_barrido(etiquetas_omega, omegas_barrido, v_fluido)
```

## Análisis y conclusiones

Al analizar las gráficas resultantes, se observa que el factor más relevante para la magnitud de la "rosca" de un balón de fútbol es la velocidad angular: variaciones de la inclinación del eje entre −30° y 30° no cambian de forma apreciable la desviación lateral final si la magnitud de $\vec\omega$ se mantiene constante. El efecto principal de la inclinación del eje está, en cambio, en cómo se reparte la fuerza de Magnus entre los ejes vertical y horizontal, lo que sí modifica la altura máxima alcanzada por el balón. Estas observaciones son consistentes con la forma de la fórmula de Magnus utilizada: al inclinar el eje se redistribuyen las componentes de $\vec\omega \times \vec v_{\mathrm{rel}}$ entre $y$ y $z$, sin cambiar su magnitud total de forma significativa en este rango.

En cuanto al viento, los efectos de la fuerza de Magnus sobre la desviación lateral final y la altura máxima aumentan con viento frontal respecto del caso sin viento. Esto también es consistente con la fórmula de Magnus: el viento en contra incrementa la velocidad relativa entre el balón y el aire, con un efecto similar al de aumentar la velocidad angular. De aquí se puede inferir, además, que un viento a favor del balón reduciría el efecto Magnus.

Con viento lateral no se observan aumentos apreciables de la altura máxima en comparación con el caso sin viento; sin embargo, el viento lateral sí afecta la desviación lateral final, reduciéndola cuando el balón se mueve en dirección contraria al viento y aumentándola cuando se mueve a favor. Este resultado no se debe únicamente al efecto Magnus, sino también, en parte, a la fuerza de arrastre.

En resumen: la magnitud de la velocidad angular determina qué tan intenso es el efecto Magnus; la inclinación del eje de rotación determina cómo se reparte esa fuerza entre los ejes vertical y horizontal, decidiendo si el efecto se nota más en la desviación lateral o en la altura máxima; y un viento paralelo a la dirección de vuelo del balón (en $x$) afecta la trayectoria de forma similar a un cambio en la magnitud de la velocidad angular, mientras que un viento lateral introduce un sesgo en la desviación lateral final, en la dirección del viento, sin afectar mayormente la altura máxima.

Para este trabajo se hicieron varias suposiciones que simplifican los cálculos, entre ellas tratar el aire como un fluido incompresible y no turbulento, y despreciar el torque que el aire ejerce sobre la rotación del balón (de ahí que $\vec\omega$ se tome constante). Un trabajo futuro podría relajar estos supuestos, contrastar las trayectorias simuladas con trayectorias reales observadas en partidos, y estudiar el efecto de los parámetros físicos del balón (masa, radio, rugosidad, coeficientes aerodinámicos) sobre los resultados.

**Nota sobre el intervalo de integración:** para ninguna de las combinaciones de parámetros simuladas en este notebook el balón llega efectivamente al suelo ($z=0$) dentro del intervalo `t_span = (0, 1.2)` s; el evento `hit_ground` está definido pero nunca se dispara. Esto significa que, pese a los nombres de las variables (`delta_y_final`, "y al impacto"), lo que en realidad se está comparando en las gráficas es el estado del balón en $t=1.2$ s, no en el momento del aterrizaje. Los resultados y las conclusiones anteriores siguen siendo válidos como comparación entre escenarios (todos se evalúan en el mismo instante), pero conviene tenerlo presente si se quiere interpretar alguna de estas cantidades como una distancia o desviación "de cancha".
