## Marco teórico

### El péndulo doble

En general, un péndulo doble es un sistema compuesto por dos péndulos simples, con el segundo colgando de la masa pendular del primero (Wikipedia, s.f.-g). Como el objetivo es generar un GIF con el movimiento del péndulo, el desarrollo se enfoca únicamente en la cinemática del problema.

Así, este marco teórico se limita a encontrar las expresiones de la posición, la velocidad y la aceleración en términos de las variables del sistema: ángulos iniciales, tiempo, masas y longitudes.

Para plantear el problema se usan coordenadas cartesianas, donde:

* $x$, $y$: posición horizontal y vertical de cada masa.
* $\theta$: ángulo respecto a la vertical.
* $l$: longitud del hilo.

El subíndice 1 se asocia al péndulo superior y el 2 al péndulo inferior. El origen de coordenadas se ubica en el punto pivote del péndulo superior, con el sentido positivo del eje vertical hacia arriba.

Con estas convenciones, la geometría del sistema relaciona las coordenadas $x$, $y$ con los ángulos de la siguiente forma:

$x_1 = l_1 \sin\theta_1$ <br> $y_1 = -l_1 \cos\theta_1$
<br>
<br>
$x_2 = x_1 + l_2 \sin\theta_2$ <br> $y_2 = y_1 - l_2 \cos\theta_2$
<br>
<br>
Derivando cada posición respecto al tiempo se obtienen las velocidades:

\begin{equation}
  \dot{x_1}  = \dot {\theta_1} l_1 \cos \theta _1
\end{equation}

\begin{equation}
  \dot{y_1}  = \dot {\theta_1} l_1 \sin \theta _1
\end{equation}

\begin{equation}
  \dot{x_2}  = \dot {x_1} +\dot {\theta_2} l_2 \cos \theta _2
\end{equation}

\begin{equation}
  \dot{y_2}  = \dot {y_1} +\dot {\theta_2} l_2 \sin \theta _2
\end{equation}

Derivando por segunda vez se obtienen las aceleraciones:

\begin{equation}
  \ddot{x_1}  =-  \dot {\theta_1}^2  l_1 \sin \theta _1 + \ddot{\theta_1} l_1 \cos\theta_1
\end{equation}

\begin{equation}
  \ddot{y_1}  = \dot {\theta_1}^2  l_1 \cos \theta _1 + \ddot{\theta_1} l_1 \sin\theta_1
\end{equation}

\begin{equation}
  \ddot{x_2}  = \ddot {x_1} -\dot {\theta_2}^2 l_2 \sin \theta _2 + \ddot{\theta_2} l_2 \cos\theta_2
\end{equation}

\begin{equation}
  \ddot{y_2}  = \ddot {y_1} +\dot {\theta_2}^2 l_2 \cos \theta _2 + \ddot{\theta_2} l_2 \sin\theta_2
\end{equation}

Aplicando las leyes de Newton se pueden plantear también las ecuaciones de fuerzas del sistema. Sobre la masa $m_1$ actúan:
\begin{equation}
  m_1\ddot{x_1}  = -T_1 \sin \theta _1 + T_2  \sin\theta_2
\end{equation}

\begin{equation}
  m_1\ddot{y_1}  = T_1 \cos\theta_1  -T_2 \cos \theta _2 -  m_{1} g
\end{equation}

y sobre la masa $m_2$ actúan:
\begin{equation}
  m_2\ddot{x_2}  = -T_2 \sin \theta _2
\end{equation}
\begin{equation}
  m_2\ddot{y_2}  = T_2 \cos \theta _2 -  m_{2} g
\end{equation}

Donde:

* $T$ es la tensión en el hilo.
* $g$ es la aceleración gravitatoria.

Con estas ecuaciones newtonianas es muy difícil despejar y resolver las ecuaciones de movimiento, ya que las tensiones $T_1$ y $T_2$ son incógnitas adicionales acopladas de forma no lineal a los ángulos. Debido a esto, se plantea la solución del problema a través del formalismo de Lagrange (Wikipedia, s.f.-e; Wikipedia, s.f.-c), que evita trabajar directamente con las tensiones.


Para aplicar el método de Lagrange se recurre al concepto de energía del sistema: energía potencial debida a la gravedad y energía cinética debida al movimiento de las masas.

### Energía potencial

Situando el nivel de referencia de energía potencial en el punto de suspensión del primer péndulo, se obtiene:

\begin{equation}
V =-m_1 g l_1 \cos \theta_1 - m_2 g (l_1 \cos \theta_1 + l_2 \cos \theta_2)
\end{equation}
\begin{equation}
V = -(m_1+m_2)g l_1 \cos \theta_1 - m_2 g l_2 \cos \theta_2
\end{equation}

### Energía cinética

A partir de las velocidades obtenidas al derivar la posición, y simplificando, la energía cinética es:
\begin{equation}
T = \frac{1}{2} m_1(\dot{x_1}^2 + \dot{y_1}^2) + \frac{1}{2} m_2(\dot{x_2}^2 + \dot{y_2}^2)  
\end{equation}


### Lagrangiano

Con las expresiones anteriores, el movimiento del sistema queda determinado por el **lagrangiano** $L = T - V$:

\begin{equation}
L = \frac{1}{2} m_1(\dot{x_1}^2 + \dot{y_1}^2) + \frac{1}{2} m_2(\dot{x_2}^2 + \dot{y_2}^2) -(m_1+m_2)g l_1 \cos \theta_1 - m_2 g l_2 \cos \theta_2
\end{equation}

Para este péndulo doble, las ecuaciones de Euler-Lagrange (Wikipedia, s.f.-e) toman la forma:

$$\frac{\partial L}{\partial \theta_1} - \frac{d}{dt}\left(\frac{\partial L}{\partial \dot{\theta_1}}\right) = 0$$
<br>
$$\frac{\partial L}{\partial \theta_2} - \frac{d}{dt}\left(\frac{\partial L}{\partial \dot{\theta_2}}\right) = 0$$

Estas ecuaciones se resolverán simbólicamente y luego se integrarán con los métodos numéricos vistos en clase.

## Restricciones de la solución

Las variables $m_1, m_2, \dot\theta_1, \dot\theta_2, l_1, l_2, \theta_1, \theta_2$ son los valores de entrada del sistema. Analizándolas, se tiene que:

* Las masas y las longitudes de los hilos deben ser valores positivos y distintos de cero.
* Todas las variables se manejan como números de punto flotante.
* Los ángulos $\theta_1$ y $\theta_2$ se ingresan en radianes.
* Las velocidades iniciales $\dot\theta_1$, $\dot\theta_2$ corresponden a velocidades **angulares**, en radianes por segundo (rad/s), pues son las componentes del vector de estado $\vec S = (\theta_1, \dot\theta_1, \theta_2, \dot\theta_2)$ que integra `odeint`.
* Las masas se toman en kilogramos y las longitudes en metros, es decir, todas las magnitudes en el Sistema Internacional (S.I.).
* La gravedad se evalúa en su valor escalar positivo; en la simulación se usa $g = 9.81\ \text{m/s}^2$.
* Como condición inicial de referencia, se toma que la masa y la longitud del péndulo superior ($l_1, m_1$) son el doble de las del péndulo inferior ($l_2, m_2$).

Las variables de salida $x_1, x_2, y_1, y_2$ se obtienen como arreglos de valores flotantes, que se usan tanto para las gráficas como para ubicar las masas en cada cuadro de la animación GIF.


## Códigos de solución


**Sección 0**: Importación de librerías


```python
# Librerías científicas y de graficación usadas en todo el notebook.
import numpy as np                      # arreglos y funciones numéricas vectorizadas
import sympy as smp                     # álgebra simbólica: se usa para derivar las ecuaciones de movimiento
from scipy.integrate import odeint      # integrador numérico de EDOs (usa el método LSODA internamente)
import matplotlib.pyplot as plt         # gráfica estática de theta vs t
from matplotlib import animation        # construcción de la animación del péndulo
from IPython.display import display     # mostrar imágenes dentro del notebook
```

**Sección 1**: Definición de Símbolos Simbólicos para las Variables del Sistema


```python
# Símbolos "estáticos" del sistema: no dependen del tiempo, así que basta con
# declararlos como símbolos ordinarios de sympy (no como funciones).
# t: tiempo | g: aceleración gravitatoria | m1, m2: masas | L1, L2: longitudes de los hilos
t, g = smp.symbols('t g')
m1, m2 = smp.symbols('m1 m2')
L1, L2 = smp.symbols('L1, L2')
```

**Sección 2**: Definición de Variables Angulares y sus Derivadas


```python
# A diferencia de t, g, m1, m2, L1, L2, los ángulos SÍ dependen del tiempo, por
# lo que theta1 y theta2 se declaran como funciones de sympy (cls=smp.Function)
# y luego se evalúan en t. Esto permite que smp.diff(..., t) calcule sus
# derivadas simbólicas de forma correcta (usando la regla de la cadena),
# algo que no sería posible si fueran símbolos comunes.
the1, the2 = smp.symbols(r'\theta_1, \theta_2', cls=smp.Function)
the1, the2 = the1(t), the2(t)
the1_d, the2_d = smp.diff(the1, t), smp.diff(the2, t)     # velocidades angulares
the1_dd, the2_dd = smp.diff(the1_d, t), smp.diff(the2_d, t)  # aceleraciones angulares
```

**Sección 3**: Posiciones de las Partículas en Términos de Variables Angulares


```python
# Posiciones (x, y) de cada masa en función de los ángulos, tal como se
# dedujeron por geometría en el marco teórico.
x1 = L1 * smp.sin(the1)
y1 = -L1 * smp.cos(the1)
x2 = L1 * smp.sin(the1) + L2 * smp.sin(the2)
y2 = -L1 * smp.cos(the1) - L2 * smp.cos(the2)
```

**Sección 4**: Energía Cinética, Potencial, y Lagrangiana del Sistema


```python
# Energía cinética T (a partir de las velocidades, obtenidas derivando x, y
# respecto a t) y energía potencial V (proporcional a la altura y de cada
# masa). El lagrangiano L = T - V condensa toda la dinámica del sistema.
T1 = 1/2 * m1 * (smp.diff(x1, t)**2 + smp.diff(y1, t)**2)
T2 = 1/2 * m2 * (smp.diff(x2, t)**2 + smp.diff(y2, t)**2)
T, V = T1 + T2, m1 * g * y1 + m2 * g * y2
L = T - V
```

**Sección 5**: Ecuaciones de Euler-Lagrange


$$\frac{\partial L}{\partial \theta_1} - \frac{d}{dt}\frac{\partial L}{\partial \dot{\theta_1}} = 0$$
$$\frac{\partial L}{\partial \theta_2} - \frac{d}{dt}\frac{\partial L}{\partial \dot{\theta_2}} = 0$$


```python
# Ecuaciones de Euler-Lagrange: cada una debe igualarse a cero (ver celda de
# arriba), así que LE1 y LE2 representan el lado izquierdo de esa igualdad.
# El .simplify() se aplica únicamente al término d/dt(dL/d theta_d) -no a la
# expresión LE completa- porque es el sub-término con productos de senos y
# cosenos anidados que más beneficio obtiene de la simplificación; simplificar
# la expresión completa de una sola vez sería considerablemente más lento para
# una ganancia equivalente.
LE1 = smp.diff(L, the1) - smp.diff(smp.diff(L, the1_d), t).simplify()
LE2 = smp.diff(L, the2) - smp.diff(smp.diff(L, the2_d), t).simplify()
```

**Sección 6**: Resolución Simbólica de las Ecuaciones de Movimiento


```python
# Se resuelve el sistema {LE1 = 0, LE2 = 0} para las aceleraciones angulares
# theta1'' y theta2''. simplify=False y rational=False evitan que sympy
# invierta tiempo adicional en simplificar o racionalizar el resultado del
# solve, ya que estas expresiones se van a "numerizar" con lambdify de todas
# formas y no se necesitan en una forma simbólica particular.
sols = smp.solve([LE1, LE2], (the1_dd, the2_dd), simplify=False, rational=False)
```

Ahora tenemos:

* $\frac{d^2 \theta_1}{dt^2} = ...$
* $\frac{d^2 \theta_2}{dt^2} = ...$

Estas son dos ecuaciones diferenciales ordinarias (EDOs) de segundo orden. En Python solo podemos resolver directamente sistemas de EDOs de primer orden, así que cualquier sistema de segundo orden se reescribe de la siguiente manera:

1. Se define $z_1 = d\theta_1/dt$ y $z_2 = d\theta_2/dt$.
2. Entonces $dz_1/dt = d^2\theta_1/dt^2$ y $dz_2/dt = d^2\theta_2/dt^2$.

Con esto se obtiene un sistema de 4 EDOs de primer orden (en lugar de 2 de segundo orden):

* $d\theta_1/dt = z_1$
* $d z_1/dt = ...$
* $d\theta_2/dt = z_2$
* $d z_2/dt = ...$

Se necesita convertir las expresiones simbólicas anteriores en funciones numéricas para poder usarlas en un solucionador numérico de Python; para esto se utiliza `smp.lambdify`.


**Sección 7**: Conversión de Soluciones en Funciones Numéricas


Las funciones `dz1dt_f` y `dz2dt_f` de abajo son las que evalúa `odeint` en cada paso de la integración: la parte costosa de todo el notebook. Como es una recurrencia temporal -el estado en el paso $i+1$ depende del estado en el paso $i$-, esta función no puede paralelizarse (no hay independencia entre iteraciones). Lo que sí se puede hacer es compilarla con **Numba** (`@njit`) para que se ejecute como código máquina nativo en lugar de bytecode interpretado de Python, ya que se invoca miles de veces durante la integración.

Para que la función sea compatible con el modo *nopython* de Numba, se genera con `modules='math'` en lugar del `'numpy'` por defecto: `dSdt` siempre se llama con números flotantes sueltos (nunca arreglos), por lo que `math.sin`/`math.cos` son exactamente equivalentes en resultado a sus versiones de NumPy, pero sin la sobrecarga de las funciones universales (*ufuncs*) pensadas para arreglos, y son las que Numba compila de forma directa.

Numba no viene garantizado en todos los entornos donde pueda ejecutarse este notebook, así que su importación se protege con `try/except`: si no está instalado, se define un decorador `njit` de respaldo que no hace nada (el código sigue siendo correcto, solo se pierde la aceleración).


```python
# --- Numba con respaldo (fallback) si no está instalado ---
# Si Numba no está disponible, "njit" se redefine como un decorador que
# simplemente devuelve la función sin modificarla, para que el resto del
# notebook funcione igual (más lento) sin necesitar Numba instalado.
try:
    from numba import njit
    NUMBA_DISPONIBLE = True
except ImportError:
    NUMBA_DISPONIBLE = False

    def njit(*args, **kwargs):
        if len(args) == 1 and callable(args[0]) and not kwargs:
            return args[0]

        def _decorador(func):
            return func
        return _decorador

print("Numba disponible:", NUMBA_DISPONIBLE)
```

```python
# Se convierten las soluciones simbólicas de las aceleraciones angulares en
# funciones numéricas. modules='math' (en vez del 'numpy' por defecto) genera
# código que usa math.sin/math.cos sobre escalares, compatible con @njit.
dz1dt_f = smp.lambdify((t, g, m1, m2, L1, L2, the1, the2, the1_d, the2_d), sols[the1_dd], modules='math')
dz2dt_f = smp.lambdify((t, g, m1, m2, L1, L2, the1, the2, the1_d, the2_d), sols[the2_dd], modules='math')

# Se compilan con Numba (o quedan sin cambios si Numba no está disponible).
# No se usa prange/paralelización: son funciones puramente escalares que se
# llaman de forma secuencial dentro de la integración, no un barrido sobre un
# arreglo de puntos independientes.
dz1dt_f = njit(dz1dt_f)
dz2dt_f = njit(dz2dt_f)
```

Se define $\vec{S} = (\theta_1, z_1, \theta_2, z_2)$. Para usar un solucionador de EDOs en Python se necesita una función que tome $\vec{S}$ y $t$, y devuelva $d\vec{S}/dt$; es decir, hay que definir $d\vec{S}/dt(\vec{S}, t)$.

* El sistema de EDOs queda completamente especificado por $d\vec{S}/dt$, que depende únicamente de $\vec{S}$ y $t$ (además de los parámetros físicos $g, m_1, m_2, L_1, L_2$).


**Sección 8**: Definición de Ecuaciones Diferenciales para Simulación Numérica


```python
def dSdt(S, t, g, m1, m2, L1, L2):
    """
    Lado derecho del sistema de EDOs de primer orden, d(vec S)/dt.

    Recibe el estado actual S = (theta1, z1, theta2, z2) y el tiempo t, y
    devuelve sus derivadas (z1, dz1/dt, z2, dz2/dt). odeint llama a esta
    función internamente en cada paso de integración, así que es la parte
    del notebook que más veces se evalúa.

    Nota de diseño: dtheta1/dt = z1 y dtheta2/dt = z2 son, por la propia
    definición de z1 y z2 (ver celda anterior), sustituciones triviales, así
    que se usan z1 y z2 directamente en lugar de pasar por una llamada de
    función adicional para cada una.
    """
    the1, z1, the2, z2 = S
    return [
        z1,
        dz1dt_f(t, g, m1, m2, L1, L2, the1, the2, z1, z2),
        z2,
        dz2dt_f(t, g, m1, m2, L1, L2, the1, the2, z1, z2),
    ]
```

**Sección 9**: Parámetros para la Simulación Numérica


```python
# t_values: instantes de tiempo (en segundos) donde se desea la solución.
# g, m1, m2, L1, L2: parámetros físicos del sistema, en unidades del S.I.
# (ver "Restricciones de la solución": L1 = 2*L2, m1 = 2*m2).
t_values = np.linspace(0, 40, 1001)
g_value, m1_value, m2_value, L1_value, L2_value = 9.81, 2, 1, 2, 1
```

```python
# Se calcula cuántos puntos de t_values caen en el primer segundo de
# simulación, para poder exportar el GIF a esa misma cantidad de cuadros por
# segundo (fps) y que la animación corra en tiempo real 1:1 respecto a la
# simulación física.
fps_realtime = len(t_values[t_values < 1])
fps_realtime
```

**Sección 10**: Resolución Numérica de las Ecuaciones Diferenciales y Extracción de Soluciones

Con las siguientes condiciones iniciales:

$\theta_1 = 1$ rad<br>
$\theta_2 = -1$ rad<br>
$\dot\theta_1 = -3$ rad/s<br>
$\dot\theta_2 = 5$ rad/s


```python
# odeint integra dSdt usando el método LSODA (Petzold, 1983; Hindmarsh, 1983),
# que alterna internamente entre un esquema explícito y uno implícito según la
# rigidez ("stiffness") local del sistema. initial_conditions sigue el orden
# de S = (theta1, z1, theta2, z2) definido más arriba.
initial_conditions = [1, -3, -1, 5]  # S0 = (theta1, dtheta1/dt, theta2, dtheta2/dt)
solution = odeint(dSdt, y0=initial_conditions, t=t_values, args=(g_value, m1_value, m2_value, L1_value, L2_value))
```

## Resultados
**Sección 11**: Cálculo de Posiciones y Visualización Animada del Sistema


```python
# Posiciones cartesianas de cada masa a partir de los ángulos ya integrados.
# Esto es una operación vectorizada de NumPy sobre todo el arreglo de tiempo a
# la vez (no un bucle en Python), así que ya se ejecuta de forma eficiente sin
# necesitar Numba ni paralelización adicional.
theta1_values, theta2_values = solution.T[0], solution.T[2]
x1_values = L1_value * np.sin(theta1_values)
y1_values = -L1_value * np.cos(theta1_values)
x2_values = L1_value * np.sin(theta1_values) + L2_value * np.sin(theta2_values)
y2_values = -L1_value * np.cos(theta1_values) - L2_value * np.cos(theta2_values)
```

### Gráfica de los ángulos en función del tiempo


```python
plt.title("Grafica t vs θ en un Péndulo Doble")
plt.plot(t_values, theta2_values, label="Posición $θ_2$")
plt.plot(t_values, theta1_values, label="Posición $θ_1$")
plt.xlabel('$t$')
plt.ylabel('$\\theta$')
plt.legend()
plt.grid()
plt.show()
```

### Animación del péndulo doble


```python
# 1. Configuración visual
fig, ax = plt.subplots(1, 1, figsize=(8, 8), facecolor='black')
ax.set_facecolor('black')
ax.get_xaxis().set_ticks([])
ax.get_yaxis().set_ticks([])
for spine in ax.spines.values():
    spine.set_visible(False)

# 2. Objetos gráficos (usamos blit=True para velocidad)
line, = ax.plot([], [], 'w-', lw=2, alpha=0.7)
mass1, = ax.plot([], [], 'o', color='cyan', markersize=12, animated=True)
mass2, = ax.plot([], [], 'o', color='magenta', markersize=10, animated=True)

ax.set_ylim(-4, 4)
ax.set_xlim(-4, 4)

param_text = (
    f"Simulación de Péndulo Doble\n"
    f"$m_1$ (Cian): {m1_value} kg, $L_1$: {L1_value} m\n"
    f"$m_2$ (Magenta): {m2_value} kg, $L_2$: {L2_value} m\n"
    f"$g$: {g_value} $m/s^2$"
)
ax.text(0.05, 0.95, param_text, transform=ax.transAxes, color='white',
        fontsize=10, verticalalignment='top', bbox=dict(boxstyle='round', facecolor='black', alpha=0.6, edgecolor='gray'))

def init():
    """Inicialización para blitting."""
    line.set_data([], [])
    mass1.set_data([], [])
    mass2.set_data([], [])
    return line, mass1, mass2

def animate(i):
    """Actualización de posiciones."""
    line.set_data([0, x1_values[i], x2_values[i]], [0, y1_values[i], y2_values[i]])
    mass1.set_data([x1_values[i]], [y1_values[i]])
    mass2.set_data([x2_values[i]], [y2_values[i]])
    return line, mass1, mass2

# 3. Sincronización de velocidad
# El intervalo debe ser 1000ms / fps para que la velocidad sea 1:1 física
ms_per_frame = 1000 / fps_realtime

ani = animation.FuncAnimation(fig, animate, init_func=init,
                               frames=len(t_values),
                               interval=ms_per_frame,
                               blit=True)

# 4. Guardado del GIF
ani.save('pendulum_animation.gif', writer='pillow', fps=fps_realtime)
```

## Conclusiones

* Ciertos sistemas cinemáticos son más sencillos de resolver a través de los métodos de Lagrange que por medio de las ecuaciones de Newton, en particular cuando existen fuerzas de restricción (como las tensiones de los hilos) que no interesa calcular explícitamente.

* La caracterización del péndulo doble desde la formulación lagrangiana exige asignar coordenadas propias de la geometría del sistema. El concepto de grado de libertad permite asociar de forma intuitiva las coordenadas generalizadas ($\theta_1$, $\theta_2$) con la configuración y el movimiento del sistema.

* Con las condiciones iniciales usadas en la simulación ($\theta_1=1$ rad, $\theta_2=-1$ rad, $\dot\theta_1=-3$ rad/s, $\dot\theta_2=5$ rad/s), la gráfica $t$ vs $\theta$ muestra que $\theta_1$ oscila en un rango acotado, aproximadamente entre $-2.6$ y $2.6$ rad, mientras que $\theta_2$ crece de forma sostenida y supera varias vueltas completas ($2\pi$) a lo largo de la simulación. Esto indica que, para estos parámetros, el péndulo inferior no oscila alrededor de un punto de equilibrio sino que gira, un comportamiento característico del régimen no lineal y sensible a condiciones iniciales del péndulo doble, y distinto del movimiento aproximadamente armónico de un péndulo simple de pequeña amplitud.

* Los métodos computacionales y numéricos simplifican los cálculos y facilitan el modelamiento de fenómenos físicos que, de otra forma, serían muy difíciles de resolver de forma puramente analítica.

* La energía potencial del sistema queda descrita por tres términos: dos corresponden al trabajo de cada masa al variar su propia coordenada generalizada, y el tercero da cuenta del acople entre las masas, ya que el trabajo asociado a $m_2$ depende también de la variación del ángulo $\theta_1$ de $m_1$.

* El lagrangiano, junto con las ecuaciones de Euler-Lagrange, condensa toda la información necesaria para determinar la evolución del sistema, permitiendo predecir el comportamiento futuro de las masas a partir únicamente de las condiciones iniciales y los parámetros físicos.

* La formulación lagrangiana muestra que, en determinadas condiciones (por ejemplo, para amplitudes pequeñas en ambos ángulos), el péndulo doble plano puede aproximar comportamientos similares a los del péndulo simple; sin embargo, como se observó en la simulación, para amplitudes moderadas a grandes el sistema se aparta rápidamente de ese régimen.
