## Física Experimental IV

### Objetivo general

Este experimento estudia la relación entre la intensidad de la luz incidente y el voltaje generado en un circuito fotoeléctrico, evaluando el efecto de la distancia entre el láser y el dispositivo, así como del color (frecuencia) de la luz empleada.

### Objetivos específicos

- Explorar el efecto fotoeléctrico y su dependencia con la intensidad de la luz incidente.
- Analizar cómo la distancia entre el láser y el dispositivo fotoeléctrico afecta el voltaje medido.
- Contrastar los resultados experimentales con las predicciones de la teoría del efecto fotoeléctrico.

### Marco teórico

El efecto fotoeléctrico consiste en la emisión de electrones desde una superficie metálica al ser iluminada con luz de frecuencia suficientemente alta. Einstein explicó este fenómeno postulando que la luz se comporta como un conjunto de paquetes discretos de energía, los fotones, cuya energía es proporcional a la frecuencia de la radiación:

$$
E = h\nu
$$

donde $h$ es la constante de Planck y $\nu$ es la frecuencia de la luz. Cuando un fotón incide sobre el material y transfiere su energía a un electrón, este puede escapar del material si la energía del fotón supera la función de trabajo $\phi$; el exceso de energía se manifiesta como energía cinética del electrón emitido. En el circuito del dispositivo fotoeléctrico, esta energía se traduce en un voltaje medible, relacionado con la frecuencia de la luz y la función de trabajo mediante

$$
eV = h\nu - \phi,
$$

donde $e$ es la carga elemental y $V$ es el voltaje generado. La intensidad de la luz, por su parte, determina el número de fotones incidentes por unidad de tiempo y, en consecuencia, la corriente de electrones emitidos, sin alterar la energía individual de cada fotón.

### Montaje experimental

El montaje consta de un dispositivo fotoeléctrico sobre el cual incide un láser de color e intensidad ajustables (rojo, verde o azul), y un voltímetro que registra el voltaje generado en el circuito. La corriente del dispositivo se varía en pasos de 5 nA, desde 20 nA hasta 0 nA, y las mediciones se repiten a dos distancias entre el láser y el dispositivo: 26.0(1) cm y 10.0(1) cm.


### Procedimiento experimental

1. **Configuración inicial:** se ajusta el láser al color deseado (rojo, verde o azul) y se fija la corriente inicial del dispositivo en 20 nA, con el láser ubicado a 26 cm del dispositivo.
2. **Medición de voltaje:** se enciende el láser y se registra el voltaje generado; luego se reduce la corriente en pasos de 5 nA, registrando el voltaje en cada paso hasta llegar a 0 nA.
3. **Variación de la distancia:** se repite el procedimiento anterior con el láser ubicado a 10 cm del dispositivo, registrando los voltajes correspondientes a cada corriente y distancia.


## Datos

Las corrientes se ajustan manualmente en el dispositivo fotoeléctrico y se expresan en nanoamperios (nA); su incertidumbre corresponde a la resolución del control de corriente. Los voltajes se miden con un multímetro digital y se expresan en milivoltios (mV) para cada color de láser (rojo, verde, azul) y cada distancia entre el láser y el dispositivo (26.0(1) cm y 10.0(1) cm).


```python
"""
Datos experimentales del efecto fotoeléctrico.
"""
import numpy as np
import matplotlib.pyplot as plt

# Corriente ajustada en el dispositivo, con incertidumbre de +/- 1 nA
corriente_nA = np.array([20, 15, 10, 5, 0])
error_corriente_nA = 1

# Voltaje medido (mV) a 26.0(1) cm de distancia
voltaje_rojo_26cm = np.array([149.8, 184.7, 226.5, 272.5, 377.8])
voltaje_verde_26cm = np.array([571.8, 586, 611, 649, 703])
voltaje_azul_26cm = np.array([781, 789, 818, 841, 871])

# Voltaje medido (mV) a 10.0(1) cm de distancia
voltaje_rojo_10cm = np.array([228.3, 245.2, 266.8, 299.8, 375.9])
voltaje_verde_10cm = np.array([626, 640, 655, 679, 709])
voltaje_azul_10cm = np.array([817, 824, 835, 847, 860])
```

## Estimación de incertidumbres en el voltaje

La incertidumbre en el voltaje se calcula siguiendo la convención de calibración registrada en el laboratorio para el multímetro digital utilizado: cuando la lectura se muestra como un número entero de milivoltios, la incertidumbre es de $0.1\text{ mV} + 0.5\%$ de la lectura; cuando la lectura incluye una cifra decimal, la incertidumbre es de $0.3\text{ mV} + 0.8\%$ de la lectura.


```python
def calcular_error_voltaje(voltajes):
    """
    Calcula la incertidumbre absoluta de una serie de voltajes medidos.

    La incertidumbre depende de si la lectura del multímetro se registró
    como un número entero de milivoltios o con una cifra decimal, según la
    convención de calibración usada en el laboratorio:
      - Lectura entera:  error = 0.1 mV + 0.5 % de la lectura.
      - Lectura decimal: error = 0.3 mV + 0.8 % de la lectura.

    Parameters
    ----------
    voltajes : array_like
        Voltajes medidos, en milivoltios.

    Returns
    -------
    numpy.ndarray
        Incertidumbre absoluta de cada voltaje, en milivoltios.
    """
    voltajes = np.asarray(voltajes, dtype=float)
    es_entero = voltajes == np.round(voltajes)
    return np.where(es_entero, 0.1 + 0.005 * voltajes, 0.3 + 0.008 * voltajes)
```

```python
# Incertidumbre asociada a cada serie de voltajes
error_rojo_26cm = calcular_error_voltaje(voltaje_rojo_26cm)
error_verde_26cm = calcular_error_voltaje(voltaje_verde_26cm)
error_azul_26cm = calcular_error_voltaje(voltaje_azul_26cm)

error_rojo_10cm = calcular_error_voltaje(voltaje_rojo_10cm)
error_verde_10cm = calcular_error_voltaje(voltaje_verde_10cm)
error_azul_10cm = calcular_error_voltaje(voltaje_azul_10cm)
```

## Visualización de resultados

La siguiente gráfica muestra el voltaje generado en función de la corriente ajustada, para cada color de láser y cada distancia, incluyendo las barras de error correspondientes.


```python
fig, ax = plt.subplots(figsize=(10, 6))

series = {
    "Rojo 26 cm": (voltaje_rojo_26cm, error_rojo_26cm, "ro-"),
    "Verde 26 cm": (voltaje_verde_26cm, error_verde_26cm, "go-"),
    "Azul 26 cm": (voltaje_azul_26cm, error_azul_26cm, "bo-"),
    "Rojo 10 cm": (voltaje_rojo_10cm, error_rojo_10cm, "r*--"),
    "Verde 10 cm": (voltaje_verde_10cm, error_verde_10cm, "g*--"),
    "Azul 10 cm": (voltaje_azul_10cm, error_azul_10cm, "b*--"),
}

# Se grafica cada serie con su propio estilo, reutilizando el mismo bucle
# para evitar repetir seis llamadas casi idénticas a errorbar.
for etiqueta, (voltaje, error, estilo) in series.items():
    ax.errorbar(
        corriente_nA, voltaje, yerr=error, xerr=error_corriente_nA,
        fmt=estilo, capsize=3, label=etiqueta,
    )

ax.set_xlabel("Corriente (nA)")
ax.set_ylabel("Voltaje (mV)")
ax.set_title("Voltaje vs. corriente a diferentes distancias y colores de láser")
ax.legend()
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## Análisis de los resultados

Los resultados muestran que, para una misma corriente ajustada, el voltaje medido es mayor cuando el láser se encuentra más cerca del dispositivo, lo que refleja una mayor intensidad de luz incidente sobre el material. Asimismo, para una distancia y corriente fijas, el voltaje aumenta con la frecuencia del láser, siendo mayor para el azul que para el verde, y mayor para el verde que para el rojo; esta tendencia es consistente con la relación $eV = h\nu - \phi$, según la cual una mayor frecuencia de la luz incidente produce electrones con mayor energía cinética y, por lo tanto, un mayor voltaje en el circuito. Las barras de error, calculadas a partir de la resolución del multímetro, son en general pequeñas en comparación con las diferencias observadas entre colores y distancias, lo que respalda la significancia física de las tendencias identificadas.


## Conclusiones

1. La disminución de la distancia entre el láser y el dispositivo incrementa la intensidad de la luz incidente sobre el material, lo que se traduce en voltajes mayores para una misma corriente ajustada.
2. Los voltajes medidos aumentan con la frecuencia de la luz, en concordancia con la teoría del efecto fotoeléctrico, según la cual la energía de los fotones —y por tanto la energía cinética de los electrones emitidos— crece con la frecuencia de la radiación incidente.
3. La calibración cuidadosa de la corriente y de la distancia entre el láser y el dispositivo resulta esencial para obtener mediciones de voltaje precisas y reproducibles.

En conjunto, el experimento permite verificar cualitativamente las predicciones del efecto fotoeléctrico: la frecuencia de la luz determina la energía de los electrones emitidos, mientras que la intensidad —modulada aquí mediante la distancia— determina la magnitud de la señal eléctrica generada, sin alterar la relación entre voltaje y frecuencia.
