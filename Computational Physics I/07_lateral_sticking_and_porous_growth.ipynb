## 1. Introducción y objetivos

Los modelos M1 a M6 describen la película con una altura por columna: es la aproximación *solid-on-solid* (SOS), en la que la película es compacta, sin huecos ni voladizos. Con sombreado (cuadernos 05 y 06), esa aproximación tiene una consecuencia que no es física. Un átomo que alcanza la pared lateral de una columna se coloca sobre ella, de modo que la columna crece en altura con el flujo que intercepta por los lados, cada vez mayor, y la superficie degenera en agujas aisladas al cabo de unas pocas decenas de capas, o de unas pocas capas con gas. En una película real, el átomo queda adherido a la pared en el punto de contacto: la columna se ensancha en lugar de crecer en altura y deja huecos debajo. Es el origen de la microestructura columnar y porosa de las películas evaporadas a baja temperatura (Dirks & Leamy, 1977; Thornton, 1977).

Este cuaderno añade al modelo M6 la **adhesión lateral**. Cuando un átomo alcanza la pared de una columna, queda apoyado en la columna vecina a la altura del contacto y se relaja desde allí con la regla común; si no salta a un vecino más bajo, queda adherido y las posiciones vacías bajo él son huecos de la película. Todo lo demás es lo del cuaderno 06. Sin adhesión lateral, el modelo M7 se reduce exactamente al M6. Es el último modelo de la cadena: tiene todos los mecanismos de la serie.

**Objetivos**

* Seguir los modelos M6 y M7 hasta 256 capas y comprobar si la adhesión lateral suprime la inestabilidad de columnas que la aproximación SOS produce con sombreado.
* Medir la porosidad de la película y su evolución con el espesor, y describir su estructura vertical y sus columnas con el registro de la ocupación de cada posición.
* Construir el mapa de la porosidad en el plano probabilidad de salto × camino libre medio, y su dependencia de la dispersión angular de la emisión.
* Medir el efecto del tamaño de la malla y del número máximo de saltos.
* Seguir M7 hasta 4096 capas en mallas de 16 a 200 celdas de lado, contrastar el crecimiento y la saturación de la rugosidad con el escalado de la clase de Kardar, Parisi y Zhang y medir la separación de los montículos de la superficie.

## 2. Marco teórico y modelo físico

### 2.1 Deposición balística

En la deposición balística, cada partícula cae en vertical sobre una columna elegida al azar y se queda en la primera posición en la que toca la película, ya sea la cima de su columna o la pared de una vecina más alta (Vold, 1959). Como una partícula puede quedar adherida a una pared por encima de la cima de su columna, la película tiene huecos y su densidad es menor que uno. En una malla, la regla se escribe con la altura de cada columna, $h_i \leftarrow \max(h_i + 1,\ h_{i-1},\ h_{i+1})$ en una dimensión, y los huecos son las posiciones vacías bajo la cima de cada columna. La superficie es rugosa a todas las escalas y pertenece a la clase de Kardar, Parisi y Zhang (Meakin et al., 1986).

Con incidencia oblicua, la película crece en columnas separadas por huecos que se inclinan hacia la fuente, y su densidad disminuye al aumentar el ángulo de incidencia (Krug & Meakin, 1989). Es lo que se observa en películas evaporadas: estructura columnar, más porosa cuanto más oblicua es la incidencia (Dirks & Leamy, 1977) y cuanto menor es la movilidad de los átomos en la superficie, es decir, cuanto menor es la temperatura del sustrato frente a la de fusión (zona 1 de Thornton, 1977). El gas de la cámara añade direcciones oblicuas al flujo (cuaderno 06), de modo que una presión mayor también debe dar películas más porosas.

### 2.2 Adhesión lateral en el modelo M7

El trazador de rayos recorre el vuelo de cada átomo celda a celda (cuaderno 05). Si el átomo cruza el borde entre la celda de una columna A y la de una columna vecina B a una altura $z_{\mathrm{e}}$ menor que la de B, ha alcanzado la pared lateral de B. El contacto ocurre al nivel $n = \lfloor z_{\mathrm{e}} \rfloor$, que es una posición libre de A (el átomo volaba por encima de su cima) contigua a una posición ocupada de B: $h_A \le n < h_B$. En los modelos SOS, el átomo se coloca sobre B y se relaja desde allí. En el modelo M7:

1. El átomo queda apoyado en A, al nivel $n$, y se relaja desde allí con la regla común de la serie, como si A tuviera altura $n$: con probabilidad $P = \min(1, T/T_{\mathrm{escala}})$ salta a un vecino de A cuya altura sea menor que $n$, hasta cinco veces.
2. Si salta, se deposita sobre el vecino como cualquier otro átomo, y A no cambia.
3. Si no salta, queda adherido a la pared: ocupa la posición $n$ de A, la altura de A pasa a ser $n + 1$ y las posiciones de A entre su altura anterior y $n$ quedan vacías.

Si el átomo llega desde arriba a la cima de B, no hay contacto lateral y todo es como en el modelo M6. Con incidencia vertical y sin gas, ningún átomo alcanza una pared, de modo que M7 se reduce a M3; sin la regla de adhesión, se reduce exactamente a M6.

**Película con huecos.** La altura $h_i$ de cada columna es ahora la de su posición ocupada más alta más uno, y las posiciones vacías por debajo son huecos. El trazador trata cada columna como sólida hasta su cima, de modo que ningún vuelo posterior alcanza un hueco: los huecos son cerrados y permanentes. Es una aproximación, porque una trayectoria muy oblicua podría entrar bajo un voladizo; esa región está, en cualquier caso, en la sombra de la columna que la cubre. Con $N$ átomos depositados y $V$ huecos, $\sum_i h_i = N + V$.

### 2.3 Observables de la película porosa

* **Espesor de material** $\overline{h} = N/L^2$, el número de capas depositadas. Es la variable de crecimiento, la misma de toda la serie.
* **Altura media de las columnas** $\langle h \rangle = (N + V)/L^2$, la del espesor aparente de la película. La rugosidad $W$ y el relieve $R = h_{\max} - \langle h \rangle$ se miden con las alturas de las columnas.
* **Porosidad** $p = V/(N + V)$, la fracción de huecos en el volumen bajo la cima de las columnas. La **porosidad incremental** entre dos espesores, $\Delta V/(\Delta N + \Delta V)$, es la de la capa depositada entre ellos: si la estructura de la película alcanza un régimen estacionario, deja de depender del espesor.
* **Perfil de densidad.** Con la ocupación de cada posición se obtienen, en cada nivel $z$, la fracción de posiciones ocupadas $\rho(z)$, la fracción de columnas cuya cima está por encima de $z$ y la porosidad dentro de esa envolvente. El perfil separa la película densa cerca del sustrato, el interior y la región de la superficie.

### 2.4 Inestabilidad de columnas con adhesión lateral

En los modelos SOS, el flujo que una columna intercepta por los lados se suma a su altura, y la columna crece más deprisa cuanto más sobresale (cuaderno 05). Con adhesión lateral, ese flujo ensancha la columna y su altura solo crece con los átomos que llegan a su cima. El mecanismo que lleva a las agujas desaparece, aunque las columnas anchas siguen haciendo sombra y compitiendo entre ellas. Por eso se espera que el relieve de M7 se mantenga acotado hasta espesores grandes, mientras que el de M6 llega al corte de 100 capas a las pocas capas.

### 2.5 Escalado con el tamaño de la malla y montículos

En una malla de lado $L$, la rugosidad de una superficie autoafín sigue la forma de escalado de Family y Vicsek (1985),

$$W(L, \overline{h}) = L^{\alpha} f\!\left(\overline{h}/L^{z}\right),$$

con $f(u) \propto u^{\beta}$ para $u \ll 1$ y $f(u)$ constante para $u \gg 1$: la rugosidad crece como $\overline{h}^{\beta}$ hasta un espesor de cruce $\overline{h}_{\times} \propto L^{z}$ y después se satura en $W_{\mathrm{sat}} \propto L^{\alpha}$, con $z = \alpha/\beta$. La deposición balística pertenece a la clase de universalidad de la ecuación de Kardar, Parisi y Zhang (KPZ; Kardar et al., 1986; Meakin et al., 1986). En 2+1 dimensiones sus exponentes solo se conocen por simulación: $\beta = 0.2415 \pm 0.0015$ (Kelling & Ódor, 2011) y $\alpha = 0.3869 \pm 0.0004$ (Pagnani & Parisi, 2015), con $z = 2 - \alpha \approx 1.61$ por la invariancia galileana de la ecuación. En la clase de Edwards y Wilkinson, la de las películas relajadas de los cuadernos 02 a 04, la rugosidad crece y se satura de forma logarítmica, y $z = 2$ (cuaderno 03).

El sombreado es una interacción no local: una región alta recibe flujo a costa de las vecinas, y un valle recibe menos cuanto más profundo es. En un modelo de sombreado con crecimiento lateral en 2+1 dimensiones, la rugosidad crece como $\overline{h}$ ($\beta = 1$) y se forman columnas altas (Drotar et al., 2000). En otros casos, la competencia produce montículos con una separación característica que crece con el espesor como una potencia, y entonces la superficie deja de ser autoafín (Pelliccione et al., 2006). En ese caso la rugosidad de saturación no tiene por qué seguir $L^{\alpha}$ con el exponente de KPZ: depende de cuántos montículos caben en la malla. Las dos descripciones se distinguen con el espectro de potencia de las alturas, $S(k)$, promediado en anillos de número de onda $k = n/L$, con $n = 1, \dots, L/2$. En una superficie autoafín, $S(k)$ es constante por debajo de la inversa de la longitud de correlación y decrece por encima, sin un máximo intermedio; en una superficie de montículos, tiene un máximo en el número de onda $k^{*}$ de su separación, $\Lambda = 1/k^{*}$, que se desplaza hacia valores menores al crecer la película. Cuando $\Lambda$ alcanza el lado de la malla, el máximo queda en $n = 1$ y la separación ya no puede medirse.

### 2.6 Hipótesis y criterios

Los criterios se fijan antes de ejecutar los experimentos:

1. **Contención del modelo M6.** Sin la regla de adhesión, el modelo M7 da exactamente las superficies de M6.
2. **Geometría del contacto lateral.** Sobre una superficie plana con un muro de 30 capas, todos los rayos que entran en la celda del muro por debajo de su cima informan de un contacto lateral desde la columna contigua, al nivel $\lfloor z_{\mathrm{e}} \rfloor$ de la altura de entrada, y ninguno de los que entran por encima de la cima informa de contacto lateral.
3. **Registro del volumen.** La película registrada con su ocupación es la de la primera réplica de M7, posición por posición: sus columnas tienen las mismas alturas, el número de posiciones ocupadas es el de átomos y el de posiciones vacías bajo las cimas es el de huecos.
4. **Supresión de la inestabilidad.** Ninguna réplica de M7 llega al corte por relieve antes de 256 capas.
5. **Régimen estacionario de la porosidad.** La porosidad incremental de M7 entre 64 y 256 capas difiere menos del $10\,\%$ de la porosidad incremental entre 16 y 64 capas.
6. **Mapa.** A 64 capas, a cada camino libre medio, la porosidad no aumenta al aumentar la probabilidad de salto, y a cada probabilidad de salto no disminuye al disminuir el camino libre medio: ninguna diferencia entre celdas vecinas va en sentido contrario en más de dos errores estándar.
7. **Dispersión angular.** Sin gas, a cada probabilidad de salto, la porosidad a 64 capas es nula con incidencia vertical y no disminuye al aumentar la dispersión angular de la emisión, con la misma tolerancia.
8. **Tamaño de la malla.** La rugosidad a 4, 16 y 64 capas y la porosidad a 64 capas en la malla de 200 celdas difieren menos del $1\,\%$ de las de 400 (Welch, $p > 0.01$).
9. **Número máximo de saltos.** Con 10 saltos máximos en lugar de 5, la rugosidad a 16 y 64 capas y la porosidad a 64 capas cambian menos del $1\,\%$ (prueba $t$ pareada, $p > 0.01$).
10. **Saturación en las mallas pequeñas.** En las mallas de 32 y 64 celdas, la rugosidad cuadrática media de M7 en la segunda mitad de la ventana de 2048 a 4096 capas no difiere de la de la primera mitad (prueba $t$ pareada por réplica, $p > 0.01$).
11. **Clase de Kardar, Parisi y Zhang.** En la malla de 200 celdas, el exponente de crecimiento entre 1024 y 4096 capas es compatible con $\beta = 0.2415$, y el exponente de rugosidad que dan las rugosidades de saturación de las dos mallas del criterio 10, si las dos están saturadas, es compatible con $\alpha = 0.3869$. En los dos casos, $|z| \le 3$ con el error estándar de la estimación y el del valor de referencia.
12. **Montículos.** En la malla de 200 celdas, el máximo del espectro de potencia está fuera de $n = 1$ a 256 capas, y la separación $\Lambda$ a 2048 capas supera la de 256 capas en más de dos errores estándar.

## 5. Funciones del modelo

La adhesión lateral está en dos funciones del motor común. El trazador (`motor.trazar_vuelo`) informa del contacto con la pared de una columna: la columna desde la que llega el átomo y el nivel del contacto. El lazo de deposición (`motor.ejecutar_motor_volumen`) aplica la regla de la sección 2.2 cuando la bandera `enable_lateral` está activa, cuenta los huecos y, si se le pasa un arreglo de ocupación, marca la posición de cada átomo depositado. La celda siguiente muestra el código del lazo.

El análisis usa, además de las funciones de los cuadernos 05 y 06, `simular_volumen`, que repite una réplica registrando la ocupación, `perfil_de_densidad`, que obtiene el perfil vertical de la película, y `ajuste_por_replica`, que ajusta una recta a cada réplica (módulo `crecimiento`). En el cuaderno se definen `espectro_por_anillos`, que calcula el espectro de potencia de las alturas en los anillos de número de onda entero de la sección 2.5, y `maximo_del_espectro`, que localiza su máximo y estima su error remuestreando las réplicas.

```python
ANCHO = 78
fila = crear_fila(48)


def parametros(L=NX, **cambios):
    """Parámetros del sistema en una malla de L x L celdas, con los cambios indicados."""
    return dataclasses.replace(PARAMS_BASE, nx=L, ny=L, **cambios)


def simular(clave, L, n_replicas, espesor_final, espesores_instantaneas=(), nombre=None, banderas=None, **opciones):
    """Ejecuta un modelo de la cadena en una malla de L x L hasta ``espesor_final`` capas.

    En la malla base el nombre es el del modelo, de modo que las réplicas son las de toda la serie; en otras
    mallas el nombre incluye el tamaño. ``banderas`` sustituye las del modelo y ``opciones`` pasa a
    ``simular_modelo`` (instantes de registro) o, si es un parámetro del sistema, a ``parametros``.
    """
    campos = {f.name for f in dataclasses.fields(PARAMS_BASE)}
    cambios = {k: opciones.pop(k) for k in list(opciones) if k in campos}
    if nombre is None:
        nombre = CLAVES[clave] if L == NX else f"{CLAVES[clave]}, L = {L}"
    return simular_modelo(parametros(L, **cambios), nombre, banderas or MODELOS[CLAVES[clave]], n_replicas, SEMILLA,
                          espesor_final, espesores_instantaneas, max_pasos_rayo=MAX_PASOS_RAYO, **opciones)


def simular_sombreado(clave, L, n_replicas, espesor_final, espesores_instantaneas=(), **opciones):
    """Simula un modelo; con sombreado, aplica el corte por relieve."""
    banderas = opciones.get("banderas") or MODELOS[CLAVES[clave]]
    if banderas["enable_shadowing"]:
        opciones.setdefault("relieve_maximo", RELIEVE_MAXIMO)
    return simular(clave, L, n_replicas, espesor_final, espesores_instantaneas, **opciones)


def p_welch(a, b):
    """Valor p de Welch, o nan si alguna muestra tiene menos de dos valores."""
    a, b = np.asarray(a, dtype=float), np.asarray(b, dtype=float)
    return contraste_welch(a, b)["p"] if a.size > 1 and b.size > 1 else np.nan


def rugosidad_en(obs, modelo, objetivo, columna="rugosidad"):
    """Valores de una columna de los observables en un espesor guardado, por réplica."""
    return obs[(obs["modelo"] == modelo) & (obs["objetivo"] == objetivo)].set_index("replica")[columna]


def rugosidad_guardada(datos, objetivo):
    """Rugosidad de cada réplica en un espesor guardado (todas las réplicas deben alcanzarlo)."""
    k = int(np.flatnonzero(datos["objetivos"] == objetivo)[0])
    return datos["instantaneas"][:, k].reshape(datos["instantaneas"].shape[0], -1).std(axis=1)


def diferencias_pareadas(base, variante):
    """Diferencia relativa media (%) y su IC del 95 % entre dos series pareadas por réplica, con el valor p."""
    comunes = base.index.intersection(variante.index)
    if len(comunes) < 2:
        return {"pares": len(comunes)}
    dif = variante[comunes].to_numpy() - base[comunes].to_numpy()
    media, ic = media_ic95(dif) if np.any(dif != 0) else (0.0, 0.0)
    p = stats.ttest_rel(variante[comunes], base[comunes]).pvalue if np.any(dif != 0) else 1.0
    return {"pares": len(comunes), "base": base[comunes].mean(), "dif. (%)": 100 * media / base[comunes].mean(),
            "IC95 (%)": 100 * ic / base[comunes].mean(), "p (t pareada)": p}


def huecos_en(datos, objetivo):
    """Número de huecos de cada réplica en un espesor guardado (``nan`` si la réplica no lo alcanzó)."""
    k = int(np.flatnonzero(datos["objetivos"] == objetivo)[0])
    i = int(np.searchsorted(datos["eventos"], datos["eventos_instantaneas"][k]))
    return datos["huecos"][:, i]


def porosidad_incremental(datos, inferior, superior):
    """Porosidad de la capa depositada entre dos espesores guardados, por réplica."""
    L2 = datos["L"] ** 2
    dv = huecos_en(datos, superior) - huecos_en(datos, inferior)
    return dv / ((superior - inferior) * L2 + dv)


def volumen_de(clave, L, espesor_final, nombre=None, banderas=None, **cambios):
    """Primera réplica de una configuración, con la ocupación de cada posición."""
    nombre = nombre or (CLAVES[clave] if L == NX else f"{CLAVES[clave]}, L = {L}")
    banderas = banderas or MODELOS[CLAVES[clave]]
    n_niveles = int(np.ceil(3 * espesor_final + RELIEVE_MAXIMO + 10))
    return simular_volumen(parametros(L, **cambios), nombre, banderas, SEMILLA, espesor_final, n_niveles,
                           max_pasos_rayo=MAX_PASOS_RAYO, relieve_maximo=RELIEVE_MAXIMO)


def espectro_por_anillos(superficies):
    """Espectro de potencia de las alturas de cada superficie de un arreglo (réplicas, L, L), promediado en los anillos
    de número de onda entero n = redondeo(L |k|), n = 1 ... L/2. La suma de la potencia de todos los modos es W^2.

    Devuelve (n, S), con S de forma (réplicas, L/2) en capas^2.
    """
    superficies = np.asarray(superficies, dtype=float)
    L = superficies.shape[-1]
    h = superficies - superficies.mean(axis=(-2, -1), keepdims=True)
    potencia = np.abs(np.fft.fft2(h)) ** 2 / L ** 4
    m = np.fft.fftfreq(L, 1.0 / L)
    anillo = np.rint(np.hypot(m[:, None], m[None, :])).astype(int).ravel()
    dentro = (anillo >= 1) & (anillo <= L // 2)
    conteo = np.bincount(anillo[dentro], minlength=L // 2 + 1)[1:]
    S = np.array([np.bincount(anillo[dentro], weights=p.ravel()[dentro], minlength=L // 2 + 1)[1:] for p in potencia])
    return np.arange(1, L // 2 + 1), S / conteo


def maximo_del_espectro(n, S, L, sorteo, n_remuestreos):
    """Máximo del espectro medio de las réplicas: su anillo n*, la separación Lambda = L / n* (celdas), el error
    estándar de Lambda remuestreando las réplicas con reemplazo y el contraste S(n*) / S(1)."""
    media = S.mean(axis=0)
    k = int(np.argmax(media))
    remuestras = [L / n[int(np.argmax(S[sorteo.integers(0, len(S), len(S))].mean(axis=0)))] for _ in range(n_remuestreos)]
    return int(n[k]), float(L / n[k]), float(np.std(remuestras, ddof=1)), float(media[k] / media[0])
```

## 6. Validación

1. **Conservación de partículas.** `simular_modelo` comprueba en cada réplica, al terminar y en cada superficie guardada, que la suma de las alturas de las columnas es el número de átomos más el de huecos, y que sin adhesión lateral no hay huecos.
2. **Contención del modelo M6.** Simulado con el nombre de M6 y la bandera `enable_lateral` apagada, el modelo M7 da exactamente las superficies de M6.
3. **Geometría del contacto lateral.** Sobre una superficie plana se levanta un muro de 30 capas que ocupa toda la fila $x = 30$, y se trazan rayos dirigidos a celdas situadas hasta diez celdas detrás del muro. Con la dirección de emisión de cada rayo se calculan las alturas $z_{\mathrm{e}}$ y $z_{\mathrm{s}}$ a las que su recta entra en la celda del muro y sale de ella. Si $z_{\mathrm{e}} < 30$, el trazador debe informar de un contacto lateral desde la columna $(29, y)$, con $y$ la fila de llegada, al nivel $\lfloor z_{\mathrm{e}} \rfloor$; si $z_{\mathrm{s}} \le 30 \le z_{\mathrm{e}}$, el rayo llega a la cima del muro sin contacto lateral; en el resto de los casos llega a su celda de mira. Se comprueban todos los rayos. La comprobación del registro de la ocupación (criterio 3) se hace en el experimento IV.

```python
# 6.1 La conservación de partículas (con los huecos) la comprueba simular_modelo en cada simulación.
# 6.2 Sin la regla de adhesión, M7 es M6: mismas superficies con el nombre de M6
L_CONTENCION, REPLICAS_CONTENCION, ESPESOR_CONTENCION = 50, 4, 8
contencion_m6 = simular("M6", L_CONTENCION, REPLICAS_CONTENCION, ESPESOR_CONTENCION, [ESPESOR_CONTENCION], n_log=10)
contencion_m7 = simular("M7", L_CONTENCION, REPLICAS_CONTENCION, ESPESOR_CONTENCION, [ESPESOR_CONTENCION], n_log=10,
                        nombre=contencion_m6["nombre"], banderas=dict(MODELOS[CLAVES["M7"]], enable_lateral=False))
contiene_m6 = np.array_equal(contencion_m6["instantaneas"], contencion_m7["instantaneas"])

# 6.3 Geometría del contacto lateral: muro de ALTURA_MURO capas en la fila x = X_MURO de una superficie plana
L_MURO, X_MURO, ALTURA_MURO, N_RAYOS = 200, 30, 30, 2000
superficie_muro = np.zeros((L_MURO, L_MURO), dtype=np.int32)
superficie_muro[X_MURO, :] = ALTURA_MURO
sorteo = np.random.default_rng(SEMILLA)
motor.sembrar_generador(SEMILLA)
conteo = {"contacto lateral": [0, 0], "llegada a la cima": [0, 0], "llegada a la celda de mira": [0, 0]}  # [rayos, conformes]
for _ in range(N_RAYOS):
    x_mira, y_mira = int(sorteo.integers(X_MURO + 1, X_MURO + 11)), int(sorteo.integers(0, L_MURO))
    cx, cy, _, theta_x, _, _, ax, ay, nivel = motor.trazar_vuelo(
        x_mira, y_mira, L_MURO, L_MURO, superficie_muro, SIGMA_INCIDENCIA_INIT, 0.0, np.inf, ALTURA_MURO,
        float(SOURCE_DISTANCE), 1, MAX_PASOS_RAYO, EMISION_GAUSSIANA, 0.0)
    a = -np.tan(np.radians(theta_x))                    # avance hacia el muro por capa de descenso
    z_e = (x_mira - X_MURO + 0.5) / a if a > 0 else np.inf   # altura a la que la recta entra en la celda del muro
    z_s = (x_mira - X_MURO - 0.5) / a if a > 0 else np.inf   # y a la que sale de ella
    if z_e < ALTURA_MURO:
        caso, conforme = "contacto lateral", (cx, ax, ay, nivel) == (X_MURO, X_MURO - 1, cy, int(np.floor(z_e)))
    elif z_s <= ALTURA_MURO:
        caso, conforme = "llegada a la cima", (cx, ax, nivel) == (X_MURO, -1, -1)
    else:
        caso, conforme = "llegada a la celda de mira", (cx, cy, nivel) == (x_mira, y_mira, -1)
    conteo[caso][0] += 1
    conteo[caso][1] += int(conforme)
df_muro = pd.DataFrame([{"caso": c, "rayos": n, "conformes": k} for c, (n, k) in conteo.items()])
geometria_ok = bool((df_muro["rayos"] > 0).all() and (df_muro["rayos"] == df_muro["conformes"]).all())

print("=" * ANCHO)
print("6. VALIDACIÓN")
print("=" * ANCHO)
fila("Sin adhesión lateral: superficies idénticas a M6", "sí" if contiene_m6 else "NO")
display(df_muro)
fila("Contacto lateral conforme a la geometría del rayo", "sí" if geometria_ok else "NO")
```

```python
# 7.8.2 Mapa (probabilidad de salto, camino libre medio) y dispersión angular
n_p, n_l = len(PROBABILIDADES_MAPA), len(CAMINOS_LIBRES_MAPA)
POROSIDAD = a_matriz(df_mapa, f"p_{ESPESOR_MAPA:g}", n_p, n_l)
W_MAPA = a_matriz(df_mapa, f"W_{ESPESOR_MAPA:g}", n_p, n_l)
SE_POROSIDAD = a_matriz(df_mapa, f"p_{ESPESOR_MAPA:g}_se", n_p, n_l)

def monotona(media, se, eje, signo):
    """Si signo * (diferencia entre celdas consecutivas a lo largo de eje) no supera dos errores estándar en ninguna celda."""
    dif = signo * np.diff(media, axis=eje)
    tolerancia = 2 * np.sqrt(np.take(se, range(1, se.shape[eje]), axis=eje) ** 2 + np.take(se, range(se.shape[eje] - 1), axis=eje) ** 2)
    return bool(np.all(np.isfinite(dif)) and np.all(dif <= tolerancia))

mapa_ok = monotona(POROSIDAD, SE_POROSIDAD, 0, 1) and monotona(POROSIDAD, SE_POROSIDAD, 1, 1)
etq_P = [f"P = {P:g}" for P in PROBABILIDADES_MAPA]
etq_l = ["∞" if np.isinf(lam) else f"{lam:g}" for lam in CAMINOS_LIBRES_MAPA]

n_pa, n_s = len(PROBABILIDADES_ANGULARES), len(DISPERSIONES_ANGULARES)
POROSIDAD_ANG = a_matriz(df_angular, f"p_{ESPESOR_MAPA:g}", n_pa, n_s)
SE_POROSIDAD_ANG = a_matriz(df_angular, f"p_{ESPESOR_MAPA:g}_se", n_pa, n_s)
angular_ok = bool(np.all(POROSIDAD_ANG[:, 0] == 0)) and monotona(POROSIDAD_ANG, np.nan_to_num(SE_POROSIDAD_ANG), 1, -1)
etq_Pa = [f"P = {P:g}" for P in PROBABILIDADES_ANGULARES]
etq_s = [f"{s:g}°" for s in DISPERSIONES_ANGULARES]

def como_tabla(matriz, filas, columnas, formato="{:.4f}"):
    return pd.DataFrame([[("—" if not np.isfinite(v) else formato.format(v)) for v in f] for f in matriz], index=filas, columns=columnas)

print("=" * ANCHO)
print(f"POROSIDAD A {ESPESOR_MAPA:g} CAPAS (filas: probabilidad de salto; columnas: camino libre medio, celdas)")
print("=" * ANCHO)
display(como_tabla(POROSIDAD, etq_P, etq_l))
print(f"Rugosidad a {ESPESOR_MAPA:g} capas:")
display(como_tabla(W_MAPA, etq_P, etq_l, "{:.3f}"))
print(f"Réplicas detenidas por el corte en el mapa: {int(df_mapa['detenidas'].sum())} de {int(df_mapa['replicas'].sum())}")
fila("Porosidad monótona en P y en lambda (2 SE)", "sí" if mapa_ok else "NO")
print()
print("=" * ANCHO)
print(f"POROSIDAD A {ESPESOR_MAPA:g} CAPAS SIN GAS (filas: probabilidad de salto; columnas: dispersión angular)")
print("=" * ANCHO)
display(como_tabla(POROSIDAD_ANG, etq_Pa, etq_s))
fila("Nula en vertical y monótona en la dispersión (2 SE)", "sí" if angular_ok else "NO")
```

```python
# 7.8.3 Estructura de la película: identidad con la réplica registrada y perfiles de densidad
e7 = ESTRUCTURAS["M7"]
volumen_ok = bool(np.array_equal(e7["superficie"], RESULTADOS["M7"]["superficies"][0])
                  and e7["depositados"] == int(RESULTADOS["M7"]["depositados"][0])
                  and e7["huecos"] == int(RESULTADOS["M7"]["huecos_finales"][0]))
filas_perfil = []
for nombre, perfil in PERFILES.items():
    e = ESTRUCTURAS[nombre]
    L = e["superficie"].shape[0]
    espesor = e["depositados"] / L ** 2
    interior = perfil[(perfil["nivel"] >= 0.25 * espesor) & (perfil["nivel"] < 0.75 * espesor)]
    filas_perfil.append({"película": nombre, "L": L, "espesor (capas)": espesor,
                         "altura media de las columnas": float(e["superficie"].mean()),
                         "porosidad global": e["huecos"] / (e["depositados"] + e["huecos"]),
                         "porosidad del interior": float(interior["porosidad"].mean()),
                         "densidad del primer nivel": float(perfil["densidad"].iloc[0])})
df_estructura = pd.DataFrame(filas_perfil)
guardar_tabla(df_estructura, "07_estructura")
print("=" * ANCHO)
print("ESTRUCTURA DE LA PELÍCULA (ocupación de cada posición)")
print("=" * ANCHO)
fila("Volumen registrado = primera réplica de M7", "sí" if volumen_ok else "NO")
print("Interior: niveles entre el 25 % y el 75 % del espesor de material.")
display(df_estructura.round(4))
```

```python
# 7.8.4 Tamaño de la malla
L_MAX = max(TAMANOS_MALLA)
filas_m = []
magnitudes = [("rugosidad", e) for e in ESPESORES_MALLA] + [("porosidad", ESPESOR_MALLA)]
for L in TAMANOS_MALLA:
    for columna, objetivo in magnitudes:
        v = obs_malla[(obs_malla["L"] == L) & (obs_malla["objetivo"] == objetivo)][columna]
        ref = obs_malla[(obs_malla["L"] == L_MAX) & (obs_malla["objetivo"] == objetivo)][columna]
        completo_m = len(v) == MALLA[L]["rugosidad"].shape[0] and len(ref) == MALLA[L_MAX]["rugosidad"].shape[0]
        filas_m.append({"L": L, "magnitud": f"{columna} a {objetivo:g} capas", "media": v.mean() if completo_m else np.nan,
                        "SE": v.sem() if completo_m else np.nan,
                        f"dif. con L = {L_MAX} (%)": 100 * (v.mean() / ref.mean() - 1) if completo_m and ref.mean() > 0 else np.nan,
                        "p Welch": p_welch(v, ref) if completo_m and L != L_MAX else np.nan})
df_malla = pd.DataFrame(filas_m)
base = df_malla[(df_malla["L"] == NX) & df_malla["p Welch"].notna()]
malla_ok = bool(len(base) == len(magnitudes) and (base[f"dif. con L = {L_MAX} (%)"].abs() < 1).all() and (base["p Welch"] > 0.01).all())
guardar_tabla(df_malla, "07_malla_resumen")
print("=" * ANCHO)
print(f"TAMAÑO DE LA MALLA (referencia L = {L_MAX})")
print("=" * ANCHO)
display(df_malla.round(4))
fila(f"L = {NX} frente a L = {L_MAX}: |dif.| < 1 % y p > 0.01", "sí" if malla_ok else "NO")
```

```python
# 7.8.6 Películas gruesas: exponentes, saturación, escalado con el tamaño de la malla y montículos (experimento VII)
# Exponente de crecimiento en cada ventana: pendiente de ln W frente a ln h, ajustada a cada réplica
filas_b = []
for L, d in GRUESA.items():
    ln_h = np.log(capas(d))
    for desde, hasta in VENTANAS_BETA:
        b, se_b = media_y_se(ajuste_por_replica(ln_h, np.log(d["rugosidad"]), np.log(desde), np.log(hasta))[:, 0])
        filas_b.append({"L": L, "desde": desde, "hasta": hasta, "beta": b, "SE": se_b})
df_beta_gruesa = pd.DataFrame(filas_b)

# Saturación: rugosidad cuadrática media de cada réplica en las dos mitades de la ventana de saturación
h_mitad = 0.5 * (VENTANA_SATURACION[0] + VENTANA_SATURACION[1])
filas_s, W_SAT = [], {}
for L, d in GRUESA.items():
    h = capas(d)
    primera = (h >= VENTANA_SATURACION[0]) & (h < h_mitad)
    segunda = (h >= h_mitad) & (h <= VENTANA_SATURACION[1])
    w2_a = np.mean(d["rugosidad"][:, primera] ** 2, axis=1)
    w2_b = np.mean(d["rugosidad"][:, segunda] ** 2, axis=1)
    W_SAT[L] = np.sqrt(np.mean(d["rugosidad"][:, primera | segunda] ** 2, axis=1))
    w, se_w = media_y_se(W_SAT[L])
    filas_s.append({"L": L, "réplicas": len(W_SAT[L]), "instantes": int(primera.sum()) + int(segunda.sum()),
                    "W en la ventana": w, "SE": se_w, "W^2: dif. entre mitades (%)": 100 * (w2_b.mean() / w2_a.mean() - 1),
                    "p (t pareada)": stats.ttest_rel(w2_b, w2_a).pvalue})
df_saturacion = pd.DataFrame(filas_s)
p_sat = dict(zip(df_saturacion["L"], df_saturacion["p (t pareada)"]))
saturadas = [L for L in MALLAS_SATURADAS if p_sat[L] > 0.01]
saturacion_ok = len(saturadas) == len(MALLAS_SATURADAS)

# Exponentes frente a los de la clase KPZ
tardia = df_beta_gruesa[(df_beta_gruesa["L"] == NX) & (df_beta_gruesa["desde"] == VENTANAS_BETA[-1][0])].iloc[0]
z_beta = (tardia["beta"] - BETA_KPZ) / np.hypot(tardia["SE"], SE_BETA_KPZ)
if len(saturadas) >= 2:
    reg_alfa = stats.linregress(np.concatenate([np.full(len(W_SAT[L]), np.log(L)) for L in saturadas]),
                                np.concatenate([np.log(W_SAT[L]) for L in saturadas]))
    alfa, se_alfa = reg_alfa.slope, reg_alfa.stderr
else:
    alfa, se_alfa = np.nan, np.nan
z_alfa = (alfa - ALFA_KPZ) / np.hypot(se_alfa, SE_ALFA_KPZ)
df_kpz = pd.DataFrame([
    {"exponente": f"beta (L = {NX}, {VENTANAS_BETA[-1][0]:g} a {VENTANAS_BETA[-1][1]:g} capas)", "estimado": tardia["beta"],
     "SE": tardia["SE"], "KPZ": BETA_KPZ, "z": z_beta},
    {"exponente": f"alfa (mallas saturadas: {', '.join(str(L) for L in saturadas) or 'ninguna'})", "estimado": alfa,
     "SE": se_alfa, "KPZ": ALFA_KPZ, "z": z_alfa},
])
kpz_ok = bool(np.isfinite(z_beta) and np.isfinite(z_alfa) and abs(z_beta) <= 3 and abs(z_alfa) <= 3)

# Espectro de potencia en anillos de número de onda entero y su máximo en cada espesor guardado
sorteo_espectro = np.random.default_rng(SEMILLA)
filas_k, ESPECTROS = [], {}
for L, d in GRUESA.items():
    for k_inst, objetivo in enumerate(d["objetivos"]):
        validas = d["instantanea_valida"][:, k_inst]
        n_anillo, S = espectro_por_anillos(d["instantaneas"][validas, k_inst])
        ESPECTROS[(L, objetivo)] = (n_anillo, S.mean(axis=0))
        n_max, separacion, se_sep, contraste = maximo_del_espectro(n_anillo, S, L, sorteo_espectro, N_REMUESTREOS)
        filas_k.append({"L": L, "espesor": objetivo, "réplicas": int(validas.sum()), "n del máximo": n_max,
                        "Lambda (celdas)": separacion, "SE": se_sep, "S(n*)/S(1)": contraste})
df_espectro = pd.DataFrame(filas_k)
interior = df_espectro[(df_espectro["L"] == NX) & (df_espectro["n del máximo"] > 1)]
reg_lambda = stats.linregress(np.log(interior["espesor"]), np.log(interior["Lambda (celdas)"])) if len(interior) >= 3 else None
temprano, tardio = (df_espectro[(df_espectro["L"] == NX) & (df_espectro["espesor"] == e)].iloc[0] for e in ESPESORES_MONTICULOS)
monticulos_ok = bool(temprano["n del máximo"] > 1
                     and tardio["Lambda (celdas)"] - temprano["Lambda (celdas)"] > 2 * np.hypot(temprano["SE"], tardio["SE"]))
guardar_tabla(df_beta_gruesa, "07_pelicula_gruesa_beta")
guardar_tabla(df_saturacion, "07_pelicula_gruesa_saturacion")
guardar_tabla(df_kpz, "07_pelicula_gruesa_kpz")
guardar_tabla(df_espectro, "07_pelicula_gruesa_espectro")

print("=" * ANCHO)
print(f"PELÍCULAS GRUESAS DE M7 (hasta {ESPESOR_GRUESA:g} capas)")
print("=" * ANCHO)
print("Exponente de crecimiento en cada ventana (filas: lado de la malla):")
display(df_beta_gruesa.pivot(index="L", columns="desde", values="beta").rename(
    columns={a: f"beta [{a:g}, {b:g}]" for a, b in VENTANAS_BETA}).round(4))
print(f"Rugosidad en la ventana de saturación [{VENTANA_SATURACION[0]:g}, {VENTANA_SATURACION[1]:g}] y diferencia entre sus mitades:")
display(df_saturacion.round(4))
fila(f"Saturación en las mallas {MALLAS_SATURADAS}", "sí" if saturacion_ok else f"NO (saturadas: {saturadas})")
print()
display(df_kpz.round(4))
fila("Exponentes compatibles con la clase KPZ (|z| <= 3)", "sí" if kpz_ok else "NO")
print()
print("Máximo del espectro de potencia en cada espesor guardado (Lambda = L / n*):")
display(df_espectro.round(3))
if reg_lambda is not None:
    fila(f"Lambda frente al espesor (L = {NX}, máximos interiores)",
         f"exponente {reg_lambda.slope:.3f} +- {reg_lambda.stderr:.3f} ({len(interior)} espesores)")
else:
    fila(f"Lambda frente al espesor (L = {NX})", f"menos de tres espesores con el máximo fuera de n = 1 ({len(interior)})")
fila(f"Montículos: máximo interior a {ESPESORES_MONTICULOS[0]:g} capas y Lambda mayor a {ESPESORES_MONTICULOS[1]:g}",
     "sí" if monticulos_ok else "NO")
```

## 9. Análisis de resultados

### 9.1 Contención y geometría del contacto lateral

Sin la regla de adhesión, M7 da exactamente las superficies de M6 (sección 6), y el criterio 1 se cumple. En la prueba del muro, los 460 rayos que entran en la celda del muro por debajo de su cima informan de un contacto lateral desde la columna $(29, y)$ al nivel $\lfloor z_{\mathrm{e}} \rfloor$, los 62 que llegan a su cima no informan de contacto y los 1478 restantes llegan a su celda de mira. Los 2000 rayos son conformes, y el criterio 2 se cumple. Son los mismos rayos de la prueba del cuaderno 05, con el mismo resultado.

### 9.2 Supresión de la inestabilidad

Ninguna réplica de M7 llega al corte, y el relieve máximo alcanzado en 256 capas es de 10.6 capas, mientras que M6 llega a 100 capas de relieve hacia las 9 capas de espesor (tabla 7.8.1 y figura 1). El criterio 4 se cumple. El relieve medio de M7 pasa de 3.6 capas a 1 capa de espesor a 7.2 a 256 capas, y en las primeras capas la película es más lisa que la de M6: un $13\,\%$ a 1 capa y un $33\,\%$ a 4 capas. El flujo que llega a una columna por los lados la ensancha en lugar de alargarla, y desaparece el mecanismo que producía las agujas.

La rugosidad de M7, sin embargo, sigue creciendo: 0.876 capas a 16 capas, 1.087 a 64 y 1.871 a 256, y más deprisa a mayor espesor. La superficie a 256 capas (figura 5, derecha) está formada por montículos y valles de unas decenas de celdas de ancho y unas pocas capas de altura, que a 16 capas no existían. La adhesión lateral elimina las agujas pero no la sombra: las regiones altas siguen recibiendo más flujo que los valles, y la diferencia se reparte ahora en anchura. El experimento VII sigue ese crecimiento hasta 4096 capas (sección 9.7).

### 9.3 Porosidad

La porosidad de M7 sube del $0.63\,\%$ a 1 capa al $1.56\,\%$ a 16 capas y se estabiliza en el $1.62\,\%$ (figura 1, derecha). La porosidad incremental, la de la capa depositada en cada intervalo, vale $1.64\,\%$ entre 16 y 64 capas y $1.62\,\%$ entre 64 y 256: difieren un $0.8\,\%$, y el criterio 5 se cumple. Las primeras capas son más densas porque crecen sobre un sustrato plano, donde las paredes son bajas y los contactos laterales escasos. A partir de unas 16 capas, la película crece con una estructura interna estacionaria, aunque su superficie siga haciéndose más rugosa.

### 9.4 Mapa y dispersión angular

A 64 capas, la porosidad va de cero con $P = 1$ al $62\,\%$ con $P = 0$ y $\lambda = 5$ celdas, y es monótona en las dos variables (tabla 7.8.2 y figura 2, izquierda); el criterio 6 se cumple. La rugosidad sigue el mismo orden, de 0.69 capas con $P = 1$ sin gas a 9.6 capas en la esquina fría y de presión alta. Sin gas, la porosidad es nula con incidencia vertical para cualquier $P$ y crece con la dispersión angular de la emisión (figura 2, derecha): con $P = 0$ pasa del $32\,\%$ a $5^\circ$ al $67\,\%$ a $45^\circ$, con $P = 0.4$ del $4.9\,\%$ al $45\,\%$, y con $P = 0.8$ del $0.16\,\%$ al $2.2\,\%$. El criterio 7 se cumple.

Los dos ceros tienen una explicación geométrica. Un átomo que cae en vertical no cruza el borde de ninguna celda y no puede alcanzar una pared: a diferencia de la deposición balística de Vold (1959), en la que la partícula ocupa una celda y se adhiere en cuanto toca a una vecina, aquí el átomo es un punto y solo alcanza una pared si su trayectoria es oblicua. La porosidad del modelo nace, por tanto, de la oblicuidad del flujo, como en las películas depositadas con incidencia oblicua (Krug & Meakin, 1989; Dirks & Leamy, 1977). Con $P = 1$, en cambio, el flujo es oblicuo, pero un hueco exige que el átomo alcance la pared de B por encima de la cima de A, es decir, $h_B \geq h_A + 2$, y la relajación con $P = 1$ no deja desniveles de dos capas entre vecinos (cuaderno 02): la porosidad es nula. En el extremo opuesto, sin movilidad, basta una dispersión de $5^\circ$ para que un tercio del volumen quede vacío.

### 9.5 Estructura de la película

La película registrada con su ocupación es la primera réplica de M7, posición por posición, y el criterio 3 se cumple (tabla 7.8.3). Los perfiles de la figura 3 muestran tres estructuras distintas. En M7, la densidad es de 0.99 desde el primer nivel hasta la región de la superficie, con una porosidad interior del $1.62\,\%$, prácticamente igual a la global ($1.61\,\%$); los huecos son posiciones sueltas o tramos horizontales cortos, repartidos de manera uniforme (figura 4, arriba). La película fría a presión alta ocupa 170 capas de altura media para 64 capas de material, con una porosidad del $63\,\%$ en el interior, constante desde unas 5 capas por encima del sustrato hasta unas 150, y una densidad de solo 0.61 ya en el primer nivel: los contactos laterales empiezan con la primera capa. Su corte vertical (figura 4, centro) es una red de tramos horizontales de material separados por grandes regiones vacías. La película caliente sin gas es compacta.

### 9.6 Tamaño de la malla y número máximo de saltos

La malla de 200 celdas difiere de la de 400 en menos de un $1\,\%$ en la rugosidad a 4, 16 y 64 capas y en la porosidad a 64 capas (la mayor diferencia, $-0.90\,\%$ en la rugosidad a 64 capas, con $p = 0.20$; tabla 7.8.4), y el criterio 8 se cumple. La de 50 celdas empieza a separarse a 16 y 64 capas ($-1.9$ y $-2.4\,\%$, con $p = 0.016$ y 0.24); es menor que la separación de los montículos que la superficie desarrolla a mayor espesor (sección 9.7). Con 10 saltos máximos, la rugosidad a 16 capas es idéntica y las diferencias a 64 capas son de un $0.07\,\%$ en la rugosidad y de un $0.004\,\%$ en la porosidad ($p = 0.34$; tabla 7.8.5), y el criterio 9 se cumple. En el experimento III, la salvaguarda del trazador detiene como mucho 20 vuelos casi rasantes por configuración, entre 6.4 millones de átomos, todos con emisiones de 25 a $45^\circ$, y ninguno en el resto de los experimentos.

### 9.7 Películas gruesas

Hasta unas 100 capas, la rugosidad de las mallas de 32 a 200 celdas es la misma, y solo la de 16 celdas se satura antes, hacia las 20 capas, en $W = 0.85$ (figura 7, arriba a la izquierda). A partir de ahí las mallas se separan en dos grupos. En las de 16 y 32 celdas la rugosidad deja de crecer en $W_{\mathrm{sat}} = 0.854$ y $1.352$ capas. En las de 64, 128 y 200 celdas crece mucho más deprisa, con exponentes de 0.66, 0.98 y 1.03 entre 256 y 1024 capas (tabla 7.8.6), y se satura después en 6.79, 8.96 y 11.3 capas. En la ventana de 2048 a 4096 capas, la rugosidad de las cinco mallas no cambia entre las dos mitades ($p \geq 0.60$), y en particular la de las mallas de 32 y 64 celdas ($p = 0.89$ y 0.84): el criterio 10 se cumple.

El crecimiento no es el de la clase KPZ. En la malla de 200 celdas, el exponente entre 1024 y 4096 capas vale $0.154 \pm 0.026$, a 3.4 errores estándar de 0.2415, porque la película ya se está saturando, y entre 256 y 1024 capas vale 1.03, cuatro veces el de KPZ. Las rugosidades de saturación de 32 y 64 celdas dan $\alpha = 2.33 \pm 0.02$, frente a 0.387: de una malla a la otra, $W_{\mathrm{sat}}$ se multiplica por cinco. El colapso con los exponentes de KPZ (figura 7, arriba a la derecha) no superpone las curvas. El criterio 11 no se cumple, y por un margen que no deja dudas.

El espectro de potencia explica los dos grupos. En la malla de 200 celdas, a 128 y 256 capas el espectro tiene un máximo en $n = 4$, una separación de los montículos de $\Lambda = 50$ celdas, con un contraste $S(n^*)/S(1)$ de 1.4 y 2.0; a 512 capas el máximo pasa a $n = 2$ ($\Lambda = 100$ celdas), y desde 1024 capas está en $n = 1$ (figura 7, abajo). En la de 128 celdas, la separación pasa de 64 a 128 celdas entre 256 y 512 capas. El máximo está fuera de $n = 1$ a 256 capas y la separación a 2048 capas es cuatro veces mayor, de modo que el criterio 12 se cumple. La superficie de M7 desarrolla montículos con una separación inicial de unas 50 celdas, que engrosan con el espesor. Mientras engrosan, la rugosidad crece con un exponente cercano a 1, el que Drotar et al. (2000) obtienen en un modelo de sombreado con crecimiento lateral. Cuando la separación alcanza el lado de la malla, el engrosamiento se detiene y la rugosidad se satura, con un valor que fija el tamaño de la malla: entre 64 y 200 celdas, $W_{\mathrm{sat}}$ crece aproximadamente como $L^{0.44}$. En las mallas de 16 y 32 celdas, menores que la separación inicial, los montículos no caben, la inestabilidad no se desarrolla y la película se satura con la rugosidad pequeña de una superficie relajada. De ahí el salto de un factor cinco entre 32 y 64 celdas, que no es un exponente de escala sino un umbral.

Que la malla de 200 celdas sea apenas cuatro veces mayor que la separación inicial limita lo que este experimento puede decir. El exponente de engrosamiento de la separación se apoya en tres espesores con el máximo fuera de $n = 1$ y vale $0.50 \pm 0.29$, sin precisión útil, y el valor de la rugosidad a 256 capas en la malla de 200 celdas está ya influido por el tamaño, porque a ese espesor caben en ella solo unos cuatro montículos por lado.

## 10. Conclusiones

1. La adhesión lateral suprime la inestabilidad de columnas: ninguna réplica de M7 supera 10.6 capas de relieve en 256 capas, mientras que M6 llega a 100 capas hacia las 9. El flujo que llega por los lados ensancha las columnas en lugar de alargarlas.
2. La película deja de ser compacta. En la condición base tiene una porosidad del $1.6\,\%$, que se estabiliza hacia las 16 capas y no cambia después ($1.64\,\%$ y $1.62\,\%$ en las capas depositadas entre 16 y 64 y entre 64 y 256 capas).
3. La porosidad la controlan la oblicuidad del flujo y la movilidad de los átomos. Es nula con incidencia vertical y con $P = 1$, crece con la dispersión angular y con la presión, y llega al $62\,\%$ en una película fría a presión alta. Es la tendencia que se observa en las películas evaporadas a baja temperatura (Thornton, 1977; Dirks & Leamy, 1977), aquí sin calibración física.
4. Sin agujas, el sombreado produce montículos. La superficie de M7 desarrolla montículos con una separación inicial de unas 50 celdas, que engrosan con el espesor; mientras engrosan, la rugosidad crece con un exponente cercano a 1, y cuando la separación alcanza el lado de la malla la rugosidad se satura. El crecimiento no es el de la clase KPZ de la deposición balística, y el criterio 11 no se cumple.
5. En mallas menores que la separación de los montículos, la inestabilidad no se desarrolla: las mallas de 16 y 32 celdas se saturan con una rugosidad de 0.85 y 1.35 capas, y la de 64 con 6.8. Los resultados de M7 a espesores grandes dependen del tamaño de la malla, y la de 200 celdas solo acomoda unos cuatro montículos por lado a 256 capas.
6. Hasta 64 capas, los resultados no dependen del tamaño de la malla ni del número máximo de saltos, y la contención de M6 y la geometría del contacto son exactas. Se cumplen once de los doce criterios de la sección 2.6.
7. Los huecos son cerrados por construcción, porque ningún vuelo entra bajo un voladizo. La limitación afecta sobre todo a las películas más porosas del mapa, cuyos huecos son grandes y estarían en parte al alcance de vuelos muy oblicuos.
