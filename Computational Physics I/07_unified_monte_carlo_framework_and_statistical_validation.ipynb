## 1. Introducción y mecanismos físicos en competencia

Este cuaderno cierra la serie que ha estudiado, uno a uno, los mecanismos físicos que determinan la morfología de una película delgada de cobre (Cu) depositada sobre vidrio por evaporación térmica en una cámara de vacío (Ohring, 2002). Los cuadernos anteriores aislaron la deposición aleatoria, la relajación superficial, la temperatura del sustrato, la incidencia angular, el sombreado geométrico y la dispersión en fase gaseosa. Este séptimo cuaderno reúne esos mecanismos en un único motor computacional, con banderas que activan o desactivan cada uno, y añade una capa de validación estadística mediante réplicas independientes.

La morfología final de una película resulta de un balance entre mecanismos que amplifican las fluctuaciones de altura y mecanismos que las atenúan.

**Incidencia angular.** Un átomo emitido con un ángulo $\theta$ respecto a la normal se desplaza lateralmente antes del impacto (Nieuwenhuizen & Haanstra, 1966). Sobre una malla con fronteras periódicas, sumar a una posición uniforme un desplazamiento aleatorio independiente de ella da de nuevo una posición uniforme, de modo que este mecanismo por sí solo no debería alterar la estadística de los impactos (sección 9 valida esta predicción).

**Sombreado geométrico.** Las regiones que por azar quedan más altas interceptan con mayor probabilidad el flujo de vapor entrante, lo que bloquea la llegada de material a los valles vecinos. Es un mecanismo de retroalimentación positiva, dominante en alto vacío, que favorece estructuras columnares y una rugosidad creciente (Karunasiri et al., 1989).

**Dispersión en fase gaseosa.** Un gas residual en la cámara reduce el camino libre medio de los átomos evaporados y desvía sus trayectorias mediante colisiones moleculares. Cada colisión ensancha la distribución angular con la que los átomos alcanzan la película, de modo que la llegada se vuelve más oblicua y refuerza, en lugar de contrarrestar, el sombreado geométrico ya presente. Bajo este modelo la dispersión gaseosa no actúa como un mecanismo estabilizador.

**Relajación superficial.** Una vez que un adátomo colisiona con la película, la energía térmica del sustrato le permite migrar lateralmente hacia sitios de menor altura (Venables, 2000). Este es el único mecanismo genuinamente estabilizador de los cuatro: actúa como un filtro de paso bajo que suaviza la rugosidad de alta frecuencia espacial (Family, 1986; Edwards & Wilkinson, 1982).

## 2. Ley de escala dinámica de Family-Vicsek

Para comparar cuantitativamente los seis modelos se recurre a la teoría de escalamiento dinámico (Family & Vicsek, 1985). Una interfaz en crecimiento es un sistema fuera del equilibrio, cuya rugosidad superficial RMS,

$$W(\overline{h}) = \sqrt{\frac{1}{N_x N_y} \sum_{i,j} \left[ h(i,j) - \overline{h} \right]^2},$$

crece con el espesor medio $\overline{h}$ (que actúa como métrica lineal del tiempo de deposición) siguiendo una ley de potencias antes de que los efectos de tamaño finito la saturen:

$$W(\overline{h}) \propto \overline{h}^{\,\beta}.$$

Tomando logaritmos,

$$\ln W = \beta \ln \overline{h} + C,$$

de modo que la pendiente de una regresión lineal en el espacio log-log entrega el exponente de crecimiento $\beta$. La sección 12 estima $\beta$ para cada modelo sobre intervalos de espesor fijados de antemano, en vez de buscar el intervalo que maximiza el coeficiente de determinación de la regresión: esa búsqueda tiende a seleccionar tramos cortos que, en una única realización, parecen más lineales por azar, sin relación necesaria con el régimen físico de crecimiento, y además puede elegir un tramo distinto en cada réplica, lo que impide promediar $\beta$ de forma consistente entre ellas.

## 3. Clases de universalidad

El valor de $\beta$ permite situar cada régimen dentro de familias de comportamiento con propiedades de escala compartidas, o clases de universalidad (Barabási & Stanley, 1995). Para una interfaz que crece sobre un sustrato bidimensional, los valores de referencia más citados en la literatura son:

* **Deposición aleatoria (RD):** $\beta = 0.5$. Sin relajación ni correlación espacial entre columnas vecinas, la rugosidad crece sin límite como $\sqrt{\overline{h}}$.
* **Clase Edwards-Wilkinson (EW):** $\beta \approx 0.24$ en un sustrato bidimensional (Family, 1986; Edwards & Wilkinson, 1982). La relajación superficial actúa como un término difusivo que amortigua el relieve de alta frecuencia espacial.
* **Sombreado o crecimiento columnar:** $\beta > 0.5$. El transporte geométrico domina sobre la relajación térmica y la película se vuelve inestable y porosa (Karunasiri et al., 1989).

Estos valores describen el régimen de crecimiento asintótico de sistemas idealmente grandes y en tiempos largos. Los espesores simulados en este cuaderno (hasta 2 capas) y el tamaño finito de la malla ($200 \times 200$ celdas) sitúan la suite en un régimen de crecimiento temprano, no necesariamente saturado, de modo que los exponentes obtenidos en la sección 12 deben leerse como una caracterización cuantitativa de este modelo particular, y no como una verificación directa de los valores asintóticos de la literatura.

## 4. Alcance del motor unificado

El motor computacional de este cuaderno comparte, para los seis modelos, la aproximación *solid-on-solid* sobre una malla $N_x \times N_y$ con fronteras periódicas (geometría toroidal) y, cuando la difusión está activa, la regla de relajación térmica introducida en el cuaderno 3 y reutilizada en los cuadernos 4 a 6: un adátomo salta a un vecino de von Neumann estrictamente más bajo con probabilidad $P(T) = \min(1, T/T_1)$, donde $T_1 = 500$ K es una temperatura de escala sin calibrar. Esta suite no reproduce la relajación instantánea y atérmica del cuaderno 2, que incorpora siempre la partícula a la celda de menor altura de su vecindad sin una probabilidad de salto, ni las fronteras abiertas del cuaderno 4; los resultados de este cuaderno no son, por tanto, una repetición exacta de esos dos cuadernos, sino una síntesis construida sobre el esquema común a los cuadernos 3 a 6.

Bajo fronteras periódicas, la incidencia angular sin sombreado no debería modificar la rugosidad global (sección 1): la sección 9 contrasta esa predicción con una prueba de hipótesis formal. La regla de relajación empleada es una variante con activación térmica de la relajación instantánea de Family (1986); su límite continuo es, como el de esa regla, la ecuación de Edwards y Wilkinson (Edwards & Wilkinson, 1982; Buceta et al., 2014), aunque esa pertenencia no se demuestra en este cuaderno.

## 8. Motor computacional unificado

El motor se organiza en tres bloques que se ensamblan en un único lazo de deposición: transporte entre la fuente y el sustrato, relajación térmica del adátomo recién llegado y cómputo de los observables morfológicos. Cada mecanismo se activa mediante una bandera booleana.

`calcular_estadisticos_superficie` obtiene el espesor medio y la rugosidad RMS con dos recorridos de la malla, el primero para la media y el segundo para las desviaciones respecto a ella, lo que evita la cancelación numérica de restar $\langle h^2 \rangle - \langle h \rangle^2$ cuando la media es grande frente a la dispersión.

`trazador_rayos_dispersion` traza el vuelo del átomo paso a paso cuando el sombreado o la dispersión gaseosa están activos. Las distancias libres entre colisiones se muestrean de una distribución exponencial de media $\lambda$ (Kalos & Whitlock, 2008), medida sobre la longitud de la trayectoria y no sobre la altura descendida, y el paso vertical se acorta cuando la pendiente supera un valor crítico, para que el rayo no avance más de una celda lateral por paso y no omita columnas del relieve. `modulo_transporte_fase_gaseosa` selecciona, según las banderas activas, entre este trazado paso a paso y un cálculo analítico directo del punto de impacto cuando ni el sombreado ni la dispersión están presentes, mucho más rápido porque no recorre el vuelo.

`modulo_relajacion_superficial` aplica la relajación térmica de la sección 4, con dos recorridos de la vecindad de von Neumann (uno para contar los vecinos más bajos y otro para localizar el elegido) en lugar de construir un arreglo de candidatos en cada salto.

`ejecutar_motor_unificado_completo` implementa el lazo secuencial de deposición y registra el espesor y la rugosidad cada `guardar_cada` eventos. Es estrictamente secuencial: el transporte y la relajación de un evento dependen del relieve que dejaron los eventos anteriores sobre la misma matriz de alturas, de modo que no se paraleliza dentro de una réplica.

`ejecutar_suite_replicas` sí se paraleliza, con `@njit(parallel=True)` y `prange` sobre el índice de réplica. A diferencia del lazo de deposición, las réplicas de un mismo modelo son independientes entre sí: cada una tiene su propia superficie y su propio generador de números aleatorios, sembrado con `semillas[rep]` al comienzo de la iteración, de modo que su resultado depende únicamente de esa semilla y no del orden ni del hilo en que Numba la ejecute.

```python
@njit
def calcular_estadisticos_superficie(superficie, nx, ny):
    """Calcula el espesor medio y la rugosidad RMS de una superficie.

    Parámetros
    ----------
    superficie : ndarray de enteros, forma (nx, ny)
        Altura de cada celda, en capas.
    nx, ny : int
        Número de celdas en cada dirección.

    Devuelve
    --------
    espesor : float
        Altura media, en capas.
    rugosidad : float
        Desviación estándar poblacional de las alturas (divisor nx * ny), en capas.
    """
    # Se calcula primero la media y después las desviaciones respecto a ella
    # (dos pasadas). Restar <h^2> - <h>^2 en una sola pasada perdería precisión
    # cuando la media es grande frente a la dispersión.
    suma_alturas = 0.0
    for i in range(nx):
        for j in range(ny):
            suma_alturas += superficie[i, j]
    espesor = suma_alturas / (nx * ny)

    suma_varianza = 0.0
    for i in range(nx):
        for j in range(ny):
            diferencia = superficie[i, j] - espesor
            suma_varianza += diferencia * diferencia
    rugosidad = np.sqrt(suma_varianza / (nx * ny))

    return espesor, rugosidad
```

```python
@njit
def trazador_rayos_dispersion(x_target, y_target, nx, ny, superficie, sigma_ang, sigma_disp,
                              mfp, max_h_actual, paso_z, distancia_fuente):
    """Determina la celda de colisión de una partícula con sombreado y dispersión en fase gaseosa.

    Los ángulos de emisión theta_x y theta_y se sortean de una distribución normal de
    media cero y desviación sigma_ang (en grados). La partícula se lanza desde un techo
    situado distancia_fuente capas por encima de la altura máxima, con una posición
    lateral tal que, sin colisiones, su recta pasa por el punto de mira (x_target,
    y_target) en z = 0. Desciende por pasos, que no superan paso_z ni permiten un
    avance lateral mayor que una celda, y se acortan para terminar en cada colisión con
    el gas y en el plano z = 0. Las distancias libres siguen una distribución
    exponencial de media mfp; en cada colisión se suma a theta_x y a theta_y un ruido
    normal de desviación sigma_disp. La partícula queda en el primer paso en que su
    altura z no supera la de la celda que sobrevuela. Las posiciones laterales son
    periódicas.

    Parámetros
    ----------
    x_target, y_target : int
        Punto de mira en el plano z = 0.
    nx, ny : int
        Número de celdas en cada dirección.
    superficie : ndarray de enteros, forma (nx, ny)
        Alturas actuales, en capas. La función no las modifica.
    sigma_ang : float
        Desviación estándar de los ángulos de emisión, en grados.
    sigma_disp : float
        Desviación estándar de la desviación angular por colisión, en grados.
    mfp : float
        Camino libre medio, en celdas. Con np.inf no hay colisiones.
    max_h_actual : int
        Altura máxima actual de la superficie, en capas.
    paso_z : float
        Paso vertical máximo del trazado, en capas.
    distancia_fuente : float
        Distancia entre el techo y la altura máxima de la película, en capas.

    Devuelve
    --------
    cx, cy : int
        Índices de la celda de colisión.
    n_col : int
        Número de colisiones con el gas sufridas durante el vuelo.
    """
    theta_x = np.random.normal(0.0, sigma_ang)
    theta_y = np.random.normal(0.0, sigma_ang)

    # Techo de vuelo, medido desde la parte más alta de la película: así la geometría
    # del vuelo no depende del espesor acumulado.
    z = float(max_h_actual) + distancia_fuente

    # Posición de lanzamiento: la recta inicial pasa por el punto de mira en z = 0.
    x = x_target + z * np.tan(theta_x * np.pi / 180.0)
    y = y_target + z * np.tan(theta_y * np.pi / 180.0)

    # Distancia de trayectoria hasta la primera colisión. Como el gas no tiene memoria,
    # basta volver a sortear la distancia después de cada colisión (transformada
    # inversa de la exponencial; 1 - random() está en (0, 1], de modo que el logaritmo
    # es finito).
    if np.isinf(mfp):
        s_libre = np.inf
    else:
        s_libre = mfp * (-np.log(1.0 - np.random.random()))

    n_col = 0
    cx = x_target
    cy = y_target

    while True:
        # Pendientes laterales por capa de descenso con la dirección actual.
        tan_x = np.tan(theta_x * np.pi / 180.0)
        tan_y = np.tan(theta_y * np.pi / 180.0)

        # Con pendientes grandes un paso de paso_z avanzaría varias celdas en horizontal
        # y podría omitir columnas: se limita el avance lateral a una celda por paso.
        # Con |tan| <= 1 / paso_z el paso es paso_z.
        pendiente = max(abs(tan_x), abs(tan_y))
        dz = paso_z
        if pendiente * dz > 1.0:
            dz = 1.0 / pendiente

        # Longitud de trayectoria recorrida en el paso. La distancia libre se mide sobre
        # la trayectoria, no sobre la altura descendida.
        ds = dz * np.sqrt(1.0 + tan_x * tan_x + tan_y * tan_y)

        # Si la colisión ocurre dentro del paso, este termina en el punto de colisión.
        hay_colision = False
        if ds >= s_libre:
            dz = dz * s_libre / ds
            ds = s_libre
            hay_colision = True

        # El descenso termina en z = 0: el sustrato siempre detiene a la partícula.
        if dz >= z:
            dz = z
            hay_colision = False

        z -= dz
        x -= dz * tan_x
        y -= dz * tan_y
        s_libre -= ds

        # Condiciones de frontera periódicas. El resto es no negativo en Python y en
        # Numba; el ajuste solo actúa como salvaguarda.
        idx_x = int(np.round(x)) % nx
        idx_y = int(np.round(y)) % ny
        if idx_x < 0:
            idx_x += nx
        if idx_y < 0:
            idx_y += ny

        # Criterio de interceptación (sombreado)
        if z <= superficie[idx_x, idx_y]:
            cx = idx_x
            cy = idx_y
            break

        # Colisión con el gas: se desvía la dirección y se sortea la siguiente distancia libre.
        if hay_colision:
            theta_x += np.random.normal(0.0, sigma_disp)
            theta_y += np.random.normal(0.0, sigma_disp)
            n_col += 1
            s_libre = mfp * (-np.log(1.0 - np.random.random()))

    return cx, cy, n_col


@njit
def modulo_transporte_fase_gaseosa(x_target, y_target, nx, ny, superficie, max_h_actual, sigma_init,
                                   sigma_scat, mfp, paso_z, distancia_fuente,
                                   enable_angle, enable_scattering, enable_shadowing):
    """Transporte unificado del átomo entre la fuente y el sustrato.

    Activa de forma selectiva la emisión angular, las colisiones con el gas y el sombreado
    por el relieve. Devuelve la celda de colisión y el número de colisiones con el gas.

    * Sin sombreado, la trayectoria no interactúa con la superficie. Con emisión angular, el
      impacto se desplaza lateralmente L * tan(theta) respecto del punto de mira (con fronteras
      periódicas), y sin ella coincide con el punto de mira. La dispersión gaseosa sin
      sombreado no está admitida (ver ``ejecutar_modelo``).
    * Con sombreado pero sin emisión angular ni gas, los rayos son verticales y ningún relieve
      los intercepta antes de la celda de mira, de modo que el impacto es esa celda y no
      hace falta trazar el vuelo.
    * En el resto de los casos se traza el vuelo con ``trazador_rayos_dispersion``.
    """
    if not enable_shadowing:
        if enable_angle:
            # Desplazamiento lateral Delta = L tan(theta) en cada eje, con theta ~ N(0, sigma_init).
            theta_x = np.random.normal(0.0, sigma_init)
            theta_y = np.random.normal(0.0, sigma_init)
            idx_x = int(np.round(x_target + distancia_fuente * np.tan(theta_x * np.pi / 180.0))) % nx
            idx_y = int(np.round(y_target + distancia_fuente * np.tan(theta_y * np.pi / 180.0))) % ny
            return idx_x, idx_y, 0
        return x_target, y_target, 0

    if not enable_angle and not enable_scattering:
        return x_target, y_target, 0

    sigma_ang = sigma_init if enable_angle else 0.0
    mfp_efectivo = mfp if enable_scattering else np.inf
    return trazador_rayos_dispersion(
        x_target, y_target, nx, ny, superficie, sigma_ang, sigma_scat, mfp_efectivo,
        max_h_actual, paso_z, distancia_fuente
    )
```

```python
@njit
def modulo_relajacion_superficial(superficie, x, y, nx, ny, temp, t_escala, max_pasos, enable_diffusion):
    """Aplica la difusión de corto alcance por descenso con fronteras periódicas.

    En cada paso, la partícula salta con probabilidad P = min(1, temp / t_escala)
    a uno de sus vecinos de von Neumann estrictamente más bajos, elegido con igual
    probabilidad. Se detiene si no hay vecinos más bajos, si el salto falla o si
    alcanza max_pasos saltos.

    Parámetros
    ----------
    superficie : ndarray de enteros, forma (nx, ny)
        Alturas actuales, en capas. La función no las modifica.
    x, y : int
        Celda de colisión.
    nx, ny : int
        Número de celdas en cada dirección.
    temp : float
        Temperatura del sustrato, en kelvin.
    t_escala : float
        Temperatura de escala de la probabilidad de salto, en kelvin.
    max_pasos : int
        Número máximo de saltos.
    enable_diffusion : bool
        Si es False, la partícula queda en la celda de colisión.

    Devuelve
    --------
    curr_x, curr_y : int
        Índices de la celda que recibe la partícula.
    """
    if not enable_diffusion:
        return x, y

    curr_x = x
    curr_y = y

    # Desplazamientos de von Neumann: izquierda, derecha, abajo, arriba
    dx = (-1, 1, 0, 0)
    dy = (0, 0, -1, 1)

    # Probabilidad de salto según la temperatura
    prob_difusion = temp / t_escala
    if prob_difusion > 1.0:
        prob_difusion = 1.0

    for _ in range(max_pasos):
        curr_h = superficie[curr_x, curr_y]

        # Primer recorrido: número de vecinos estrictamente más bajos, con fronteras periódicas.
        conteo_validos = 0
        for i in range(4):
            nx_c = (curr_x + dx[i]) % nx
            ny_c = (curr_y + dy[i]) % ny
            if nx_c < 0:
                nx_c += nx
            if ny_c < 0:
                ny_c += ny
            if superficie[nx_c, ny_c] < curr_h:
                conteo_validos += 1

        # Sin microvalles en la vecindad, la partícula se deposita.
        if conteo_validos == 0:
            break

        # Intento de salto. Un segundo recorrido localiza al vecino elegido sin
        # crear arreglos temporales en cada paso.
        if np.random.random() < prob_difusion:
            idx = np.random.randint(0, conteo_validos)
            next_x = curr_x
            next_y = curr_y
            contador = 0
            for i in range(4):
                nx_c = (curr_x + dx[i]) % nx
                ny_c = (curr_y + dy[i]) % ny
                if nx_c < 0:
                    nx_c += nx
                if ny_c < 0:
                    ny_c += ny
                if superficie[nx_c, ny_c] < curr_h:
                    if contador == idx:
                        next_x = nx_c
                        next_y = ny_c
                        break
                    contador += 1
            curr_x = next_x
            curr_y = next_y
        else:
            break

    return curr_x, curr_y
```

```python
@njit
def ejecutar_motor_unificado_completo(nx, ny, n_eventos, temp, t_escala, max_pasos, sigma_init, sigma_scat, mfp,
                                      paso_z, distancia_fuente, enable_angle, enable_scattering,
                                      enable_shadowing, enable_diffusion, guardar_cada):
    """Lazo secuencial de deposición Monte Carlo del motor unificado.

    Cada evento elige un punto de mira uniforme, transporta la partícula hasta su celda de
    colisión, la relaja térmicamente y la deposita. Registra el espesor y la rugosidad cada
    ``guardar_cada`` eventos. El generador pseudoaleatorio debe haberse sembrado antes de
    llamar a esta función.

    Devuelve
    --------
    superficie : ndarray int32, forma (nx, ny)
        Alturas finales, en capas.
    hist_ev : ndarray int32
        Número de eventos transcurridos en cada muestreo.
    hist_esp, hist_rug : ndarray float64
        Espesor medio y rugosidad RMS, en capas, en cada muestreo.
    n_colisiones : int
        Número total de colisiones con el gas de todas las partículas.
    """
    superficie = np.zeros((nx, ny), dtype=np.int32)
    tamano_hist = n_eventos // guardar_cada

    hist_ev = np.zeros(tamano_hist, dtype=np.int32)
    hist_esp = np.zeros(tamano_hist, dtype=np.float64)
    hist_rug = np.zeros(tamano_hist, dtype=np.float64)

    idx_h = 0
    max_h_actual = 0
    n_colisiones = 0
    for h in range(1, n_eventos + 1):
        # Punto de mira, equiprobable e independiente de la altura actual.
        x_i = np.random.randint(0, nx)
        y_i = np.random.randint(0, ny)

        # El transporte y la relajación dependen del relieve que dejaron los eventos anteriores,
        # por lo que los eventos deben procesarse en orden.
        x_col, y_col, n_col = modulo_transporte_fase_gaseosa(
            x_i, y_i, nx, ny, superficie, max_h_actual, sigma_init, sigma_scat, mfp, paso_z,
            distancia_fuente, enable_angle, enable_scattering, enable_shadowing
        )
        n_colisiones += n_col
        x_f, y_f = modulo_relajacion_superficial(
            superficie, x_col, y_col, nx, ny, temp, t_escala, max_pasos, enable_diffusion
        )
        superficie[x_f, y_f] += 1

        # Las alturas solo crecen: basta comparar con la celda que acaba de recibir la partícula.
        if superficie[x_f, y_f] > max_h_actual:
            max_h_actual = superficie[x_f, y_f]

        if h % guardar_cada == 0:
            esp, rug = calcular_estadisticos_superficie(superficie, nx, ny)
            hist_ev[idx_h] = h
            hist_esp[idx_h] = esp
            hist_rug[idx_h] = rug
            idx_h += 1

    return superficie, hist_ev, hist_esp, hist_rug, n_colisiones


@njit(parallel=True)
def ejecutar_suite_replicas(semillas, nx, ny, n_eventos, temp, t_escala, max_pasos, sigma_init, sigma_scat, mfp,
                            paso_z, distancia_fuente, enable_angle, enable_scattering, enable_shadowing,
                            enable_diffusion, guardar_cada):
    """Ejecuta en paralelo réplicas independientes del motor unificado para un mismo modelo.

    Las réplicas no comparten estado (cada una tiene su propia superficie y su propia
    semilla), de modo que el bucle es genuinamente paralelizable. Cada réplica siembra el
    generador de números aleatorios del hilo que la ejecuta con ``semillas[rep]`` antes de
    correr, por lo que su resultado depende únicamente de su semilla y no del orden ni del
    hilo en que se ejecute. Dentro de cada réplica los eventos son secuenciales.

    Parámetros
    ----------
    semillas : ndarray de enteros, forma (n_replicas,)
        Semilla de cada réplica.

    El resto de los parámetros son los de ``ejecutar_motor_unificado_completo``.

    Devuelve
    --------
    superficies : ndarray int32, forma (n_replicas, nx, ny)
    historias_espesor, historias_rugosidad : ndarray float64, forma (n_replicas, n_eventos // guardar_cada)
    espesores_finales, rugosidades_finales : ndarray float64, forma (n_replicas,)
    maximos_finales, minimos_finales : ndarray int32, forma (n_replicas,)
    colisiones_por_atomo : ndarray float64, forma (n_replicas,)
        Número medio de colisiones con el gas por partícula depositada.
    """
    n_replicas = semillas.shape[0]
    tamano_hist = n_eventos // guardar_cada

    superficies = np.zeros((n_replicas, nx, ny), dtype=np.int32)
    historias_espesor = np.zeros((n_replicas, tamano_hist), dtype=np.float64)
    historias_rugosidad = np.zeros((n_replicas, tamano_hist), dtype=np.float64)
    espesores_finales = np.zeros(n_replicas, dtype=np.float64)
    rugosidades_finales = np.zeros(n_replicas, dtype=np.float64)
    maximos_finales = np.zeros(n_replicas, dtype=np.int32)
    minimos_finales = np.zeros(n_replicas, dtype=np.int32)
    colisiones_por_atomo = np.zeros(n_replicas, dtype=np.float64)

    for rep in prange(n_replicas):
        np.random.seed(semillas[rep])

        superficie, _, h_esp, h_rug, n_col = ejecutar_motor_unificado_completo(
            nx, ny, n_eventos, temp, t_escala, max_pasos, sigma_init, sigma_scat, mfp, paso_z,
            distancia_fuente, enable_angle, enable_scattering, enable_shadowing, enable_diffusion,
            guardar_cada
        )

        superficies[rep] = superficie
        historias_espesor[rep, :] = h_esp
        historias_rugosidad[rep, :] = h_rug

        esp_f, rug_f = calcular_estadisticos_superficie(superficie, nx, ny)
        espesores_finales[rep] = esp_f
        rugosidades_finales[rep] = rug_f
        maximos_finales[rep] = superficie.max()
        minimos_finales[rep] = superficie.min()
        colisiones_por_atomo[rep] = n_col / n_eventos

    return (superficies, historias_espesor, historias_rugosidad, espesores_finales,
            rugosidades_finales, maximos_finales, minimos_finales, colisiones_por_atomo)
```

## 9. Orquestación y validación de las réplicas

`ejecutar_modelo` es la función de nivel Python que prepara y valida cada corrida antes de invocar el motor compilado. Deriva las semillas de las réplicas con `SeedSequence`, a partir de una semilla raíz común y del nombre del modelo, de modo que cada modelo recorre un flujo de semillas propio e independiente. También comprueba, para cada réplica, que la suma de alturas coincide con el número de eventos: es una verificación exacta de la conservación de partículas, no estadística, y se repite de forma independiente en la sección 10 sobre las superficies ya almacenadas.

La función rechaza explícitamente una combinación de banderas que activaría la dispersión gaseosa sin el sombreado, porque `modulo_transporte_fase_gaseosa` solo traza el vuelo paso a paso cuando el sombreado o la dispersión están activos: sin sombreado, el cálculo analítico directo ignoraría la bandera de dispersión sin advertirlo. Ninguno de los seis modelos de la sección 10 activa esa combinación, pero la comprobación evita que una configuración futura falle en silencio.

```python
def ejecutar_modelo(nombre, flags, n_eventos, n_replicas, semilla_raiz):
    """Ejecuta la suite de réplicas de un modelo y valida sus banderas y su conservación de partículas.

    Deriva las semillas de las réplicas con ``SeedSequence`` a partir de ``semilla_raiz``, de
    modo que cada modelo tiene su propio flujo de semillas independiente y reproducible,
    y delega el cómputo en ``ejecutar_suite_replicas``. La dispersión gaseosa solo tiene efecto
    si el sombreado está activo, porque el módulo de transporte solo traza el vuelo paso a paso
    cuando hay sombreado o dispersión (sección 5); por eso se rechaza aquí una combinación que
    activaría `enable_scattering` sin que el vuelo se trace.

    Parámetros
    ----------
    nombre : str
        Identificador del modelo, usado solo para los mensajes de progreso.
    flags : dict
        Banderas físicas (`enable_angle`, `enable_scattering`, `enable_shadowing`,
        `enable_diffusion`) y la temperatura `temp` del modelo.
    n_eventos : int
        Partículas depositadas por réplica.
    n_replicas : int
        Número de réplicas independientes.
    semilla_raiz : int
        Semilla de la que se derivan las semillas de las réplicas.

    Devuelve
    --------
    dict con las superficies, las historias temporales, los promedios, las desviaciones,
    los errores estándar, los intervalos de confianza y las métricas finales de todas las réplicas.
    """
    if flags["enable_scattering"] and not flags["enable_shadowing"]:
        raise ValueError(f"{nombre}: la dispersión gaseosa requiere el sombreado activo.")

    semillas = np.random.SeedSequence([semilla_raiz, hash(nombre) & 0xFFFFFFFF]).generate_state(n_replicas)

    t_i = time.time()
    (superficies, historias_espesor, historias_rugosidad, espesores_finales, rugosidades_finales,
     maximos_finales, minimos_finales, colisiones_por_atomo) = ejecutar_suite_replicas(
        semillas, NX, NY, n_eventos, flags["temp"], T_ESCALA, MAX_PASOS_DIFUSION,
        SIGMA_INCIDENCIA_INIT, SIGMA_SCATTERING_GAS, MFP_PRESION_INTERMEDIA, PASO_Z, SOURCE_DISTANCE,
        flags["enable_angle"], flags["enable_scattering"], flags["enable_shadowing"],
        flags["enable_diffusion"], GUARDAR_CADA
    )

    # La conservación del número de partículas debe cumplirse en cada réplica, con
    # independencia de las banderas activas: es una comprobación exacta, no estadística.
    if np.any(espesores_finales * (NX * NY) != n_eventos):
        raise RuntimeError(f"{nombre}: la suma de alturas no coincide con n_eventos en alguna réplica.")

    n = n_replicas
    t_critico = stats.t.ppf(1 - (1 - NIVEL_CONFIANZA) / 2, df=n - 1)

    std_espesor = np.std(historias_espesor, axis=0, ddof=1)
    std_rugosidad = np.std(historias_rugosidad, axis=0, ddof=1)
    error_rugosidad = std_rugosidad / np.sqrt(n)

    print(f" -> {nombre}: {n_replicas} réplicas en {time.time() - t_i:.1f} s")

    historial_eventos = np.arange(GUARDAR_CADA, n_eventos + 1, GUARDAR_CADA, dtype=np.int32)
    return {
        "superficie": superficies[0],
        "historial_eventos": historial_eventos,
        "historias_espesor": historias_espesor,
        "historias_rugosidad": historias_rugosidad,
        "espesor_promedio": np.mean(historias_espesor, axis=0),
        "rugosidad_promedio": np.mean(historias_rugosidad, axis=0),
        "std_espesor": std_espesor,
        "std_rugosidad": std_rugosidad,
        "error_rugosidad": error_rugosidad,
        "ic_rugosidad": t_critico * error_rugosidad,
        "espesores_finales": espesores_finales,
        "rugosidades_finales": rugosidades_finales,
        "maximos_finales": maximos_finales,
        "minimos_finales": minimos_finales,
        "colisiones_por_atomo": colisiones_por_atomo,
    }
```

## 10. Configuración de los modelos

Los seis modelos forman una jerarquía de mecanismos, cada uno construido sobre el anterior. M1 no activa ningún mecanismo (deposición aleatoria). M2 y M3 añaden solo la relajación térmica, a 300 K y 400 K respectivamente, para separar el efecto de la relajación en sí del de la temperatura. M4 añade la incidencia angular a 400 K, sin sombreado. M5 añade el sombreado geométrico (transporte balístico con relieve). M6 añade la dispersión en fase gaseosa al camino libre medio `MFP_PRESION_INTERMEDIA`.

```python
# Cada modelo activa un subconjunto de mecanismos sobre el mismo motor. Todos comparten
# fronteras periódicas y, cuando la difusión está activa, la misma regla de relajación
# térmica de los cuadernos 3 a 6 (salto a un vecino de von Neumann más bajo con
# probabilidad P(T) = min(1, T / T_ESCALA)); esta suite no reproduce la relajación
# instantánea y sin temperatura del cuaderno 2 ni las fronteras abiertas del cuaderno 4.
CONFIGURACIONES_MODELOS = {
    "M1: Deposición aleatoria": {
        "enable_angle": False, "enable_scattering": False, "enable_shadowing": False,
        "enable_diffusion": False, "temp": 0,
    },
    "M2: Relajación térmica (300 K)": {
        "enable_angle": False, "enable_scattering": False, "enable_shadowing": False,
        "enable_diffusion": True, "temp": TEMP_MEDIA,
    },
    "M3: Relajación térmica (400 K)": {
        "enable_angle": False, "enable_scattering": False, "enable_shadowing": False,
        "enable_diffusion": True, "temp": TEMP_ALTA,
    },
    "M4: Incidencia angular": {
        "enable_angle": True, "enable_scattering": False, "enable_shadowing": False,
        "enable_diffusion": True, "temp": TEMP_ALTA,
    },
    "M5: Sombreado geométrico": {
        "enable_angle": True, "enable_scattering": False, "enable_shadowing": True,
        "enable_diffusion": True, "temp": TEMP_ALTA,
    },
    "M6: Dispersión en fase gaseosa": {
        "enable_angle": True, "enable_scattering": True, "enable_shadowing": True,
        "enable_diffusion": True, "temp": TEMP_ALTA,
    },
}
```

## 11. Ejecución de la suite estadística

Cada modelo se ejecuta con `REPLICAS_POR_MODELO` réplicas independientes. La tabla siguiente resume, para cada uno, el espesor medio (idéntico en todas las réplicas por conservación de partículas), la rugosidad final promedio con su desviación entre réplicas y su intervalo de confianza, las alturas extremas y, cuando corresponde, el número medio de colisiones con el gas por átomo depositado.

```python
RESULTADOS_MODELOS = {}
tabla_metricas_finales = []

print(f"Ejecutando la suite con {REPLICAS_POR_MODELO} réplicas por modelo "
      f"({N_EVENTOS_COMPARATIVA} eventos cada una)...")
t_inicio_suite = time.time()

for nombre, flags in CONFIGURACIONES_MODELOS.items():
    datos = ejecutar_modelo(nombre, flags, N_EVENTOS_COMPARATIVA, REPLICAS_POR_MODELO, SEMILLA)
    RESULTADOS_MODELOS[nombre] = datos

    n = REPLICAS_POR_MODELO
    t_critico = stats.t.ppf(1 - (1 - NIVEL_CONFIANZA) / 2, df=n - 1)
    rug = datos["rugosidades_finales"]
    tabla_metricas_finales.append({
        "Modelo": nombre,
        "Espesor medio (capas)": np.mean(datos["espesores_finales"]),
        "Rugosidad media W (capas)": np.mean(rug),
        "Desv. entre réplicas": np.std(rug, ddof=1),
        f"IC{100*NIVEL_CONFIANZA:.0f}% de W": t_critico * np.std(rug, ddof=1) / np.sqrt(n),
        "Pico máximo medio (capas)": np.mean(datos["maximos_finales"]),
        "Valle mínimo medio (capas)": np.mean(datos["minimos_finales"]),
        "Colisiones por átomo": np.mean(datos["colisiones_por_atomo"]),
    })

print(f"\nSuite completada en {time.time() - t_inicio_suite:.1f} s.")

df_comparativo = pd.DataFrame(tabla_metricas_finales)
display(df_comparativo.round(4))
```

## 12. Validación del motor unificado

Se contrastan tres límites del modelo. Primero, la conservación de partículas en todas las réplicas de todos los modelos, ya comprobada durante la ejecución (sección 9) y repetida aquí sobre las superficies almacenadas. Segundo, el límite de deposición aleatoria: M1 no activa ningún mecanismo, de modo que sus alturas deben seguir una distribución de Poisson de media $\overline{h}_{\text{teo}} = N / (N_x N_y)$, con rugosidad $W_{\text{RD}} = \sqrt{\overline{h}_{\text{teo}}}$ y una fluctuación de la media de las réplicas de $\sigma_W / \sqrt{2 N_x N_y} / \sqrt{n}$ (cuaderno 1). Tercero, la invariancia de la incidencia angular bajo fronteras periódicas (sección 1): la rugosidad final de M3 y M4 no debería diferir de forma sistemática, lo que se contrasta con una prueba t de Welch para dos muestras independientes, que no supone varianzas iguales entre ambos modelos.

```python
# Validación 1: conservación del número de partículas en todas las réplicas de todos los
# modelos. ejecutar_modelo ya la comprueba al terminar cada modelo (sección 7); aquí se
# repite sobre las superficies almacenadas como comprobación independiente de que ninguna
# réplica se sobrescribió al guardar los resultados.
conservacion_ok = all(
    np.array_equal(datos["superficie"].sum(), N_EVENTOS_COMPARATIVA)
    for datos in RESULTADOS_MODELOS.values()
)

# Validación 2: límite de deposición aleatoria. M1 no tiene difusión, dispersión angular
# ni sombreado, de modo que sus alturas deben seguir una distribución de Poisson de media
# h_teo = N / M, con rugosidad W_RD = sqrt(h_teo) y una fluctuación de la media de las
# réplicas de sigma_W / sqrt(n_replicas), donde sigma_W = W_RD / sqrt(2M) (cuaderno 1).
espesor_teo = N_EVENTOS_COMPARATIVA / (NX * NY)
w_rd_teo = np.sqrt(espesor_teo)
w_rd_sim = np.mean(RESULTADOS_MODELOS["M1: Deposición aleatoria"]["rugosidades_finales"])
error_medio_rd = (w_rd_teo / np.sqrt(2 * NX * NY)) / np.sqrt(REPLICAS_POR_MODELO)
z_rd = (w_rd_sim - w_rd_teo) / error_medio_rd

# Validación 3: equivalencia entre M3 y M4 bajo fronteras periódicas. Sobre una malla
# periódica, sumar a una posición uniforme un desplazamiento lateral aleatorio
# independiente da de nuevo una posición uniforme (sección 2.1): la incidencia angular no
# debería alterar, en promedio, la rugosidad final frente al modelo sin ella. Se contrasta
# con la prueba t de Welch para dos muestras independientes, que no supone varianzas
# iguales entre M3 y M4.
rug_m3 = RESULTADOS_MODELOS["M3: Relajación térmica (400 K)"]["rugosidades_finales"]
rug_m4 = RESULTADOS_MODELOS["M4: Incidencia angular"]["rugosidades_finales"]
t_stat, p_valor_m3m4 = stats.ttest_ind(rug_m3, rug_m4, equal_var=False)

ANCHO = 70


def fila(etiqueta, valor):
    print(f"{etiqueta:<44}: {valor}")


print("=" * ANCHO)
print("VALIDACIONES DEL MOTOR UNIFICADO".center(ANCHO))
print("=" * ANCHO)
fila("Conservación de partículas (todos los modelos)", "correcta" if conservacion_ok else "FALLA")
print("-" * ANCHO)
fila("Rugosidad RD teórica (sqrt(h))", f"{w_rd_teo:.4f} capas")
fila("Rugosidad simulada (M1, media de réplicas)", f"{w_rd_sim:.4f} capas")
fila("Desviación z frente a RD", f"{z_rd:+.2f}")
print("-" * ANCHO)
fila("Rugosidad media M3 (sin incidencia angular)", f"{np.mean(rug_m3):.4f} capas")
fila("Rugosidad media M4 (con incidencia angular)", f"{np.mean(rug_m4):.4f} capas")
fila("Prueba t de Welch, M3 frente a M4", f"t = {t_stat:+.2f}, p = {p_valor_m3m4:.3f}")
print("=" * ANCHO)
if not conservacion_ok:
    raise RuntimeError("Fallo de conservación de partículas: revisar el motor unificado.")
if abs(z_rd) <= 3:
    print("M1 es compatible con la deposición aleatoria (|z| <= 3).")
else:
    print("M1 se aparta de la deposición aleatoria (|z| > 3): revisar el generador y el algoritmo.")
if p_valor_m3m4 > ALFA_PRUEBA:
    print(f"No hay evidencia de diferencia entre M3 y M4 al nivel alfa = {ALFA_PRUEBA} "
          "(compatible con fronteras periódicas).")
else:
    print(f"M3 y M4 difieren de forma significativa al nivel alfa = {ALFA_PRUEBA}: revisar el "
          "argumento de invariancia bajo fronteras periódicas.")
```

## 15. Ley de Family-Vicsek: extracción del exponente de crecimiento

Para cada modelo se ajusta $\beta$ por separado en dos intervalos de espesor fijados de antemano (`VENTANAS_AJUSTE`): uno temprano, que excluye el transitorio de muy baja cobertura, y uno tardío, más cercano al espesor final de la suite. El ajuste se hace réplica por réplica y después se promedia, de modo que el intervalo de confianza de $\beta$ refleja la dispersión entre réplicas independientes y no la incertidumbre de una única regresión.

```python
def beta_ventana(historia_h, historia_w, h_min, h_max):
    """Pendiente beta de la regresión log(W) frente a log(h) en la ventana [h_min, h_max).

    La ventana se fija de antemano en capas (VENTANAS_AJUSTE), en lugar de buscar la
    posición que maximiza el coeficiente de determinación: maximizar R^2 sobre una
    ventana deslizante favorece intervalos cortos que, por azar en una única
    realización, parecen más lineales, sin relación necesaria con el régimen físico de
    crecimiento, y además puede seleccionar una ventana distinta en cada réplica, lo que
    impide promediar beta de forma consistente entre ellas.
    """
    mascara = (historia_h >= h_min) & (historia_h < h_max)
    if mascara.sum() < 3:
        return np.nan
    pendiente, _, _, _, _ = stats.linregress(np.log(historia_h[mascara]), np.log(historia_w[mascara]))
    return pendiente


tabla_escalamiento_fv = []
for nombre, datos in RESULTADOS_MODELOS.items():
    historias_h = datos["historias_espesor"]
    historias_w = datos["historias_rugosidad"]
    delta_h = np.mean(datos["maximos_finales"]) - np.mean(datos["minimos_finales"])

    fila_resultado = {"Modelo": nombre, "Rugosidad final (capas)": np.mean(datos["rugosidades_finales"]),
                       "Rango pico-valle (capas)": delta_h}
    for h_min, h_max in VENTANAS_AJUSTE:
        betas_replica = np.array([
            beta_ventana(h, w, h_min, h_max) for h, w in zip(historias_h, historias_w)
        ])
        betas_validos = betas_replica[~np.isnan(betas_replica)]
        n_validos = len(betas_validos)
        beta_media = np.mean(betas_validos) if n_validos > 0 else np.nan
        error_beta = np.std(betas_validos, ddof=1) / np.sqrt(n_validos) if n_validos > 1 else np.nan
        ic_beta = stats.t.ppf(1 - (1 - NIVEL_CONFIANZA) / 2, df=n_validos - 1) * error_beta if n_validos > 1 else np.nan
        fila_resultado[f"beta [{h_min:g}, {h_max:g})"] = beta_media
        fila_resultado[f"IC{100*NIVEL_CONFIANZA:.0f}% beta [{h_min:g}, {h_max:g})"] = ic_beta
    tabla_escalamiento_fv.append(fila_resultado)

df_fv = pd.DataFrame(tabla_escalamiento_fv)
display(df_fv.round(4))
```

```python
# Ley de Family-Vicsek: rugosidad frente al espesor en escala log-log, con la recta de
# ajuste de la ventana tardía (VENTANAS_AJUSTE[-1]) superpuesta sobre la curva de cada modelo
h_min_tardio, h_max_tardio = VENTANAS_AJUSTE[-1]

fig, ax = plt.subplots(figsize=(9, 5.5))
for (nombre, datos), color, fila_beta in zip(RESULTADOS_MODELOS.items(), colores, tabla_escalamiento_fv):
    h = datos["espesor_promedio"]
    w = datos["rugosidad_promedio"]
    ic = datos["ic_rugosidad"]
    ax.errorbar(h, w, yerr=ic, fmt="o", ms=3, alpha=0.35, color=color)

    beta_tardio = fila_beta[f"beta [{h_min_tardio:g}, {h_max_tardio:g})"]
    mascara = (h >= h_min_tardio) & (h < h_max_tardio)
    if mascara.sum() >= 3 and not np.isnan(beta_tardio):
        h_ventana = h[mascara]
        _, intercepto, _, _, _ = stats.linregress(np.log(h_ventana), np.log(w[mascara]))
        ax.loglog(h_ventana, np.exp(intercepto) * h_ventana ** beta_tardio, lw=2.2, color=color,
                  label=rf"{nombre}: $\beta$ = {beta_tardio:.3f}")

ax.set_title(rf"Ley de Family-Vicsek: ajuste en la ventana $\overline{{h}} \in [{h_min_tardio:g}, {h_max_tardio:g})$",
             fontweight="bold")
ax.set_xlabel(r"Espesor medio ($\overline{h}$, capas)")
ax.set_ylabel(r"Rugosidad RMS ($W$, capas)")
ax.grid(True, which="both", linestyle=":", alpha=0.5)
ax.legend(bbox_to_anchor=(1.02, 1), loc="upper left", fontsize=8)
fig.tight_layout()
fig.savefig("unificado_family_vicsek.png", bbox_inches="tight")
plt.show()
```

## 16. Análisis de resultados

### 16.1 Validación
Las tres comprobaciones de la sección 12 se cumplen. La conservación de partículas es exacta en las 15 réplicas de los seis modelos. M1 da una rugosidad de 1.1186 capas frente a la referencia teórica $\sqrt{1.25} = 1.1180$ (z = +0.55), compatible con la deposición aleatoria. La rugosidad final de M3 y M4 (0.6280 y 0.6279 capas) es prácticamente idéntica, y la prueba t de Welch no encuentra diferencia significativa (p = 0.91, muy por encima de $\alpha = 0.01$): bajo fronteras periódicas, la incidencia angular sin sombreado no altera la rugosidad global, como predice el argumento de la sección 1. El exponente de M1 se mantiene en 0.50 en ambas ventanas de ajuste (0.501 y 0.501), el resultado esperado para deposición aleatoria en cualquier tramo de su crecimiento y una validación adicional, independiente, del propio método de extracción de $\beta$.

Como referencia cruzada con el cuaderno anterior, M5 (sombreado geométrico, sin dispersión, $\sigma_\theta = 15°$, $T = 400$ K, 50 000 eventos) da una rugosidad de 0.6536 capas, muy cercana a los 0.653-0.656 capas del control balístico del cuaderno 6 bajo condiciones equivalentes.

### 16.2 Jerarquía de mecanismos
La rugosidad final ordena los modelos como M3 $\approx$ M4 (0.628) < M5 (0.654) < M6 $\approx$ M2 (0.74-0.75) < M1 (1.12). La relajación térmica es, con diferencia, el mecanismo más influyente del conjunto: pasar de la deposición aleatoria (M1) a la relajación a 400 K (M3) reduce la rugosidad en un 44 %. Dentro de la relajación térmica, la temperatura importa: a 300 K (M2) la rugosidad es un 19 % mayor que a 400 K (M3), porque la probabilidad de salto $P(T) = \min(1, T/T_1)$ es menor y una fracción mayor de los adátomos queda atrapada en su celda de llegada.

Añadir el sombreado geométrico a la relajación de 400 K (M5 frente a M3) sube la rugosidad un 4 %: en este régimen de espesores, la relajación térmica compensa la mayor parte, aunque no la totalidad, de la retroalimentación positiva del sombreado. Añadir además la dispersión en fase gaseosa (M6 frente a M5) sube la rugosidad otro 13 %, hasta 0.739 capas, un efecto mucho mayor que el del sombreado por sí solo. En términos comparables, ese aumento equivale aproximadamente al que produce enfriar el sustrato de 400 a 300 K sin sombreado ni gas (M2): dos mecanismos de naturaleza distinta, relajación más lenta y dispersión gaseosa, terminan produciendo una rugosidad similar. Cada átomo de M6 sufre en promedio 7.23 colisiones con el gas, bastante más que las $L/\lambda = 4$ de un vuelo vertical sobre una superficie plana, porque su recorrido real es oblicuo y más largo, igual que se observó en el cuaderno 6.

### 16.3 Exponente de crecimiento
El exponente $\beta$ refuerza la misma jerarquía. En la ventana tardía ($\overline{h} \in [0.75, 1.25)$), M1 da 0.501, muy cerca del valor teórico de deposición aleatoria. Los modelos con relajación y sin sombreado ni dispersión (M3 y M4) dan los valores más bajos, 0.116 y 0.117, cercanos al valor de referencia de la clase Edwards-Wilkinson en dos dimensiones ($\beta \approx 0.24$, aunque en un régimen de crecimiento distinto: ver la sección 3). M2 (relajación a 300 K) da un valor intermedio, 0.219, coherente con una relajación menos eficiente. M5 (sombreado) sube a 0.157, y M6 (dispersión) sube más, a 0.236, el segundo valor más alto de la tabla después de M1. En todos los modelos con mecanismos activos, $\beta$ decrece de la ventana temprana a la tardía, señal de que la rugosidad se aleja del crecimiento sin límite de RD a medida que la película engrosa.

### 16.4 Interpretación física
La dispersión en fase gaseosa no actúa, en este modelo, como un mecanismo de suavizado. Cada colisión ensancha la distribución angular con la que los átomos llegan a la película; cuanto más oblicua es esa llegada, más flujo interceptan las regiones ya elevadas y menos alcanza los valles vecinos, el mismo mecanismo de sombreado reforzado que ya se cuantificó en el cuaderno 6. En este cuaderno esa reafirmación es visible incluso con la relajación térmica más eficiente de la serie (400 K): el aumento de rugosidad de M5 a M6 (13 %) es mayor que el de M3 a M5 (4 %), de modo que, en el rango de presiones estudiado, la dispersión gaseosa pesa más que el sombreado puro sobre la morfología final.

### 16.5 Alcance y limitaciones
Los resultados corresponden a espesores tempranos (hasta 1.25 capas) sobre una malla de tamaño finito, y no verifican los valores asintóticos de la literatura para las clases de universalidad (sección 3). El motor comparte, en los seis modelos, la relajación térmica activada de los cuadernos 3 a 6, y no la relajación instantánea del cuaderno 2 ni las fronteras abiertas del cuaderno 4, de modo que las comparaciones con esos dos cuadernos son solo cualitativas. $T_1$ sigue sin calibrar frente a un sustrato real. El número de réplicas (15) y la reducción de escala respecto a los parámetros originalmente previstos para esta suite responden a que la ejecución se realizó en Python puro, sin Numba disponible en este entorno; con Numba, los mismos modelos podrían ejecutarse con más réplicas o a mayor espesor en un tiempo comparable al de esta corrida.

## 17. Conclusiones

* El motor unificado integra correctamente los cuatro mecanismos de los cuadernos 3 a 6 bajo un mismo esquema de fronteras periódicas y relajación térmica activada, y supera las tres validaciones planteadas: conservación de partículas, límite de deposición aleatoria e invariancia de la incidencia angular bajo fronteras periódicas (M3 frente a M4, p = 0.91). La rugosidad de M5 coincide, dentro de un margen pequeño, con el control balístico equivalente del cuaderno 6, lo que respalda la coherencia física del motor a través de los dos últimos cuadernos de la serie.
* La relajación térmica es el mecanismo dominante: por sí sola reduce la rugosidad final un 44 % frente a la deposición aleatoria, y su eficiencia depende de la temperatura (19 % de diferencia entre 300 K y 400 K). El sombreado geométrico y, sobre todo, la dispersión en fase gaseosa actúan en sentido contrario: la dispersión gaseosa añade más rugosidad que el sombreado puro, porque ensancha la distribución angular de llegada y refuerza, en lugar de contrarrestar, el sombreado ya presente.
* El exponente de crecimiento $\beta$, estimado sobre ventanas de espesor fijadas de antemano, ordena los modelos de forma consistente con su rugosidad final, y decrece con el espesor en todos los modelos que activan algún mecanismo, salvo en la deposición aleatoria, donde se mantiene en 0.50 como predice la teoría. Esto es coherente con un régimen de crecimiento temprano, aún no saturado, y no debe interpretarse como una medición de los exponentes asintóticos de la literatura.
