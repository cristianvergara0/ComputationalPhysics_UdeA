## 1. Introducción y objetivos

Los cuadernos 01 a 07 construyen, mecanismo a mecanismo, un modelo de Monte Carlo del crecimiento de una película delgada de cobre por evaporación térmica. Cada cuaderno añade un mecanismo al modelo del anterior, de modo que el modelo M$k$ contiene al M$(k-1)$: con el mecanismo nuevo apagado, da exactamente las mismas superficies. El modelo M7 reúne todos los mecanismos de la serie.

| Modelo | Mecanismo que añade | Cuaderno |
|:--|:--|:--:|
| M1 | Deposición aleatoria: cada átomo cae en vertical sobre una columna elegida al azar | 01 |
| M2 | Relajación por descenso hacia vecinos más bajos, con probabilidad de salto $P = 1$ | 02 |
| M3 | Probabilidad de salto dependiente de la temperatura, $P = \min(1, T/T_{\mathrm{escala}})$, a 400 K | 03 |
| M4 | Incidencia angular: dirección de emisión con una dispersión de $15^\circ$ por eje | 04 |
| M5 | Sombreado geométrico: el relieve intercepta los vuelos | 05 |
| M6 | Colisiones con el gas de la cámara (camino libre medio de 10 celdas) | 06 |
| M7 | Adhesión lateral: los átomos que alcanzan la pared de una columna quedan adheridos y dejan huecos | 07 |

Cada cuaderno estudia su mecanismo con detalle. Este reúne los siete modelos para responder a la pregunta que da sentido a la cadena: si cada modelo es un avance sobre el anterior, es decir, si el mecanismo que añade cambia la película de forma apreciable y en qué régimen. Las respuestas no son triviales. Un mecanismo puede no tener efecto por sí solo y ser necesario para otro: la incidencia angular, por ejemplo, solo cambia la película a través del sombreado. Y el efecto de cada mecanismo depende de los que ya están presentes. Por eso la comparación de la cadena se completa con un diseño factorial de los cinco mecanismos de transporte y de incorporación.

**Objetivos**

* Comprobar en un mismo cuaderno que cada modelo de la cadena contiene al anterior.
* Simular los siete modelos en las mismas condiciones, hasta 256 capas, y comparar su rugosidad, su relieve, su estructura y su escalamiento.
* Medir el efecto de cada mecanismo sobre el modelo anterior en cada espesor y decidir, con un criterio fijado de antemano, si cada modelo es un avance.
* Medir con un diseño factorial $2^5$ los efectos de la relajación, la incidencia angular, el sombreado, el gas y la adhesión lateral, y sus interacciones, sobre la rugosidad, el relieve y la porosidad.
* Analizar por separado la mitad del diseño con relajación, la de los modelos de la cadena a partir de M3.

## 2. Marco teórico

### 2.1 Escalamiento dinámico y clases de universalidad

La rugosidad de una película que crece sobre un sustrato de lado $L$ sigue la forma de Family y Vicsek (1985): crece como $W \sim \overline{h}^{\beta}$ hasta saturarse en $W_{\mathrm{sat}} \sim L^{\alpha}$. El exponente de crecimiento distingue las clases de universalidad (Barabási & Stanley, 1995). En la deposición aleatoria (M1) las columnas son independientes y $W = \overline{h}^{1/2}$, sin saturación. La relajación hacia vecinos más bajos (M2 a M4) lleva a la clase de Edwards y Wilkinson (1982), que en dos dimensiones de sustrato tiene $\beta = 0$: la rugosidad crece de forma logarítmica, $W^2 = a \ln \overline{h} + b$. La deposición balística, con adhesión lateral, pertenece a la clase de Kardar, Parisi y Zhang (Kardar et al., 1986; Meakin et al., 1986), con $\beta \approx 0.24$ en dos dimensiones de sustrato (Barabási & Stanley, 1995). El sombreado no tiene una clase propia en estos modelos: produce la inestabilidad de columnas de los cuadernos 05 y 06, en la que el relieve crece de forma aproximadamente exponencial.

El exponente efectivo $\beta_{\mathrm{ef}} = d\ln W / d\ln\overline{h}$, medido en ventanas de espesor, resume el régimen de cada modelo: cerca de $1/2$ sin correlaciones laterales, cerca de cero con relajación eficaz y grande cuando domina la inestabilidad.

### 2.2 El efecto de cada mecanismo en la cadena

El efecto del mecanismo $k$ sobre la cadena es el cociente $W(\mathrm{M}k)/W(\mathrm{M}(k-1))$ a un mismo espesor, con los mismos parámetros y con las réplicas de cada modelo. Es un efecto condicionado: mide lo que cambia el mecanismo $k$ cuando están presentes todos los anteriores. La cadena es un camino concreto entre todas las combinaciones de mecanismos, y otro orden daría otros efectos si los mecanismos interactúan.

### 2.3 Diseño factorial

En un diseño factorial $2^k$ completo, cada factor toma un nivel bajo y uno alto y se simulan las $2^k$ combinaciones con $R$ réplicas cada una (Montgomery, 2017; Box et al., 2005). Con los niveles codificados $x_f = \pm 1$, el efecto de un subconjunto $\mathcal{S}$ de factores es el contraste

$$E_{\mathcal{S}} = \frac{1}{2^{k-1}} \sum_{\text{celdas}} \Bigl(\prod_{f \in \mathcal{S}} x_f\Bigr)\,\bar y_{\text{celda}}.$$

Para un factor, es la diferencia entre la respuesta media con el factor alto y con el bajo, promediada sobre los demás; para dos, la mitad de la diferencia entre el efecto de uno con el otro alto y con el otro bajo. La varianza de cada efecto se obtiene de la varianza de cada celda sin suponerla igual en todas, y los grados de libertad con la aproximación de Satterthwaite (1946). Se analizan tres respuestas. Con $y = \ln W$ y con $y = \ln R$, el logaritmo del relieve $R = h_{\max} - \langle h \rangle$, $e^{E}$ es el factor por el que cambia la respuesta al pasar del nivel bajo al alto. La porosidad $p$ se analiza sin transformar, porque es exactamente nula en muchas celdas (sección 2.4), y su efecto es una diferencia de fracciones de volumen.

Los cinco factores son la relajación (D: $P = 0$ o $P = 0.8$), la incidencia angular (A: $0$ o $15^\circ$), el sombreado (S), el gas (G: sin gas o $\lambda = 10$ celdas) y la adhesión lateral (L). La celda con todos los factores bajos es M1, y la cadena M3, M4, M5, M6 y M7 recorre las celdas en que se activan, en ese orden, D, A, S, G y L. M2 no está en el diseño, porque difiere de M3 solo en el valor de $P$.

### 2.4 El sombreado como mecanismo habilitante

Si el relieve no intercepta los vuelos, la columna que recibe cada átomo es la que cruza su trayectoria a la altura de la columna más alta. Con fronteras periódicas y puntos de mira uniformes, esa columna es uniforme en la malla sea cual sea la dirección de llegada, de modo que la película no depende de la incidencia angular, del gas ni de la adhesión lateral, que solo actúa cuando un vuelo es interceptado por una pared. Y si los vuelos son verticales (sin incidencia angular ni gas), el sombreado no cambia nada, porque un vuelo vertical llega a la columna a la que apunta. Así, la rugosidad solo depende de A, G, S y L cuando hay sombreado y alguna fuente de oblicuidad. En el factorial, los efectos de A, G y L dentro de la mitad sin sombreado, y los de S y L dentro de la cuarta parte sin oblicuidad, deben ser nulos. Por la misma razón, la porosidad es nula en todas las celdas sin adhesión lateral, sin sombreado o sin oblicuidad: sin adhesión lateral no hay huecos, y sin sombreado o con vuelos verticales ningún átomo alcanza la pared de una columna.

Sin relajación, la película no tiene ningún mecanismo que se oponga a la inestabilidad de columnas (cuaderno 05), y las celdas con $D = 0$, sombreado y oblicuidad pueden dominar las interacciones de orden alto del diseño completo. Por eso la mitad con relajación ($D = 1$), la de los modelos M3 a M7 de la cadena, se analiza también por separado, como un factorial $2^4$ en A, S, G y L.

### 2.5 Hipótesis y criterios

Los criterios se fijan antes de ejecutar los experimentos:

1. **Contención.** Cada modelo, con su mecanismo apagado y con el nombre del anterior, da exactamente las superficies del anterior; M5 con incidencia vertical da las de M3.
2. **Avance de la cadena.** Para M2, M3, M5, M6 y M7, el cociente con el modelo anterior difiere de uno (Welch, $p < 0.01$) en al menos uno de los espesores guardados que alcanzan todas las réplicas de los dos modelos. Para M4 no difiere en ninguno ($p > 0.01$): la incidencia angular solo actúa a través del sombreado.
3. **Inestabilidad y su supresión.** Todas las réplicas de M5 y de M6 llegan al corte por relieve antes de 256 capas, y ninguna de M7.
4. **Sombreado habilitante.** En el factorial, dentro de la mitad sin sombreado ningún efecto que contenga A, G o L tiene $|t| > 3$, y dentro de la cuarta parte sin oblicuidad (sin incidencia angular ni gas) ningún efecto que contenga S o L tiene $|t| > 3$.
5. **Efectos con sombreado.** Dentro de la mitad con sombreado, los efectos principales de A y de G son positivos ($t > 3$), y el de D es negativo ($t < -3$).
6. **Mitad con relajación.** Dentro de la mitad con relajación, los efectos principales de S y de G en $\ln W$ son positivos ($t > 3$), y el de L es negativo ($t < -3$).
7. **Porosidad.** La porosidad es nula en todas las réplicas de las celdas sin adhesión lateral, sin sombreado o sin incidencia angular ni gas, y positiva en todas las réplicas de las demás.

## 5. Funciones del modelo

Los siete modelos usan el mismo motor (`motor.ejecutar_motor_volumen`, a través de `simular_modelo`), con una bandera por mecanismo: `enable_diffusion` y la temperatura `temp` para la relajación, `enable_angle`, `enable_shadowing`, `enable_scattering` y `enable_lateral`. El diccionario `MODELOS` de la celda anterior define la cadena con esas banderas. Sin sombreado, el gas solo cambia la columna a la que llega el átomo, que es la que cruza su trayectoria a la altura de la columna más alta, y la adhesión lateral no actúa.

El análisis usa `simular_modelo`, `observables_de_instantaneas`, `estadistica_por_instante`, `espesor_de_inicio`, `ajuste_por_replica` y `media_y_se` (módulo `crecimiento`), `contraste_welch`, `cociente_con_error` y `efectos_factorial` (módulo `estadistica`) y `pendiente_local` (módulo `analisis`).

```python
ANCHO = 78
fila = crear_fila(48)


def parametros(L=NX, **cambios):
    """Parámetros del sistema en una malla de L x L celdas, con los cambios indicados."""
    return dataclasses.replace(PARAMS_BASE, nx=L, ny=L, **cambios)


def simular(clave, L, n_replicas, espesor_final, espesores_instantaneas=(), nombre=None, banderas=None, **opciones):
    """Ejecuta un modelo de la cadena en una malla de L x L hasta ``espesor_final`` capas.

    En la malla base el nombre es el del modelo, de modo que las réplicas son las de toda la serie; en otras
    mallas el nombre incluye el tamaño. ``banderas`` sustituye las del modelo y ``opciones`` pasa a
    ``simular_modelo`` (instantes de registro) o, si es un parámetro del sistema, a ``parametros``.
    """
    campos = {f.name for f in dataclasses.fields(PARAMS_BASE)}
    cambios = {k: opciones.pop(k) for k in list(opciones) if k in campos}
    if nombre is None:
        nombre = CLAVES[clave] if L == NX else f"{CLAVES[clave]}, L = {L}"
    return simular_modelo(parametros(L, **cambios), nombre, banderas or MODELOS[CLAVES[clave]], n_replicas, SEMILLA,
                          espesor_final, espesores_instantaneas, max_pasos_rayo=MAX_PASOS_RAYO, **opciones)


def simular_sombreado(clave, L, n_replicas, espesor_final, espesores_instantaneas=(), **opciones):
    """Simula un modelo; con sombreado, aplica el corte por relieve."""
    banderas = opciones.get("banderas") or MODELOS[CLAVES[clave]]
    if banderas["enable_shadowing"]:
        opciones.setdefault("relieve_maximo", RELIEVE_MAXIMO)
    return simular(clave, L, n_replicas, espesor_final, espesores_instantaneas, **opciones)


def p_welch(a, b):
    """Valor p de Welch, o nan si alguna muestra tiene menos de dos valores."""
    a, b = np.asarray(a, dtype=float), np.asarray(b, dtype=float)
    return contraste_welch(a, b)["p"] if a.size > 1 and b.size > 1 else np.nan


def rugosidad_en(obs, modelo, objetivo, columna="rugosidad"):
    """Valores de una columna de los observables en un espesor guardado, por réplica."""
    return obs[(obs["modelo"] == modelo) & (obs["objetivo"] == objetivo)].set_index("replica")[columna]


def rugosidad_guardada(datos, objetivo):
    """Rugosidad de cada réplica en un espesor guardado (todas las réplicas deben alcanzarlo)."""
    k = int(np.flatnonzero(datos["objetivos"] == objetivo)[0])
    return datos["instantaneas"][:, k].reshape(datos["instantaneas"].shape[0], -1).std(axis=1)


def diferencias_pareadas(base, variante):
    """Diferencia relativa media (%) y su IC del 95 % entre dos series pareadas por réplica, con el valor p."""
    comunes = base.index.intersection(variante.index)
    if len(comunes) < 2:
        return {"pares": len(comunes)}
    dif = variante[comunes].to_numpy() - base[comunes].to_numpy()
    media, ic = media_ic95(dif) if np.any(dif != 0) else (0.0, 0.0)
    p = stats.ttest_rel(variante[comunes], base[comunes]).pvalue if np.any(dif != 0) else 1.0
    return {"pares": len(comunes), "base": base[comunes].mean(), "dif. (%)": 100 * media / base[comunes].mean(),
            "IC95 (%)": 100 * ic / base[comunes].mean(), "p (t pareada)": p}


def banderas_factorial(d, a, s, g, l):
    """Banderas de una celda del factorial (niveles 0 o 1 de D, A, S, G y L)."""
    return {"enable_angle": bool(a), "enable_scattering": bool(g), "enable_shadowing": bool(s), "enable_diffusion": bool(d),
            "temp": TEMP_BASE if d else 0, "enable_lateral": bool(l)}


def efectos_que_contienen(tabla, letras):
    """Filas de una tabla de efectos cuyo rótulo contiene alguno de los factores de ``letras``."""
    return tabla[tabla["efecto"].apply(lambda e: any(f in e.split("×") for f in letras))]
```

## 6. Validación

1. **Conservación de partículas.** `simular_modelo` comprueba en cada réplica que la suma de las alturas de las columnas es el número de átomos más el de huecos.
2. **Contención de la cadena.** En una malla de $40 \times 40$ celdas y hasta 8 capas, cada modelo se simula con su mecanismo apagado y con el nombre del anterior, y se comparan las superficies: M2 sin saltos frente a M1; M3 con $T = T_{\mathrm{escala}}$ frente a M2; M4 sin incidencia angular frente a M3; M5 con incidencia vertical frente a M3 (sin sombreado, M5 es M4 por definición); M6 sin gas ($\lambda = \infty$) frente a M5, y M7 sin adhesión lateral frente a M6.

```python
# 6.1 La conservación de partículas la comprueba simular_modelo en cada simulación.
# 6.2 Contención de la cadena: modelo con su mecanismo apagado y con el nombre del anterior
def con_nombre(clave, nombre_de, banderas=None, **cambios):
    referencia = simular(nombre_de, L_CONTENCION, REPLICAS_CONTENCION, ESPESOR_CONTENCION, [ESPESOR_CONTENCION], n_log=8)
    prueba = simular(clave, L_CONTENCION, REPLICAS_CONTENCION, ESPESOR_CONTENCION, [ESPESOR_CONTENCION], n_log=8,
                     nombre=referencia["nombre"], banderas=banderas, **cambios)
    return np.array_equal(referencia["instantaneas"], prueba["instantaneas"])

pasos_contencion = [
    ("M2 sin saltos = M1", con_nombre("M2", "M1", max_pasos_difusion=0)),
    ("M3 con T = T_escala = M2", con_nombre("M3", "M2", banderas=dict(MODELOS[CLAVES["M3"]], temp=T_ESCALA))),
    ("M4 sin incidencia angular = M3", con_nombre("M4", "M3", banderas=dict(MODELOS[CLAVES["M4"]], enable_angle=False))),
    ("M5 con incidencia vertical = M3", con_nombre("M5", "M3", banderas=dict(MODELOS[CLAVES["M5"]], enable_angle=False))),
    ("M6 sin gas = M5", con_nombre("M6", "M5", mfp=np.inf)),
    ("M7 sin adhesión lateral = M6", con_nombre("M7", "M6", banderas=dict(MODELOS[CLAVES["M7"]], enable_lateral=False))),
]
df_contencion = pd.DataFrame(pasos_contencion, columns=["paso", "superficies idénticas"])
contencion_ok = bool(df_contencion["superficies idénticas"].all())
guardar_tabla(df_contencion, "08_contencion")

print("=" * ANCHO)
print("6. VALIDACIÓN: CONTENCIÓN DE LA CADENA")
print("=" * ANCHO)
display(df_contencion)
fila("Cada modelo contiene al anterior", "sí" if contencion_ok else "NO")
```

Se simulan los siete modelos en la malla de $200 \times 200$ celdas con 10 réplicas, hasta 256 capas o hasta que el relieve supera 100 capas. Con los nombres de la cadena, las réplicas son las de los cuadernos 01 a 07.

### 7.2 Experimento II: diseño factorial $2^5$

Se simulan las 32 combinaciones de los niveles de D, A, S, G y L a 2 capas, con 10 réplicas cada una. El factorial se simula sin corte por relieve, para que todas las celdas lleguen a 2 capas y el diseño quede completo, y la tabla de celdas informa del relieve medio de cada una: en las celdas más inestables puede superar, a ese espesor, las 100 capas del corte de los cuadernos 05 a 07.

```python
# Experimento II: diseño factorial 2^5 (relajación, incidencia angular, sombreado, gas, adhesión lateral)
print("Experimento II: diseño factorial...")
filas_fact = []
t0 = time.time()
for codigo in itertools.product((0, 1), repeat=len(FACTORES)):
    nombre = "Factorial (" + ", ".join(f"{f} = {v}" for f, v in zip(FACTORES, codigo)) + ")"
    d = simular("M7", NX, REPLICAS_FACTORIAL, ESPESOR_FACTORIAL, [ESPESOR_FACTORIAL], n_log=10, nombre=nombre,
                banderas=banderas_factorial(*codigo))           # sin corte por relieve: todas las réplicas llegan
    superficies = d["instantaneas"][:, 0].reshape(REPLICAS_FACTORIAL, -1)
    suma = superficies.sum(axis=1, dtype=np.int64)
    for r in range(REPLICAS_FACTORIAL):
        filas_fact.append({**dict(zip(FACTORES, codigo)), "replica": r, "rugosidad": float(superficies[r].std()),
                           "relieve": float(superficies[r].max() - suma[r] / N_CELDAS),
                           "porosidad": float((suma[r] - d["eventos_instantaneas"][0]) / suma[r])})
df_factorial = pd.DataFrame(filas_fact)
guardar_tabla(df_factorial, "08_factorial_replicas")
print(f"Factorial completado: {2 ** len(FACTORES)} celdas en {time.time() - t0:.0f} s.")
```

### 7.3 Resumen cuantitativo

Las celdas siguientes reúnen las cifras con las que se contrastan los criterios de la sección 2.5. La última los resume.

```python
# 7.3.1 Los siete modelos en cada espesor guardado
columnas = ["rugosidad", "relieve", "xi", "escalones", "porosidad"]
resumen_modelos = (obs_largo.groupby(["modelo", "objetivo"])[columnas].agg(["mean", "sem", "count"]))
filas_m = []
for c, d in RESULTADOS.items():
    inicio = espesor_de_inicio(d, RELIEVE_INESTABLE)
    filas_m.append({"modelo": CLAVES[c], "detenidas": int(d["detenida"].sum()), "espesor final medio": float(d["espesor_final"].mean()),
                    "h_20 mediana": mediana_censurada(inicio),
                    "colisiones por átomo": float(np.mean(d["colisiones_por_atomo"])),
                    "porosidad final": float(np.nanmean([d["porosidad"][r][np.isfinite(d["porosidad"][r])][-1] for r in range(REPLICAS)])),
                    "segundos": d["segundos"]})
df_modelos = pd.DataFrame(filas_m)
tabla_W = obs_largo.pivot_table(index="modelo", columns="objetivo", values="rugosidad", aggfunc="mean")
tabla_n = obs_largo.pivot_table(index="modelo", columns="objetivo", values="rugosidad", aggfunc="count")
cortes_ok = bool(RESULTADOS["M5"]["detenida"].all() and RESULTADOS["M6"]["detenida"].all() and not RESULTADOS["M7"]["detenida"].any())
guardar_tabla(df_modelos, "08_modelos")
guardar_tabla(resumen_modelos.reset_index(), "08_resumen_instantaneas")

print("=" * ANCHO)
print("LOS SIETE MODELOS")
print("=" * ANCHO)
display(df_modelos.round(4))
print("Rugosidad media en cada espesor guardado (capas); réplicas que lo alcanzan:")
display(tabla_W.round(4))
display(tabla_n)
fila("M5 y M6 llegan al corte en todas las réplicas; M7 en ninguna", "sí" if cortes_ok else "NO")
```

```python
# 7.3.2 Efecto de cada mecanismo sobre el modelo anterior (cociente de rugosidades y prueba de Welch)
claves = list(CLAVES)
filas_inc = []
for anterior, actual in zip(claves[:-1], claves[1:]):
    for objetivo in ESPESORES_INSTANTANEAS:
        a = rugosidad_en(obs_largo, anterior, objetivo)
        b = rugosidad_en(obs_largo, actual, objetivo)
        if len(a) < REPLICAS or len(b) < REPLICAS:
            continue
        c, se = cociente_con_error(b.mean(), b.sem(), a.mean(), a.sem())
        filas_inc.append({"paso": f"{anterior} -> {actual}", "espesor": objetivo, "W anterior": a.mean(), "W actual": b.mean(),
                          "cociente": c, "SE": se, "p Welch": p_welch(a, b)})
df_incrementos = pd.DataFrame(filas_inc)
avance = df_incrementos.groupby("paso")["p Welch"].min()
nulos = {"M3 -> M4"}
avance_ok = bool(all((avance[p] < 0.01) != (p in nulos) for p in avance.index) and len(avance) == len(claves) - 1)
guardar_tabla(df_incrementos, "08_incrementos")

print("=" * ANCHO)
print("EFECTO DE CADA MECANISMO SOBRE EL MODELO ANTERIOR")
print("=" * ANCHO)
display(df_incrementos.round(4))
display(pd.DataFrame({"menor p de Welch": avance, "avance (p < 0.01)": avance < 0.01}))
fila("Avance en todos los pasos salvo M3 -> M4", "sí" if avance_ok else "NO")
```

```python
# 7.3.3 Exponente de crecimiento efectivo de cada modelo en ventanas de espesor (ajuste por réplica)
filas_b = []
for c, d in RESULTADOS.items():
    h = capas(d)
    for inferior, superior in VENTANAS_BETA:
        pendientes = ajuste_por_replica(np.log(h), np.log(d["rugosidad"]), np.log(inferior), np.log(superior))[:, 0]
        media, se = media_y_se(pendientes)
        filas_b.append({"modelo": CLAVES[c], "ventana": f"{inferior:g}-{superior:g}", "réplicas": int(np.isfinite(pendientes).sum()),
                        "beta efectivo": media, "SE": se})
df_beta = pd.DataFrame(filas_b)
guardar_tabla(df_beta, "08_beta_efectivo")
print("=" * ANCHO)
print("EXPONENTE EFECTIVO EN VENTANAS DE ESPESOR (solo réplicas que llegan al final de la ventana)")
print("=" * ANCHO)
display(df_beta.pivot(index="modelo", columns="ventana", values="beta efectivo")
        .reindex(index=[CLAVES[c] for c in CLAVES], columns=[f"{a:g}-{b:g}" for a, b in VENTANAS_BETA]).round(3))
```

```python
# 7.3.4 Efectos del diseño factorial en ln W
efectos = efectos_factorial(df_factorial, FACTORES, "rugosidad", np.log)
mitad_sin = efectos_factorial(df_factorial[df_factorial["S"] == 0], ["D", "A", "G", "L"], "rugosidad", np.log)
mitad_con = efectos_factorial(df_factorial[df_factorial["S"] == 1], ["D", "A", "G", "L"], "rugosidad", np.log)
sin_oblicuidad = efectos_factorial(df_factorial[(df_factorial["A"] == 0) & (df_factorial["G"] == 0)], ["D", "S", "L"], "rugosidad", np.log)
habilitante_ok = bool((efectos_que_contienen(mitad_sin, "AGL")["t"].abs() <= 3).all()
                      and (efectos_que_contienen(sin_oblicuidad, "SL")["t"].abs() <= 3).all())
principal = mitad_con.set_index("efecto")
con_sombreado_ok = bool(principal.loc["A", "t"] > 3 and principal.loc["G", "t"] > 3 and principal.loc["D", "t"] < -3)
celdas = df_factorial.groupby(FACTORES)[["rugosidad", "relieve", "porosidad"]].mean().reset_index()
camino = {"M1": (0, 0, 0, 0, 0), "M3": (1, 0, 0, 0, 0), "M4": (1, 1, 0, 0, 0), "M5": (1, 1, 1, 0, 0), "M6": (1, 1, 1, 1, 0),
          "M7": (1, 1, 1, 1, 1)}
df_camino = pd.DataFrame([{"modelo": m, **dict(zip(FACTORES, cod)),
                           **celdas.set_index(FACTORES).loc[cod, ["rugosidad", "relieve", "porosidad"]].to_dict()} for m, cod in camino.items()])
for tabla, nombre in ((efectos, "08_efectos"), (mitad_sin, "08_efectos_sin_sombreado"), (mitad_con, "08_efectos_con_sombreado"),
                      (sin_oblicuidad, "08_efectos_sin_oblicuidad"), (celdas, "08_factorial_celdas")):
    guardar_tabla(tabla, nombre)

print("=" * ANCHO)
print(f"DISEÑO FACTORIAL 2^5 A {ESPESOR_FACTORIAL:g} CAPAS: EFECTOS EN ln W (|t| > 3)")
print("=" * ANCHO)
display(efectos[efectos["t"].abs() > 3].sort_values("valor", key=np.abs, ascending=False).round(4))
print("Mitad sin sombreado (factorial 2^4 en D, A, G y L), efectos con |t| > 3:")
display(mitad_sin[mitad_sin["t"].abs() > 3].round(4))
print("Mitad con sombreado (factorial 2^4 en D, A, G y L), efectos principales:")
display(mitad_con[mitad_con["orden"] == 1].round(4))
print("Cuarta parte sin incidencia angular ni gas (factorial 2^3 en D, S y L), efectos con |t| > 3:")
display(sin_oblicuidad[sin_oblicuidad["t"].abs() > 3].round(4))
print("La cadena dentro del diseño (medias de las celdas):")
display(df_camino.round(4))
print(f"Las cinco celdas de mayor relieve a {ESPESOR_FACTORIAL:g} capas (medias de las réplicas):")
display(celdas.sort_values("relieve", ascending=False).head(5).round(4))
fila("Sin sombreado, A, G y L no tienen efecto", "sí" if habilitante_ok else "NO")
fila("Con sombreado: A y G aumentan W, D la reduce", "sí" if con_sombreado_ok else "NO")
```

```python
# 7.3.5 Relieve y porosidad, y mitad con relajación del diseño factorial
efectos_R = efectos_factorial(df_factorial, FACTORES, "relieve", np.log)
efectos_p = efectos_factorial(df_factorial, FACTORES, "porosidad")
con_relajacion = df_factorial[df_factorial["D"] == 1]
relajacion_W = efectos_factorial(con_relajacion, ["A", "S", "G", "L"], "rugosidad", np.log)
relajacion_R = efectos_factorial(con_relajacion, ["A", "S", "G", "L"], "relieve", np.log)
relajacion_p = efectos_factorial(con_relajacion, ["A", "S", "G", "L"], "porosidad")
principal_D1 = relajacion_W.set_index("efecto")
relajacion_ok = bool(principal_D1.loc["S", "t"] > 3 and principal_D1.loc["G", "t"] > 3 and principal_D1.loc["L", "t"] < -3)

# Porosidad: nula sin adhesión lateral, sin sombreado o sin oblicuidad; positiva en las demás celdas
sin_huecos = (df_factorial["L"] == 0) | (df_factorial["S"] == 0) | ((df_factorial["A"] == 0) & (df_factorial["G"] == 0))
porosidad_ok = bool((df_factorial.loc[sin_huecos, "porosidad"] == 0).all() and (df_factorial.loc[~sin_huecos, "porosidad"] > 0).all())
celdas_porosas = (df_factorial[~sin_huecos].groupby(FACTORES)["porosidad"].agg(["mean", "sem", "min"]).reset_index()
                  .rename(columns={"mean": "porosidad media", "sem": "SE", "min": "mínima"}))

def principales(tablas):
    """Efectos principales de varias respuestas, lado a lado (valor y t)."""
    salida = None
    for nombre, tabla in tablas.items():
        t = tabla[tabla["orden"] == 1].set_index("efecto")[["valor", "t"]].rename(columns={"valor": nombre, "t": f"t {nombre}"})
        salida = t if salida is None else salida.join(t)
    return salida

for tabla, nombre in ((efectos_R, "08_efectos_relieve"), (efectos_p, "08_efectos_porosidad"),
                      (relajacion_W, "08_efectos_con_relajacion"), (relajacion_R, "08_efectos_con_relajacion_relieve"),
                      (relajacion_p, "08_efectos_con_relajacion_porosidad"), (celdas_porosas, "08_celdas_porosas")):
    guardar_tabla(tabla, nombre)

print("=" * ANCHO)
print(f"DISEÑO FACTORIAL 2^5 A {ESPESOR_FACTORIAL:g} CAPAS: EFECTOS PRINCIPALES EN LAS TRES RESPUESTAS")
print("=" * ANCHO)
display(principales({"ln W": efectos, "ln R": efectos_R, "porosidad": efectos_p}).round(4))
print("Efectos en ln R con |t| > 3, de mayor a menor valor absoluto (los diez primeros):")
display(efectos_R[efectos_R["t"].abs() > 3].sort_values("valor", key=np.abs, ascending=False).head(10).round(4))
print("Efectos en la porosidad con |t| > 3:")
display(efectos_p[efectos_p["t"].abs() > 3].sort_values("valor", key=np.abs, ascending=False).round(5))
print("Celdas con huecos posibles (adhesión lateral, sombreado y alguna fuente de oblicuidad):")
display(celdas_porosas.round(5))
fila("Porosidad nula donde no puede haber huecos y positiva en las demás celdas", "sí" if porosidad_ok else "NO")
print()
print("=" * ANCHO)
print("MITAD CON RELAJACIÓN (D = 1): FACTORIAL 2^4 EN A, S, G Y L")
print("=" * ANCHO)
display(principales({"ln W": relajacion_W, "ln R": relajacion_R, "porosidad": relajacion_p}).round(4))
print("Efectos en ln W con |t| > 3:")
display(relajacion_W[relajacion_W["t"].abs() > 3].sort_values("valor", key=np.abs, ascending=False).round(4))
fila("Con relajación: S y G aumentan W, L la reduce", "sí" if relajacion_ok else "NO")
```

## 9. Análisis de resultados

### 9.1 Contención de la cadena

En los seis pasos de la cadena, el modelo con su mecanismo apagado y con el nombre del anterior da exactamente las superficies del anterior (sección 6), y el criterio 1 se cumple. Cada cuaderno añade un mecanismo y nada más.

### 9.2 Los siete modelos y el efecto de cada mecanismo

La tabla 7.3.2 y la figura 2 resumen el efecto de cada paso. La relajación (M1 a M2) es el de mayor efecto: reduce la rugosidad a la mitad a 1 capa y 22 veces a 256 capas. Bajar la temperatura de 500 a 400 K (M2 a M3) la aumenta entre un 18 y un $22\,\%$ en todos los espesores. La incidencia angular sin sombreado (M3 a M4) no tiene efecto: los cocientes están entre 0.991 y 1.001, con $p \geq 0.21$. El sombreado (M4 a M5) aumenta la rugosidad un $3\,\%$ a 1 capa y un 10 a $11\,\%$ a 4 y 16 capas, y el gas (M5 a M6) un $13\,\%$ a 1 capa y un $45\,\%$ a 4. La adhesión lateral (M6 a M7) la reduce un $13\,\%$ a 1 capa y un $33\,\%$ a 4. Cada mecanismo, salvo la incidencia angular sin sombreado, cambia la película con $p < 10^{-13}$ en algún espesor, y el criterio 2 se cumple.

Las 10 réplicas de M5 y de M6 llegan al corte, a 64.5 y 9.2 capas en promedio, y ninguna de M7 (tabla 7.3.1), de modo que el criterio 3 se cumple. Los exponentes efectivos por ventanas (tabla 7.3.3) separan los regímenes: 0.50 en M1 en todas las ventanas; entre 0.03 y 0.11 en M2, M3 y M4, decrecientes con el espesor, como corresponde al crecimiento logarítmico; 0.15 en M5 entre 1 y 4 capas; 0.32 en M6 en la misma ventana, antes del corte. En M7 el exponente es de 0.13, 0.11 y 0.16 en las ventanas de 1 a 64 capas y sube a 0.39 entre 64 y 256 capas, cuando se forman los montículos descritos en el cuaderno 07. A 256 capas, la rugosidad de M7, 1.87 capas, es más del doble de la de M3.

Las superficies de la figura 3 resumen la cadena: alturas sin correlación en M1; dominios de pocas capas de desnivel en M2, M3 y M4, indistinguibles a la vista entre M3 y M4; algo más de contraste en M5 a 16 capas; un campo de columnas aisladas, cada una con su sombra, en M6 al detenerse, y en M7 una superficie sin columnas, de rugosidad parecida a la de M5 a ese espesor (0.87 y 0.86 capas). En M6 y M7 cada átomo sufre 6.3 colisiones por término medio: con la capa de gas del cuaderno 06, el número de colisiones no depende del relieve.

### 9.3 Diseño factorial: rugosidad

A 2 capas, los 31 efectos del diseño superan tres errores estándar (tabla 7.3.4): con 10 réplicas por celda en la malla de $200 \times 200$, el error de cada efecto es de 0.0007 en $\ln W$, y la significación no discrimina. Lo que importa es el tamaño de los efectos. El mayor es el de la relajación ($E_D = -0.96$: un factor $e^{-0.96} = 0.38$ en la rugosidad), seguido por el sombreado ($+0.28$), el gas ($+0.20$), la interacción del sombreado con el gas ($+0.20$), la de la relajación con el sombreado ($-0.20$), la del sombreado con la adhesión lateral ($-0.19$) y la adhesión lateral ($-0.19$). La incidencia angular tiene un efecto pequeño ($+0.048$): con $15^\circ$ por eje aporta poca oblicuidad comparada con la del gas.

Dentro de la mitad sin sombreado, el único efecto apreciable es el de la relajación ($-0.76$), y dentro de la cuarta parte sin incidencia angular ni gas, también ($-0.76$): ni A, ni G, ni L cambian la película si el relieve no intercepta los vuelos, ni S y L si los vuelos son verticales. El criterio 4 se cumple. Dentro de la mitad con sombreado, los efectos principales son $+0.095$ para A, $+0.406$ para G, $-1.158$ para D y $-0.372$ para L, y el criterio 5 se cumple.

El papel habilitante del sombreado deja una huella exacta en los efectos. Si un factor solo actúa cuando hay sombreado, su efecto principal es la mitad de su efecto dentro de la mitad con sombreado, y su interacción con S vale lo mismo. Así ocurre: G y S×G valen 0.204 y 0.203; L y S×L, $-0.186$ los dos; A y A×S, 0.048 los dos, y lo mismo sucede con los pares de interacciones de mayor orden, como D×G y D×S×G ($-0.141$ los dos).

Las interacciones de orden alto las dominan dos celdas. Sin relajación ni adhesión lateral, con sombreado y gas, el relieve a 2 capas es de 168 capas sin incidencia angular y de 225 con ella, y la rugosidad de 5.2 y 5.8 capas. Basta activar la relajación o la adhesión lateral para que esas agujas desaparezcan: ninguna celda con adhesión lateral está entre las cinco de mayor relieve, y la tercera ya tiene solo 21 capas. Dentro del diseño, la cadena recorre las celdas M1, M3, M4, M5, M6 y M7 con rugosidades de 1.41, 0.66, 0.66, 0.71, 0.86 y 0.69 capas a 2 capas (figura 4, derecha): una caída fuerte con la relajación, ningún cambio con la incidencia angular, aumentos con el sombreado y el gas y una disminución con la adhesión lateral, como en el experimento I.

### 9.4 Relieve y porosidad

Los efectos principales sobre $\ln R$ siguen el mismo orden que sobre $\ln W$, pero el sombreado y la adhesión lateral pesan más (tabla 7.3.5): $-1.08$ para D, $+0.61$ para S, $-0.58$ para L, $+0.44$ para G y $+0.11$ para A, este último con $t = 8$, frente a $t$ de 30 a 80 en los demás. El sombreado multiplica el relieve por $e^{0.61} = 1.8$ y la rugosidad por $e^{0.28} = 1.3$: el relieve, que mide la columna más alta, registra la inestabilidad antes que la rugosidad. La interacción S×L ($-0.57$) vale casi lo mismo que el efecto de L, la huella de un mecanismo que solo actúa con sombreado.

La porosidad es exactamente nula en todas las réplicas de las 26 celdas sin adhesión lateral, sin sombreado o sin incidencia angular ni gas, y positiva en todas las réplicas de las 6 restantes, y el criterio 7 se cumple. Sin relajación, esas celdas tienen porosidades del $24\,\%$ (solo gas), $12\,\%$ (solo incidencia angular) y $26\,\%$ (las dos); con relajación, del $0.98$, $0.38$ y $1.04\,\%$: la relajación reduce la porosidad entre 25 y 30 veces. Como la porosidad solo es distinta de cero en las celdas con S = L = 1, sus efectos se agrupan en conjuntos de valor idéntico: los de S, L y S×L valen 0.040, la mitad de la porosidad media de las ocho celdas con sombreado y adhesión lateral, y los de D, D×S, D×L y D×S×L, $-0.037$.

### 9.5 Mitad con relajación

Dentro de la mitad con relajación, la de los modelos M3 a M7 de la cadena, los efectos principales en $\ln W$ son $+0.080$ para S, $+0.063$ para G, $-0.063$ para L y $+0.012$ para A, todos con $|t| > 15$ (tabla 7.3.5 y figura 4, centro), y el criterio 6 se cumple. Son tres o cuatro veces menores que en el diseño completo, y de cinco a siete veces menores que en la mitad sin relajación, cuyos efectos se obtienen restando los de esta mitad del doble de los del diseño completo: a 2 capas, la relajación contiene la mayor parte del efecto de los mecanismos de transporte. La huella del sombreado habilitante se mantiene (G y S×G valen 0.063 y 0.062; L y S×L, $-0.063$ los dos), y las mayores interacciones son de segundo orden: S×L, S×G y G×L. Las interacciones de orden alto del diseño completo proceden de las celdas sin relajación. En el relieve, los efectos de S, G y L dentro de esta mitad son de $+0.30$, $+0.21$ y $-0.26$; el de A ($+0.06$, $t = 2.8$) no se distingue del ruido.

## 10. Conclusiones

1. La cadena es coherente: cada modelo contiene exactamente al anterior, y cada mecanismo cambia la película de forma apreciable salvo la incidencia angular sin sombreado, que solo actúa a través de él. Se cumplen los siete criterios de la sección 2.5.
2. La relajación es el mecanismo de mayor efecto. Lleva la película de la deposición aleatoria al crecimiento logarítmico de la clase de Edwards-Wilkinson y la hace 22 veces más lisa a 256 capas; la temperatura de 400 K, con $P = 0.8$, la hace un $20\,\%$ más rugosa que con $P = 1$.
3. El sombreado es el mecanismo habilitante de la serie. Sin él, ni la incidencia angular, ni el gas, ni la adhesión lateral cambian la película, y el diseño factorial lo confirma con efectos nulos fuera de la mitad con sombreado e interacciones con S iguales a los efectos principales. Con él, la incidencia oblicua y el gas desencadenan una inestabilidad de columnas, a 35 capas sin gas y a 3.8 capas con gas.
4. La adhesión lateral suprime las agujas e introduce una porosidad del $1.6\,\%$, pero no detiene el crecimiento de la rugosidad: de los modelos que llegan a 256 capas, M7 es el más rugoso después de M1, porque el sombreado produce ahora montículos que engrosan con el espesor (cuaderno 07).
5. En el diseño factorial a 2 capas, la relajación domina ($-0.96$ en $\ln W$ y $-1.08$ en $\ln R$), el sombreado y el gas aumentan la rugosidad sobre todo juntos, la adhesión lateral la reduce y la incidencia angular de $15^\circ$ tiene un efecto pequeño. Las interacciones de orden alto proceden de dos celdas sin relajación ni adhesión lateral, en las que crecen agujas de cientos de capas; dentro de la mitad con relajación, los efectos son tres o cuatro veces menores que en el diseño completo y las mayores interacciones son de segundo orden.
6. La porosidad solo aparece con adhesión lateral, sombreado y alguna fuente de oblicuidad, y la relajación la reduce entre 25 y 30 veces.
7. Los resultados se refieren al modelo, sin calibración para el cobre. Sus principales limitaciones son la aproximación SOS en M5 y M6, que deja crecer agujas sin límite; los huecos cerrados de M7; el tamaño de la malla frente a la separación de los montículos de M7 a espesores grandes, y el espesor único del diseño factorial, elegido para que lo alcancen todas las celdas.
